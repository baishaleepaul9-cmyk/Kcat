```python
import requests
from tqdm.auto import tqdm

dataset_path = "/content/catpred-db.tar.gz"

response = requests.get(download_url, stream=True)
response.raise_for_status()

total_size = int(response.headers.get("content-length", 0))

with open(dataset_path, "wb") as f:
    with tqdm(
        total=total_size,
        unit="B",
        unit_scale=True,
        desc="Downloading CatPred-DB"
    ) as pbar:

        for chunk in response.iter_content(chunk_size=1024 * 1024):
            if chunk:
                f.write(chunk)
                pbar.update(len(chunk))

print("\nDownload complete!")
```

```python
import numpy as np
import pandas as pd

# Standard amino acids
AMINO_ACIDS = list("ACDEFGHIKLMNPQRSTVWY")

def sequence_features(seq):
    """
    Calculate interpretable biochemical features
    from a protein sequence.
    """

    seq = str(seq).upper()
    length = len(seq)

    features = {}

    # Sequence length
    features["seq_length"] = length

    # Amino-acid composition
    for aa in AMINO_ACIDS:
        features[f"aa_{aa}"] = seq.count(aa) / length

    # Biochemical groups
    hydrophobic = set("AILMFWVY")
    charged = set("DEKR")
    positive = set("KR")
    negative = set("DE")
    aromatic = set("FWY")
    polar = set("STNQ")
    small = set("AGST")

    features["hydrophobic_fraction"] = (
        sum(aa in hydrophobic for aa in seq) / length
    )

    features["charged_fraction"] = (
        sum(aa in charged for aa in seq) / length
    )

    features["positive_fraction"] = (
        sum(aa in positive for aa in seq) / length
    )

    features["negative_fraction"] = (
        sum(aa in negative for aa in seq) / length
    )

    features["aromatic_fraction"] = (
        sum(aa in aromatic for aa in seq) / length
    )

    features["polar_fraction"] = (
        sum(aa in polar for aa in seq) / length
    )

    features["small_residue_fraction"] = (
        sum(aa in small for aa in seq) / length
    )

    return features
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors, Lipinski, Crippen
import numpy as np
import pandas as pd

def smiles_features(smiles):
    mol = Chem.MolFromSmiles(smiles)

    if mol is None:
        return {
            "mol_weight": np.nan,
            "logp": np.nan,
            "tpsa": np.nan,
            "hbd": np.nan,
            "hba": np.nan,
            "rotatable_bonds": np.nan,
            "rings": np.nan,
            "heavy_atoms": np.nan
        }

    return {
        "mol_weight": Descriptors.MolWt(mol),
        "logp": Crippen.MolLogP(mol),
        "tpsa": Descriptors.TPSA(mol),
        "hbd": Lipinski.NumHDonors(mol),
        "hba": Lipinski.NumHAcceptors(mol),
        "rotatable_bonds": Lipinski.NumRotatableBonds(mol),
        "rings": Lipinski.RingCount(mol),
        "heavy_atoms": Lipinski.HeavyAtomCount(mol)
    }

substrate_feature_df = pd.DataFrame(
    df["reactant_smiles"].apply(smiles_features).tolist(),
    index=df.index
)

print("Substrate feature shape:", substrate_feature_df.shape)
display(substrate_feature_df.head())
```

```python
# ============================================================
# RECREATE SEQUENCE FEATURES + BUILD FINAL FEATURE MATRIX
# ============================================================

import numpy as np
import pandas as pd

AMINO_ACIDS = list("ACDEFGHIKLMNPQRSTVWY")

def sequence_features(seq):
    seq = str(seq).upper()
    length = len(seq)

    # Avoid division by zero
    if length == 0:
        return {}

    features = {}

    # Sequence length
    features["seq_length"] = length

    # Amino-acid composition
    for aa in AMINO_ACIDS:
        features[f"aa_{aa}"] = seq.count(aa) / length

    # Biochemical groups
    hydrophobic = set("AILMFWVY")
    charged = set("DEKR")
    positive = set("KR")
    negative = set("DE")
    aromatic = set("FWY")
    polar = set("STNQ")
    small = set("AGST")

    features["hydrophobic_fraction"] = (
        sum(aa in hydrophobic for aa in seq) / length
    )

    features["charged_fraction"] = (
        sum(aa in charged for aa in seq) / length
    )

    features["positive_fraction"] = (
        sum(aa in positive for aa in seq) / length
    )

    features["negative_fraction"] = (
        sum(aa in negative for aa in seq) / length
    )

    features["aromatic_fraction"] = (
        sum(aa in aromatic for aa in seq) / length
    )

    features["polar_fraction"] = (
        sum(aa in polar for aa in seq) / length
    )

    features["small_residue_fraction"] = (
        sum(aa in small for aa in seq) / length
    )

    return features


# ------------------------------------------------------------
# Generate sequence features
# ------------------------------------------------------------

sequence_feature_df = pd.DataFrame(
    df["sequence"].apply(sequence_features).tolist(),
    index=df.index
)

print("Sequence features:", sequence_feature_df.shape)


# ------------------------------------------------------------
# Experimental/context features
# ------------------------------------------------------------

context_features = df[[
    "temperature",
    "ph"
]].copy()

context_features["temperature"] = (
    context_features["temperature"]
    .fillna(context_features["temperature"].median())
)

context_features["ph"] = (
    context_features["ph"]
    .fillna(context_features["ph"].median())
)


# ------------------------------------------------------------
# Combine all features
# ------------------------------------------------------------

X = pd.concat(
    [
        sequence_feature_df,
        substrate_feature_df,
        context_features
    ],
    axis=1
)

y = df["log10_value"].copy()


# ------------------------------------------------------------
# Final checks
# ------------------------------------------------------------

print("\n==============================")
print("FINAL FEATURE MATRIX")
print("==============================")

print("X shape:", X.shape)
print("y shape:", y.shape)

print(
    "Missing values:",
    X.isna().sum().sum()
)

print(
    "Infinite values:",
    np.isinf(
        X.select_dtypes(include=np.number)
    ).sum().sum()
)

print("\nNumber of features:", X.shape[1])

print("\nFeature groups:")
print("Sequence:", sequence_feature_df.shape[1])
print("Substrate:", substrate_feature_df.shape[1])
print("Context:", context_features.shape[1])
```

```python
# ============================================================
# STEP 12 — CREATE FEATURES FOR TRAIN / VALIDATION / TEST
# ============================================================

# Training-only imputation values
train_temp_median = kcat_train["temperature"].median()
train_ph_median = kcat_train["ph"].median()

print("Training temperature median:", train_temp_median)
print("Training pH median:", train_ph_median)


def create_features(data, temp_median, ph_median):

    data = data.copy()

    # -----------------------------
    # Protein sequence features
    # -----------------------------
    sequence_feature_df = pd.DataFrame(
        data["sequence"].apply(sequence_features).tolist(),
        index=data.index
    )

    # -----------------------------
    # Substrate features
    # -----------------------------
    substrate_feature_df = pd.DataFrame(
        data["reactant_smiles"].apply(smiles_features).tolist(),
        index=data.index
    )

    # -----------------------------
    # Experimental conditions
    # -----------------------------
    context_features = data[
        ["temperature", "ph"]
    ].copy()

    context_features["temperature"] = (
        context_features["temperature"]
        .fillna(temp_median)
    )

    context_features["ph"] = (
        context_features["ph"]
        .fillna(ph_median)
    )

    # -----------------------------
    # Combine
    # -----------------------------
    features = pd.concat(
        [
            sequence_feature_df,
            substrate_feature_df,
            context_features
        ],
        axis=1
    )

    return features


# Create feature matrices
X_train = create_features(
    kcat_train,
    train_temp_median,
    train_ph_median
)

X_val = create_features(
    kcat_val,
    train_temp_median,
    train_ph_median
)

X_test = create_features(
    kcat_test,
    train_temp_median,
    train_ph_median
)


# Targets
y_train = kcat_train["log10_value"].copy()
y_val = kcat_val["log10_value"].copy()
y_test = kcat_test["log10_value"].copy()


# ============================================================
# CHECKS
# ============================================================

print("\nTRAIN")
print("X:", X_train.shape)
print("y:", y_train.shape)

print("\nVALIDATION")
print("X:", X_val.shape)
print("y:", y_val.shape)

print("\nTEST")
print("X:", X_test.shape)
print("y:", y_test.shape)

print("\nMissing values:")
print("Train:", X_train.isna().sum().sum())
print("Validation:", X_val.isna().sum().sum())
print("Test:", X_test.isna().sum().sum())
```

```python
# ============================================================
# STEP 14 — BASELINE MODEL COMPARISON
# ============================================================

import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import ElasticNet
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
from scipy.stats import pearsonr

from xgboost import XGBRegressor


# ------------------------------------------------------------
# Evaluation function
# ------------------------------------------------------------

def evaluate_model(model, X, y, dataset_name):

    pred = model.predict(X)

    mae = mean_absolute_error(y, pred)
    rmse = np.sqrt(mean_squared_error(y, pred))
    r2 = r2_score(y, pred)
    pearson_r, pearson_p = pearsonr(y, pred)

    return {
        "Dataset": dataset_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p
    }


# ------------------------------------------------------------
# Models
# ------------------------------------------------------------

models = {

    "ElasticNet": Pipeline([
        ("scaler", StandardScaler()),
        ("model", ElasticNet(
            alpha=0.01,
            l1_ratio=0.5,
            max_iter=10000,
            random_state=42
        ))
    ]),

    "ExtraTrees": ExtraTreesRegressor(
        n_estimators=300,
        max_features="sqrt",
        min_samples_leaf=2,
        n_jobs=-1,
        random_state=42
    ),

    "XGBoost": XGBRegressor(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        n_jobs=-1,
        random_state=42
    )
}


# ------------------------------------------------------------
# Train and evaluate
# ------------------------------------------------------------

results = []
trained_models = {}

for name, model in models.items():

    print(f"\n{'='*50}")
    print(f"Training {name}")
    print(f"{'='*50}")

    model.fit(X_train, y_train)

    trained_models[name] = model

    # Validation
    val_result = evaluate_model(
        model,
        X_val,
        y_val,
        "Validation"
    )
    val_result["Model"] = name
    results.append(val_result)

    print(
        f"Validation | "
        f"R²={val_result['R2']:.4f} | "
        f"RMSE={val_result['RMSE']:.4f} | "
        f"MAE={val_result['MAE']:.4f} | "
        f"Pearson r={val_result['Pearson_r']:.4f}"
    )


results_df = pd.DataFrame(results)

print("\n==============================")
print("VALIDATION RESULTS")
print("==============================")

display(
    results_df[
        [
            "Model",
            "R2",
            "RMSE",
            "MAE",
            "Pearson_r",
            "Pearson_p"
        ]
    ].sort_values("R2", ascending=False)
)
```

```python
# ============================================================
# STEP 15 — EXTRA TREES FEATURE IMPORTANCE
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Retrieve trained ExtraTrees model
extra_model = trained_models["ExtraTrees"]

# Feature importance
importance_df = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": extra_model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
).reset_index(drop=True)

print("Top 20 features:")
display(importance_df.head(20))
```

```python
# ============================================================
# FEATURE GROUPING + GROUPED IMPORTANCE
# ============================================================

# Make sure importance_df exists
print("Columns currently available:")
print(importance_df.columns.tolist())


# Define feature groups
def feature_group(feature):

    # Amino-acid composition
    if feature.startswith("aa_"):
        return "Amino-acid composition"

    # Protein-level descriptors
    elif feature in [
        "seq_length",
        "hydrophobic_fraction",
        "charged_fraction",
        "positive_fraction",
        "negative_fraction",
        "aromatic_fraction",
        "polar_fraction",
        "small_residue_fraction"
    ]:
        return "Protein properties"

    # Substrate descriptors
    elif feature in [
        "mol_weight",
        "logp",
        "tpsa",
        "hbd",
        "hba",
        "rotatable_bonds",
        "rings",
        "heavy_atoms"
    ]:
        return "Substrate properties"

    # Experimental conditions
    elif feature in [
        "temperature",
        "ph"
    ]:
        return "Experimental conditions"

    else:
        return "Other"


# Add group information
importance_df["Feature_Group"] = (
    importance_df["Feature"].apply(feature_group)
)


# Calculate total importance by group
group_importance = (
    importance_df
    .groupby("Feature_Group")["Importance"]
    .sum()
    .sort_values(ascending=False)
)


print("\n====================================")
print("FEATURE IMPORTANCE BY GROUP")
print("====================================")

display(
    group_importance.to_frame("Total_Importance")
)


print("\nTotal importance:", group_importance.sum())
```

```python
# ============================================================
# LIGHTWEIGHT MODEL INTERPRETATION
# PERMUTATION IMPORTANCE
# ============================================================

from sklearn.inspection import permutation_importance
import pandas as pd
import numpy as np

extra_model = trained_models["ExtraTrees"]

# Use only 300 validation samples for speed
np.random.seed(42)

n_samples = min(300, len(X_val))

idx = np.random.choice(
    len(X_val),
    size=n_samples,
    replace=False
)

X_perm = X_val.iloc[idx]
y_perm = y_val.iloc[idx]

print("Permutation sample:", X_perm.shape)

# Calculate permutation importance
perm_result = permutation_importance(
    extra_model,
    X_perm,
    y_perm,
    n_repeats=3,
    random_state=42,
    scoring="r2",
    n_jobs=-1
)

# Create results table
importance_df = pd.DataFrame({
    "Feature": X_perm.columns,
    "Importance": perm_result.importances_mean,
    "Std": perm_result.importances_std
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
).reset_index(drop=True)

print("\nTop 20 features:")
display(importance_df.head(20))
```

```python
# ============================================================
# FIGURE — TOP 20 PERMUTATION IMPORTANCE
# ============================================================

import matplotlib.pyplot as plt

top20 = (
    importance_df
    .head(20)
    .sort_values("Importance")
)

plt.figure(figsize=(8, 7))

plt.barh(
    top20["Feature"],
    top20["Importance"]
)

plt.xlabel("Permutation Importance (Δ R²)")
plt.ylabel("Feature")
plt.title("Top Predictive Features for Enzyme kcat")

plt.tight_layout()
plt.show()
```

```python
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import randint

param_dist = {
    "n_estimators": [200, 300, 500],
    "max_features": ["sqrt", "log2", 0.5, 0.8, 1.0],
    "min_samples_leaf": [1, 2, 3, 5],
    "min_samples_split": [2, 5, 10],
    "max_depth": [None, 10, 20, 30]
}

et = ExtraTreesRegressor(
    random_state=42,
    n_jobs=-1
)

search = RandomizedSearchCV(
    estimator=et,
    param_distributions=param_dist,
    n_iter=15,
    scoring="r2",
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

search.fit(X_train, y_train)

print("Best parameters:")
print(search.best_params_)

print("\nBest CV R²:")
print(search.best_score_)
```

```python
# ============================================================
# SEQUENCE-CLUSTER OVERLAP BETWEEN TRAIN AND TEST
# ============================================================

cluster_cols = [
    "sequence_40cluster",
    "sequence_60cluster",
    "sequence_80cluster",
    "sequence_99cluster"
]

for col in cluster_cols:

    train_clusters = set(kcat_train[col].dropna().unique())
    test_clusters = set(kcat_test[col].dropna().unique())

    unseen_test_clusters = test_clusters - train_clusters
    shared_clusters = test_clusters & train_clusters

    unseen_samples = kcat_test[col].isin(unseen_test_clusters).sum()

    print("\n" + "=" * 55)
    print(col)
    print("=" * 55)

    print("Training clusters :", len(train_clusters))
    print("Test clusters     :", len(test_clusters))
    print("Shared clusters   :", len(shared_clusters))
    print("Unseen clusters   :", len(unseen_test_clusters))
    print("Test samples in unseen clusters:", unseen_samples)
```

```python
# ============================================================
# SEQUENCE-AWARE GENERALIZATION ANALYSIS
# ============================================================

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr
import numpy as np
import pandas as pd

cluster_cols = [
    "sequence_40cluster",
    "sequence_60cluster",
    "sequence_80cluster",
    "sequence_99cluster"
]

generalization_results = []

for col in cluster_cols:

    # Clusters present in training
    train_clusters = set(
        kcat_train[col].dropna().unique()
    )

    # Identify test samples belonging to clusters
    # completely absent from training
    unseen_mask = ~kcat_test[col].isin(train_clusters)

    n_unseen = unseen_mask.sum()

    y_true_unseen = y_test[unseen_mask.values]
    y_pred_unseen = y_test_pred[unseen_mask.values]

    if n_unseen >= 3:

        r2 = r2_score(y_true_unseen, y_pred_unseen)
        rmse = np.sqrt(
            mean_squared_error(
                y_true_unseen,
                y_pred_unseen
            )
        )
        mae = mean_absolute_error(
            y_true_unseen,
            y_pred_unseen
        )

        r, p = pearsonr(
            y_true_unseen,
            y_pred_unseen
        )

    else:
        r2 = np.nan
        rmse = np.nan
        mae = np.nan
        r = np.nan
        p = np.nan

    generalization_results.append({
        "Cluster threshold": col.replace(
            "sequence_", ""
        ).replace("cluster", "%"),
        "Unseen test samples": n_unseen,
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    })

generalization_df = pd.DataFrame(
    generalization_results
)

print("=" * 70)
print("SEQUENCE-AWARE GENERALIZATION")
print("=" * 70)

display(generalization_df)
```

```python
# ============================================================
# FEATURE GROUP ABLATION
# ============================================================

feature_groups = {
    "Sequence": sequence_features,
    "Substrate": substrate_features,
    "Context": context_features,

    "Sequence + Substrate":
        sequence_features + substrate_features,

    "Sequence + Context":
        sequence_features + context_features,

    "Substrate + Context":
        substrate_features + context_features,

    "All features":
        sequence_features + substrate_features + context_features
}

for name, features in feature_groups.items():
    print(f"{name}: {len(features)} features")
```

```python
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr
import numpy as np
import pandas as pd

ablation_results = []

for name, features in feature_groups.items():

    print("\n" + "=" * 60)
    print(f"Training: {name}")
    print(f"Features: {len(features)}")

    model = ExtraTreesRegressor(
        n_estimators=300,
        max_features=1.0,
        min_samples_split=2,
        min_samples_leaf=1,
        max_depth=None,
        random_state=42,
        n_jobs=-1
    )

    model.fit(
        X_train[features],
        y_train
    )

    y_pred = model.predict(
        X_val[features]
    )

    r2 = r2_score(y_val, y_pred)
    rmse = np.sqrt(
        mean_squared_error(y_val, y_pred)
    )
    mae = mean_absolute_error(
        y_val, y_pred
    )
    r, p = pearsonr(
        y_val, y_pred
    )

    ablation_results.append({
        "Feature_Set": name,
        "N_Features": len(features),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    })

    print(f"R²          : {r2:.4f}")
    print(f"RMSE        : {rmse:.4f}")
    print(f"MAE         : {mae:.4f}")
    print(f"Pearson r   : {r:.4f}")

ablation_df = pd.DataFrame(ablation_results)

print("\n" + "=" * 70)
print("FEATURE GROUP ABLATION RESULTS")
print("=" * 70)

display(
    ablation_df
    .sort_values("R2", ascending=False)
    .reset_index(drop=True)
)
```

```python
# ============================================================
# PERFORMANCE ACROSS kcat RANGES
# ============================================================

import pandas as pd
import numpy as np
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

analysis_df = pd.DataFrame({
    "Observed": np.asarray(y_test),
    "Predicted": np.asarray(y_test_pred)
})

# Divide into tertiles based ONLY on the test observations
analysis_df["kcat_group"] = pd.qcut(
    analysis_df["Observed"],
    q=3,
    labels=["Low kcat", "Medium kcat", "High kcat"]
)

results = []

for group, subset in analysis_df.groupby("kcat_group", observed=True):

    y_true = subset["Observed"].values
    y_pred = subset["Predicted"].values

    r2 = r2_score(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r, p = pearsonr(y_true, y_pred)

    results.append({
        "kcat_group": group,
        "N": len(subset),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    })

range_results = pd.DataFrame(results)

display(range_results)
```

```python
# ============================================================
# RMSE ACROSS kcat RANGES
# ============================================================

plt.figure(figsize=(7, 5))

plt.bar(
    range_results["kcat_group"],
    range_results["RMSE"]
)

plt.ylabel("RMSE")
plt.xlabel("Observed kcat group")
plt.title("Prediction Error Across kcat Ranges")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# SUBSTRATE-SPECIFIC PERFORMANCE ANALYSIS
# CatPred — Enzyme kcat Prediction
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

from scipy.stats import pearsonr


# ============================================================
# 1. CHECK REQUIRED OBJECTS
# ============================================================

print("=" * 65)
print("SUBSTRATE-SPECIFIC PERFORMANCE ANALYSIS")
print("=" * 65)

print("\nTest data:", X_test.shape)
print("Target:", y_test.shape)
print("Model:", type(final_model).__name__)


# ============================================================
# 2. SUBSTRATE FEATURES
# ============================================================

substrate_features = [
    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms"
]

print("\nSubstrate features:")
print(substrate_features)


# Make sure all features exist
missing = [
    f for f in substrate_features
    if f not in X_test.columns
]

if missing:
    raise ValueError(
        f"Missing substrate features: {missing}"
    )


# ============================================================
# 3. GENERATE PREDICTIONS
# ============================================================

y_true = np.asarray(y_test)
y_pred = final_model.predict(X_test)

print("\nPredictions generated:", len(y_pred))


# ============================================================
# 4. FUNCTION FOR PERFORMANCE METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    # Need at least 2 observations
    if len(y_true) < 2:
        return {
            "N": len(y_true),
            "R2": np.nan,
            "RMSE": np.nan,
            "MAE": np.nan,
            "Pearson_r": np.nan,
            "Pearson_p": np.nan
        }

    r2 = r2_score(y_true, y_pred)

    rmse = np.sqrt(
        mean_squared_error(y_true, y_pred)
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    r, p = pearsonr(
        y_true,
        y_pred
    )

    return {
        "N": len(y_true),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    }


# ============================================================
# 5. SUBSTRATE-SPECIFIC ANALYSIS
# ============================================================

results = []

for feature in substrate_features:

    print("\n" + "=" * 60)
    print("SUBSTRATE:", feature)
    print("=" * 60)

    # Median calculated ONLY from test set
    median_value = X_test[feature].median()

    print("Median:", median_value)

    # Low / High groups
    low_mask = X_test[feature] <= median_value
    high_mask = X_test[feature] > median_value

    y_low = y_true[low_mask.values]
    pred_low = y_pred[low_mask.values]

    y_high = y_true[high_mask.values]
    pred_high = y_pred[high_mask.values]

    # Calculate metrics
    low_metrics = calculate_metrics(
        y_low,
        pred_low
    )

    high_metrics = calculate_metrics(
        y_high,
        pred_high
    )

    # Store
    results.append({
        "Substrate_Feature": feature,
        "Group": "Low",
        "Threshold": median_value,
        **low_metrics
    })

    results.append({
        "Substrate_Feature": feature,
        "Group": "High",
        "Threshold": median_value,
        **high_metrics
    })

    print(
        f"Low  | N={low_metrics['N']} | "
        f"R²={low_metrics['R2']:.4f} | "
        f"RMSE={low_metrics['RMSE']:.4f} | "
        f"MAE={low_metrics['MAE']:.4f} | "
        f"r={low_metrics['Pearson_r']:.4f}"
    )

    print(
        f"High | N={high_metrics['N']} | "
        f"R²={high_metrics['R2']:.4f} | "
        f"RMSE={high_metrics['RMSE']:.4f} | "
        f"MAE={high_metrics['MAE']:.4f} | "
        f"r={high_metrics['Pearson_r']:.4f}"
    )


# ============================================================
# 6. RESULTS TABLE
# ============================================================

substrate_results = pd.DataFrame(results)

print("\n")
print("=" * 75)
print("SUBSTRATE-SPECIFIC PERFORMANCE RESULTS")
print("=" * 75)

display(
    substrate_results.round(4)
)


# ============================================================
# 7. PIVOT TABLE FOR R²
# ============================================================

r2_table = substrate_results.pivot(
    index="Substrate_Feature",
    columns="Group",
    values="R2"
)

print("\n" + "=" * 60)
print("R² BY SUBSTRATE GROUP")
print("=" * 60)

display(r2_table.round(4))


# ============================================================
# 8. PLOT R²
# ============================================================

plt.figure(figsize=(10, 6))

x = np.arange(len(r2_table))
width = 0.35

plt.bar(
    x - width/2,
    r2_table["Low"],
    width,
    label="Low"
)

plt.bar(
    x + width/2,
    r2_table["High"],
    width,
    label="High"
)

plt.xticks(
    x,
    r2_table.index,
    rotation=45,
    ha="right"
)

plt.ylabel("R²")
plt.xlabel("Substrate descriptor")
plt.title(
    "Model Performance Across Substrate Property Groups"
)

plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 9. PLOT RMSE
# ============================================================

rmse_table = substrate_results.pivot(
    index="Substrate_Feature",
    columns="Group",
    values="RMSE"
)

plt.figure(figsize=(10, 6))

plt.bar(
    x - width/2,
    rmse_table["Low"],
    width,
    label="Low"
)

plt.bar(
    x + width/2,
    rmse_table["High"],
    width,
    label="High"
)

plt.xticks(
    x,
    rmse_table.index,
    rotation=45,
    ha="right"
)

plt.ylabel("RMSE")
plt.xlabel("Substrate descriptor")
plt.title(
    "Prediction Error Across Substrate Property Groups"
)

plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 10. SAVE RESULTS TO GOOGLE DRIVE
# ============================================================

SAVE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project"

import os

os.makedirs(
    os.path.join(
        SAVE_DIR,
        "results"
    ),
    exist_ok=True
)

substrate_results.to_csv(
    os.path.join(
        SAVE_DIR,
        "results",
        "substrate_specific_performance.csv"
    ),
    index=False
)

r2_table.to_csv(
    os.path.join(
        SAVE_DIR,
        "results",
        "substrate_specific_R2.csv"
    )
)

rmse_table.to_csv(
    os.path.join(
        SAVE_DIR,
        "results",
        "substrate_specific_RMSE.csv"
    )
)

print("\n" + "=" * 65)
print("SUBSTRATE ANALYSIS SAVED")
print("=" * 65)

print(
    os.path.join(
        SAVE_DIR,
        "results",
        "substrate_specific_performance.csv"
    )
)
```

```python
# ============================================================
# CATPRED kcat — RESIDUAL & ERROR ANALYSIS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score
)
from scipy.stats import pearsonr

print("=" * 70)
print("CATPRED kcat — RESIDUAL / ERROR ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. CHECK REQUIRED VARIABLES
# ------------------------------------------------------------

required = ["y_test", "test_predictions"]

missing = [x for x in required if x not in globals()]

if missing:
    raise NameError(
        f"Missing variables: {missing}\n"
        "Make sure the model evaluation cell has been run first."
    )

print("\n✓ y_test available")
print("✓ test_predictions available")


# ------------------------------------------------------------
# 2. PREPARE DATA
# ------------------------------------------------------------

y_actual = np.asarray(y_test).ravel()
y_pred = np.asarray(test_predictions).ravel()

if len(y_actual) != len(y_pred):
    raise ValueError(
        f"Length mismatch: y_test={len(y_actual)}, "
        f"predictions={len(y_pred)}"
    )

residuals = y_actual - y_pred
absolute_error = np.abs(residuals)
squared_error = residuals ** 2

residual_df = pd.DataFrame({
    "Actual_kcat": y_actual,
    "Predicted_kcat": y_pred,
    "Residual": residuals,
    "Absolute_Error": absolute_error,
    "Squared_Error": squared_error
})

print(f"\nTest samples: {len(residual_df)}")


# ------------------------------------------------------------
# 3. OVERALL ERROR METRICS
# ------------------------------------------------------------

r2 = r2_score(y_actual, y_pred)
rmse = np.sqrt(mean_squared_error(y_actual, y_pred))
mae = mean_absolute_error(y_actual, y_pred)
pearson_r, pearson_p = pearsonr(y_actual, y_pred)

print("\n" + "=" * 70)
print("OVERALL TEST-SET PERFORMANCE")
print("=" * 70)

print(f"R²          : {r2:.4f}")
print(f"RMSE        : {rmse:.4f}")
print(f"MAE         : {mae:.4f}")
print(f"Pearson r   : {pearson_r:.4f}")
print(f"Pearson p   : {pearson_p:.4e}")


# ------------------------------------------------------------
# 4. RESIDUAL SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RESIDUAL SUMMARY")
print("=" * 70)

print(f"Mean residual       : {np.mean(residuals):.4f}")
print(f"Median residual     : {np.median(residuals):.4f}")
print(f"Std residual        : {np.std(residuals):.4f}")
print(f"Min residual        : {np.min(residuals):.4f}")
print(f"Max residual        : {np.max(residuals):.4f}")
print(f"Mean absolute error : {np.mean(absolute_error):.4f}")


# ------------------------------------------------------------
# 5. BIAS CHECK
# ------------------------------------------------------------

overprediction = np.sum(residuals < 0)
underprediction = np.sum(residuals > 0)
approximately_correct = np.sum(residuals == 0)

print("\n" + "=" * 70)
print("PREDICTION DIRECTION")
print("=" * 70)

print(
    f"Under-predictions : {underprediction} "
    f"({underprediction / len(residuals) * 100:.2f}%)"
)

print(
    f"Over-predictions  : {overprediction} "
    f"({overprediction / len(residuals) * 100:.2f}%)"
)

print(
    f"Exact matches     : {approximately_correct} "
    f"({approximately_correct / len(residuals) * 100:.2f}%)"
)


# ------------------------------------------------------------
# 6. LARGEST ERRORS
# ------------------------------------------------------------

largest_errors = (
    residual_df
    .sort_values("Absolute_Error", ascending=False)
    .head(20)
    .reset_index()
    .rename(columns={"index": "Original_Test_Index"})
)

print("\n" + "=" * 70)
print("TOP 20 LARGEST PREDICTION ERRORS")
print("=" * 70)

display(largest_errors)


# ------------------------------------------------------------
# 7. TOP UNDER-PREDICTIONS
# ------------------------------------------------------------

under_predictions = (
    residual_df
    .sort_values("Residual", ascending=False)
    .head(10)
    .reset_index()
    .rename(columns={"index": "Original_Test_Index"})
)

print("\n" + "=" * 70)
print("TOP 10 UNDER-PREDICTIONS")
print("(Actual kcat substantially higher than predicted)")
print("=" * 70)

display(under_predictions)


# ------------------------------------------------------------
# 8. TOP OVER-PREDICTIONS
# ------------------------------------------------------------

over_predictions = (
    residual_df
    .sort_values("Residual", ascending=True)
    .head(10)
    .reset_index()
    .rename(columns={"index": "Original_Test_Index"})
)

print("\n" + "=" * 70)
print("TOP 10 OVER-PREDICTIONS")
print("(Predicted kcat substantially higher than actual)")
print("=" * 70)

display(over_predictions)


# ------------------------------------------------------------
# 9. CREATE SAVE DIRECTORIES
# ------------------------------------------------------------

SAVE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project"

RESULTS_DIR = os.path.join(SAVE_DIR, "results")
FIGURES_DIR = os.path.join(SAVE_DIR, "figures")

os.makedirs(RESULTS_DIR, exist_ok=True)
os.makedirs(FIGURES_DIR, exist_ok=True)


# ------------------------------------------------------------
# 10. SAVE COMPLETE RESIDUAL DATA
# ------------------------------------------------------------

residual_path = os.path.join(
    RESULTS_DIR,
    "residual_error_analysis.csv"
)

residual_df.to_csv(residual_path, index=False)

print(f"\n✓ Saved: {residual_path}")


# ------------------------------------------------------------
# 11. SAVE LARGEST ERRORS
# ------------------------------------------------------------

largest_path = os.path.join(
    RESULTS_DIR,
    "top_20_prediction_errors.csv"
)

largest_errors.to_csv(largest_path, index=False)

under_path = os.path.join(
    RESULTS_DIR,
    "top_10_under_predictions.csv"
)

under_predictions.to_csv(under_path, index=False)

over_path = os.path.join(
    RESULTS_DIR,
    "top_10_over_predictions.csv"
)

over_predictions.to_csv(over_path, index=False)

print("✓ Saved largest-error tables")


# ------------------------------------------------------------
# 12. ACTUAL vs PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(8, 7))

plt.scatter(
    y_actual,
    y_pred,
    alpha=0.5
)

min_val = min(y_actual.min(), y_pred.min())
max_val = max(y_actual.max(), y_pred.max())

plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    linestyle="--"
)

plt.xlabel("Observed kcat")
plt.ylabel("Predicted kcat")
plt.title("Observed vs Predicted kcat")

plt.tight_layout()

actual_pred_path = os.path.join(
    FIGURES_DIR,
    "observed_vs_predicted_kcat.png"
)

plt.savefig(actual_pred_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"✓ Saved: {actual_pred_path}")


# ------------------------------------------------------------
# 13. RESIDUALS vs PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(8, 7))

plt.scatter(
    y_pred,
    residuals,
    alpha=0.5
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Predicted kcat")
plt.ylabel("Residual (Observed − Predicted)")
plt.title("Residuals vs Predicted kcat")

plt.tight_layout()

residual_pred_path = os.path.join(
    FIGURES_DIR,
    "residuals_vs_predicted_kcat.png"
)

plt.savefig(residual_pred_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"✓ Saved: {residual_pred_path}")


# ------------------------------------------------------------
# 14. RESIDUAL DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.hist(
    residuals,
    bins=40
)

plt.axvline(
    0,
    linestyle="--"
)

plt.xlabel("Residual (Observed − Predicted)")
plt.ylabel("Frequency")
plt.title("Distribution of Prediction Residuals")

plt.tight_layout()

residual_dist_path = os.path.join(
    FIGURES_DIR,
    "residual_distribution.png"
)

plt.savefig(residual_dist_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"✓ Saved: {residual_dist_path}")


# ------------------------------------------------------------
# 15. ABSOLUTE ERROR vs OBSERVED kcat
# ------------------------------------------------------------

plt.figure(figsize=(8, 6))

plt.scatter(
    y_actual,
    absolute_error,
    alpha=0.5
)

plt.xlabel("Observed kcat")
plt.ylabel("Absolute Prediction Error")
plt.title("Absolute Prediction Error vs Observed kcat")

plt.tight_layout()

abs_error_path = os.path.join(
    FIGURES_DIR,
    "absolute_error_vs_observed_kcat.png"
)

plt.savefig(abs_error_path, dpi=300, bbox_inches="tight")
plt.show()

print(f"✓ Saved: {abs_error_path}")


# ------------------------------------------------------------
# 16. ERROR BY kcat RANGE
# ------------------------------------------------------------

# Use tertiles so the analysis does not assume arbitrary
# biological thresholds.

q1 = np.quantile(y_actual, 1/3)
q2 = np.quantile(y_actual, 2/3)

conditions = [
    y_actual <= q1,
    (y_actual > q1) & (y_actual <= q2),
    y_actual > q2
]

labels = [
    "Low kcat",
    "Medium kcat",
    "High kcat"
]

range_results = []

for label, condition in zip(labels, conditions):

    if np.sum(condition) == 0:
        continue

    actual_group = y_actual[condition]
    pred_group = y_pred[condition]

    group_r2 = r2_score(actual_group, pred_group)
    group_rmse = np.sqrt(
        mean_squared_error(actual_group, pred_group)
    )
    group_mae = mean_absolute_error(
        actual_group,
        pred_group
    )

    if len(actual_group) > 1:
        group_r, group_p = pearsonr(
            actual_group,
            pred_group
        )
    else:
        group_r = np.nan
        group_p = np.nan

    range_results.append({
        "kcat_group": label,
        "N": len(actual_group),
        "R2": group_r2,
        "RMSE": group_rmse,
        "MAE": group_mae,
        "Pearson_r": group_r,
        "Pearson_p": group_p
    })

range_df = pd.DataFrame(range_results)

print("\n" + "=" * 70)
print("ERROR PERFORMANCE ACROSS kcat RANGES")
print("=" * 70)

display(range_df)


range_path = os.path.join(
    RESULTS_DIR,
    "residual_performance_by_kcat_range.csv"
)

range_df.to_csv(range_path, index=False)

print(f"✓ Saved: {range_path}")


# ------------------------------------------------------------
# 17. SAVE SUMMARY
# ------------------------------------------------------------

summary_path = os.path.join(
    RESULTS_DIR,
    "residual_analysis_summary.txt"
)

with open(summary_path, "w") as f:

    f.write("CATPRED kcat — RESIDUAL / ERROR ANALYSIS\n")
    f.write("=" * 60 + "\n\n")

    f.write(f"Test samples: {len(y_actual)}\n\n")

    f.write("OVERALL PERFORMANCE\n")
    f.write("-" * 40 + "\n")
    f.write(f"R2: {r2:.6f}\n")
    f.write(f"RMSE: {rmse:.6f}\n")
    f.write(f"MAE: {mae:.6f}\n")
    f.write(f"Pearson r: {pearson_r:.6f}\n")
    f.write(f"Pearson p: {pearson_p:.6e}\n\n")

    f.write("RESIDUAL SUMMARY\n")
    f.write("-" * 40 + "\n")
    f.write(f"Mean residual: {np.mean(residuals):.6f}\n")
    f.write(f"Median residual: {np.median(residuals):.6f}\n")
    f.write(f"Std residual: {np.std(residuals):.6f}\n")
    f.write(f"Minimum residual: {np.min(residuals):.6f}\n")
    f.write(f"Maximum residual: {np.max(residuals):.6f}\n\n")

    f.write("PREDICTION DIRECTION\n")
    f.write("-" * 40 + "\n")
    f.write(f"Under-predictions: {underprediction}\n")
    f.write(f"Over-predictions: {overprediction}\n")
    f.write(f"Exact matches: {approximately_correct}\n")

range_df.to_csv(
    os.path.join(
        RESULTS_DIR,
        "residual_kcat_range_summary.csv"
    ),
    index=False
)

print(f"✓ Saved: {summary_path}")


# ------------------------------------------------------------
# 18. FINAL STATUS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("RESIDUAL ANALYSIS COMPLETE")
print("=" * 70)

print("\nResults saved to:")
print(SAVE_DIR)

print("\nResults:")
print("✓ residual_error_analysis.csv")
print("✓ top_20_prediction_errors.csv")
print("✓ top_10_under_predictions.csv")
print("✓ top_10_over_predictions.csv")
print("✓ residual_performance_by_kcat_range.csv")
print("✓ residual_kcat_range_summary.csv")
print("✓ residual_analysis_summary.txt")

print("\nFigures:")
print("✓ observed_vs_predicted_kcat.png")
print("✓ residuals_vs_predicted_kcat.png")
print("✓ residual_distribution.png")
print("✓ absolute_error_vs_observed_kcat.png")

print("\nNext step after this:")
print("→ Interpret residual/error patterns")
print("→ Identify systematic failure regions")
print("→ Prepare final model interpretation")
```

```python
# ================================================================
# CATPRED kcat — PROTEIN / SEQUENCE OVERLAP AUDIT
# ================================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("CATPRED kcat — PROTEIN / SEQUENCE OVERLAP AUDIT")
print("=" * 70)

# ------------------------------------------------
# 1. Check that original dataframe is available
# ------------------------------------------------

if "df" not in globals():
    raise NameError(
        "Original dataframe 'df' is not available. "
        "Reload the original dataset first."
    )

print(f"\nOriginal dataset shape: {df.shape}")

# ------------------------------------------------
# 2. Recover identifiers using the split indices
# ------------------------------------------------

train_meta = df.loc[X_train.index].copy()
val_meta   = df.loc[X_val.index].copy()
test_meta  = df.loc[X_test.index].copy()

print("\nSplit sizes:")
print(f"Train: {len(train_meta)}")
print(f"Validation: {len(val_meta)}")
print(f"Test: {len(test_meta)}")

# ------------------------------------------------
# 3. Function for overlap analysis
# ------------------------------------------------

def overlap_report(column, name=None):

    if name is None:
        name = column

    print("\n" + "-" * 70)
    print(f"{name}")
    print("-" * 70)

    train_vals = set(
        train_meta[column]
        .dropna()
        .astype(str)
        .unique()
    )

    val_vals = set(
        val_meta[column]
        .dropna()
        .astype(str)
        .unique()
    )

    test_vals = set(
        test_meta[column]
        .dropna()
        .astype(str)
        .unique()
    )

    train_val = train_vals & val_vals
    train_test = train_vals & test_vals
    val_test = val_vals & test_vals
    all_three = train_vals & val_vals & test_vals

    print(f"Train unique:      {len(train_vals):,}")
    print(f"Validation unique: {len(val_vals):,}")
    print(f"Test unique:       {len(test_vals):,}")

    print(f"\nTrain ∩ Validation: {len(train_val):,}")
    print(f"Train ∩ Test:       {len(train_test):,}")
    print(f"Validation ∩ Test:  {len(val_test):,}")
    print(f"All three splits:   {len(all_three):,}")

    return {
        "feature": name,
        "train_unique": len(train_vals),
        "val_unique": len(val_vals),
        "test_unique": len(test_vals),
        "train_val_overlap": len(train_val),
        "train_test_overlap": len(train_test),
        "val_test_overlap": len(val_test),
        "all_three_overlap": len(all_three)
    }


# ------------------------------------------------
# 4. UniProt overlap
# ------------------------------------------------

results = []

results.append(
    overlap_report(
        "uniprot",
        "UniProt / Protein ID"
    )
)

# ------------------------------------------------
# 5. Sequence-cluster overlap
# ------------------------------------------------

cluster_columns = [
    "sequence_20cluster",
    "sequence_40cluster",
    "sequence_60cluster",
    "sequence_80cluster",
    "sequence_99cluster"
]

for col in cluster_columns:

    if col in df.columns:

        results.append(
            overlap_report(
                col,
                col
            )
        )

# ------------------------------------------------
# 6. Exact sequence overlap
# ------------------------------------------------

print("\n" + "-" * 70)
print("Exact Protein Sequence Overlap")
print("-" * 70)

train_sequences = set(
    train_meta["sequence"]
    .dropna()
    .astype(str)
    .unique()
)

val_sequences = set(
    val_meta["sequence"]
    .dropna()
    .astype(str)
    .unique()
)

test_sequences = set(
    test_meta["sequence"]
    .dropna()
    .astype(str)
    .unique()
)

print(f"Train unique sequences:      {len(train_sequences):,}")
print(f"Validation unique sequences: {len(val_sequences):,}")
print(f"Test unique sequences:       {len(test_sequences):,}")

print(
    f"\nTrain ∩ Validation: "
    f"{len(train_sequences & val_sequences):,}"
)

print(
    f"Train ∩ Test: "
    f"{len(train_sequences & test_sequences):,}"
)

print(
    f"Validation ∩ Test: "
    f"{len(val_sequences & test_sequences):,}"
)

# ------------------------------------------------
# 7. Number of rows affected by protein overlap
# ------------------------------------------------

print("\n" + "=" * 70)
print("ROW-LEVEL OVERLAP")
print("=" * 70)

train_uniprot = set(
    train_meta["uniprot"]
    .dropna()
    .astype(str)
)

val_uniprot = set(
    val_meta["uniprot"]
    .dropna()
    .astype(str)
)

test_uniprot = set(
    test_meta["uniprot"]
    .dropna()
    .astype(str)
)

train_test_mask = (
    test_meta["uniprot"]
    .astype(str)
    .isin(train_uniprot)
)

val_test_mask = (
    test_meta["uniprot"]
    .astype(str)
    .isin(val_uniprot)
)

print(
    f"Test rows whose UniProt occurs in TRAIN: "
    f"{train_test_mask.sum():,} / {len(test_meta):,}"
)

print(
    f"Test rows whose UniProt occurs in VALIDATION: "
    f"{val_test_mask.sum():,} / {len(test_meta):,}"
)

# ------------------------------------------------
# 8. Save audit
# ------------------------------------------------

audit_df = pd.DataFrame(results)

save_path = (
    "/content/drive/MyDrive/"
    "CatPred_kcat_Project/results/"
    "protein_sequence_overlap_audit.csv"
)

audit_df.to_csv(save_path, index=False)

print("\n" + "=" * 70)
print("AUDIT COMPLETE")
print("=" * 70)

print(f"\nSaved to:")
print(save_path)

print("\nWe will use these results to decide the correct")
print("protein-level / sequence-level validation strategy.")
```

```python
# ============================================================
# CATPRED — CORRECT TRAIN/VALIDATION/TEST OVERLAP AUDIT
# ============================================================

import pandas as pd
import numpy as np

print("=" * 70)
print("CATPRED kcat — CORRECTED PROTEIN / SEQUENCE OVERLAP AUDIT")
print("=" * 70)

# ------------------------------------------------------------
# 1. Check actual split sizes
# ------------------------------------------------------------

print("\nACTUAL SPLIT SIZES")
print("-" * 70)

print("X_train:", X_train.shape)
print("X_val  :", X_val.shape)
print("X_test :", X_test.shape)

print("\nTarget sizes:")
print("y_train:", y_train.shape)
print("y_val  :", y_val.shape)
print("y_test :", y_test.shape)

print("\nOriginal dataset:", df.shape if 'df' in globals() else "df not found")

# ------------------------------------------------------------
# 2. Make sure row counts are consistent
# ------------------------------------------------------------

total_split = len(X_train) + len(X_val) + len(X_test)

print("\nTotal rows across splits:", total_split)

if 'df' in globals():
    print("Original dataset rows :", len(df))

    if total_split == len(df):
        print("✓ Split sizes sum correctly to original dataset")
    else:
        print("⚠ WARNING: split sizes do NOT sum to original dataset")

# ------------------------------------------------------------
# 3. Build metadata corresponding to each split
# ------------------------------------------------------------
#
# IMPORTANT:
# We need the original metadata for the actual rows.
#
# If X_train/X_val/X_test retained their original pandas indices,
# we can directly use those indices.
# ------------------------------------------------------------

if not isinstance(X_train, pd.DataFrame):
    raise TypeError("X_train must be a pandas DataFrame for this audit.")

if not isinstance(X_val, pd.DataFrame):
    raise TypeError("X_val must be a pandas DataFrame for this audit.")

if not isinstance(X_test, pd.DataFrame):
    raise TypeError("X_test must be a pandas DataFrame for this audit.")

print("\nIndex information:")
print("Train index:", X_train.index[:5].tolist())
print("Val index  :", X_val.index[:5].tolist())
print("Test index :", X_test.index[:5].tolist())

# ------------------------------------------------------------
# 4. Recover original metadata
# ------------------------------------------------------------

metadata_columns = [
    'uniprot',
    'sequence',
    'sequence_20cluster',
    'sequence_40cluster',
    'sequence_60cluster',
    'sequence_80cluster',
    'sequence_99cluster'
]

if 'df' not in globals():
    raise NameError(
        "Original dataframe 'df' is not available. "
        "Load the original dataset first."
    )

missing_cols = [c for c in metadata_columns if c not in df.columns]

if missing_cols:
    raise ValueError(
        f"Missing metadata columns: {missing_cols}"
    )

# ------------------------------------------------------------
# 5. Extract metadata using ORIGINAL ROW INDICES
# ------------------------------------------------------------

train_meta = df.loc[X_train.index, metadata_columns].copy()
val_meta   = df.loc[X_val.index, metadata_columns].copy()
test_meta  = df.loc[X_test.index, metadata_columns].copy()

print("\nMetadata recovered:")
print("Train:", train_meta.shape)
print("Val  :", val_meta.shape)
print("Test :", test_meta.shape)

# ------------------------------------------------------------
# 6. Function for overlap analysis
# ------------------------------------------------------------

def overlap_analysis(column):

    train_values = set(
        train_meta[column].dropna().astype(str)
    )

    val_values = set(
        val_meta[column].dropna().astype(str)
    )

    test_values = set(
        test_meta[column].dropna().astype(str)
    )

    train_val = train_values & val_values
    train_test = train_values & test_values
    val_test = val_values & test_values

    all_three = train_values & val_values & test_values

    return {
        "Feature": column,
        "Train_unique": len(train_values),
        "Validation_unique": len(val_values),
        "Test_unique": len(test_values),
        "Train_Validation_overlap": len(train_val),
        "Train_Test_overlap": len(train_test),
        "Validation_Test_overlap": len(val_test),
        "All_three_overlap": len(all_three)
    }

# ------------------------------------------------------------
# 7. Run audit
# ------------------------------------------------------------

results = []

for col in metadata_columns:
    results.append(overlap_analysis(col))

overlap_df = pd.DataFrame(results)

print("\n")
print("=" * 70)
print("OVERLAP RESULTS")
print("=" * 70)

print(overlap_df.to_string(index=False))

# ------------------------------------------------------------
# 8. Exact sequence row-level audit
# ------------------------------------------------------------

train_sequences = set(
    train_meta['sequence'].dropna().astype(str)
)

val_sequences = set(
    val_meta['sequence'].dropna().astype(str)
)

test_sequences = set(
    test_meta['sequence'].dropna().astype(str)
)

print("\n")
print("=" * 70)
print("EXACT SEQUENCE OVERLAP")
print("=" * 70)

print(
    "Train ∩ Validation:",
    len(train_sequences & val_sequences)
)

print(
    "Train ∩ Test:",
    len(train_sequences & test_sequences)
)

print(
    "Validation ∩ Test:",
    len(val_sequences & test_sequences)
)

# ------------------------------------------------------------
# 9. Row-level test leakage
# ------------------------------------------------------------

test_uniprot = test_meta['uniprot'].astype(str)

train_uniprot = set(
    train_meta['uniprot'].dropna().astype(str)
)

test_rows_with_train_protein = test_uniprot.isin(
    train_uniprot
).sum()

print("\n")
print("=" * 70)
print("ROW-LEVEL TEST OVERLAP")
print("=" * 70)

print(
    f"Test rows whose UniProt occurs in TRAIN: "
    f"{test_rows_with_train_protein} / {len(test_meta)}"
)

# ------------------------------------------------------------
# 10. Save
# ------------------------------------------------------------

SAVE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project/results"

import os
os.makedirs(SAVE_DIR, exist_ok=True)

output_file = os.path.join(
    SAVE_DIR,
    "corrected_protein_sequence_overlap_audit.csv"
)

overlap_df.to_csv(output_file, index=False)

print("\n✓ Saved:")
print(output_file)

print("\n")
print("=" * 70)
print("CORRECTED OVERLAP AUDIT COMPLETE")
print("=" * 70)
```

```python
# ================================================================
# CATPRED kcat — CREATE & VERIFY CORRECT DATA SPLIT
# ================================================================

import numpy as np
import pandas as pd
import os
from sklearn.model_selection import train_test_split

print("=" * 70)
print("CATPRED kcat — CREATING CORRECT SPLIT")
print("=" * 70)

# ------------------------------------------------
# 1. Original dataset
# ------------------------------------------------

print("\nOriginal dataset:")
print(df.shape)

assert len(df) == 18751, (
    f"Expected 18,751 rows, found {len(df)}"
)

# Preserve original row IDs
all_indices = df.index.to_numpy()

print("Original index range:",
      all_indices.min(), "to", all_indices.max())

# ------------------------------------------------
# 2. FIRST SPLIT
#    80% development + 20% final test
# ------------------------------------------------

train_val_idx, test_idx = train_test_split(
    all_indices,
    test_size=0.20,
    random_state=42,
    shuffle=True
)

# ------------------------------------------------
# 3. SECOND SPLIT
#    From the 80% development set:
#    87.5% train + 12.5% validation
#
#    This gives:
#    Train      = 70%
#    Validation = 10%
#    Test       = 20%
# ------------------------------------------------

train_idx, val_idx = train_test_split(
    train_val_idx,
    test_size=0.125,
    random_state=42,
    shuffle=True
)

# ------------------------------------------------
# 4. Convert to sets for verification
# ------------------------------------------------

train_set = set(train_idx)
val_set = set(val_idx)
test_set = set(test_idx)

# ------------------------------------------------
# 5. Check overlap
# ------------------------------------------------

print("\n" + "=" * 70)
print("SPLIT OVERLAP CHECK")
print("=" * 70)

print("Train ∩ Validation:",
      len(train_set & val_set))

print("Train ∩ Test:",
      len(train_set & test_set))

print("Validation ∩ Test:",
      len(val_set & test_set))

print("All three:",
      len(train_set & val_set & test_set))

# ------------------------------------------------
# 6. Assert zero overlap
# ------------------------------------------------

assert len(train_set & val_set) == 0
assert len(train_set & test_set) == 0
assert len(val_set & test_set) == 0

print("\n✓ ZERO ROW OVERLAP CONFIRMED")

# ------------------------------------------------
# 7. Check total number of observations
# ------------------------------------------------

total = (
    len(train_idx)
    + len(val_idx)
    + len(test_idx)
)

print("\nSplit sizes:")
print("Train      :", len(train_idx))
print("Validation :", len(val_idx))
print("Test       :", len(test_idx))
print("Total      :", total)
print("Original   :", len(df))

assert total == len(df)

print("\n✓ All 18,751 observations accounted for")

# ------------------------------------------------
# 8. Percentages
# ------------------------------------------------

print("\nPercentages:")

print(
    f"Train      : {len(train_idx)/len(df)*100:.2f}%"
)

print(
    f"Validation : {len(val_idx)/len(df)*100:.2f}%"
)

print(
    f"Test       : {len(test_idx)/len(df)*100:.2f}%"
)

# ------------------------------------------------
# 9. Create metadata subsets
# ------------------------------------------------

train_meta = df.loc[train_idx].copy()
val_meta   = df.loc[val_idx].copy()
test_meta  = df.loc[test_idx].copy()

# ------------------------------------------------
# 10. Protein overlap audit
# ------------------------------------------------

print("\n" + "=" * 70)
print("UNIPROT OVERLAP AUDIT")
print("=" * 70)

train_uniprot = set(
    train_meta["uniprot"].dropna().astype(str)
)

val_uniprot = set(
    val_meta["uniprot"].dropna().astype(str)
)

test_uniprot = set(
    test_meta["uniprot"].dropna().astype(str)
)

print("Unique UniProt:")
print("Train:", len(train_uniprot))
print("Val  :", len(val_uniprot))
print("Test :", len(test_uniprot))

print("\nUniProt overlap:")
print(
    "Train ∩ Validation:",
    len(train_uniprot & val_uniprot)
)

print(
    "Train ∩ Test:",
    len(train_uniprot & test_uniprot)
)

print(
    "Validation ∩ Test:",
    len(val_uniprot & test_uniprot)
)

# ------------------------------------------------
# 11. Exact sequence overlap
# ------------------------------------------------

print("\n" + "=" * 70)
print("EXACT SEQUENCE OVERLAP")
print("=" * 70)

train_seq = set(
    train_meta["sequence"].dropna().astype(str)
)

val_seq = set(
    val_meta["sequence"].dropna().astype(str)
)

test_seq = set(
    test_meta["sequence"].dropna().astype(str)
)

print("Unique sequences:")
print("Train:", len(train_seq))
print("Val  :", len(val_seq))
print("Test :", len(test_seq))

print("\nExact sequence overlap:")
print(
    "Train ∩ Validation:",
    len(train_seq & val_seq)
)

print(
    "Train ∩ Test:",
    len(train_seq & test_seq)
)

print(
    "Validation ∩ Test:",
    len(val_seq & test_seq)
)

# ------------------------------------------------
# 12. Sequence-cluster overlap
# ------------------------------------------------

print("\n" + "=" * 70)
print("SEQUENCE CLUSTER OVERLAP")
print("=" * 70)

cluster_columns = [
    "sequence_20cluster",
    "sequence_40cluster",
    "sequence_60cluster",
    "sequence_80cluster",
    "sequence_99cluster"
]

for col in cluster_columns:

    train_c = set(
        train_meta[col].dropna().astype(str)
    )

    val_c = set(
        val_meta[col].dropna().astype(str)
    )

    test_c = set(
        test_meta[col].dropna().astype(str)
    )

    print(f"\n{col}")

    print("Train unique:", len(train_c))
    print("Val unique  :", len(val_c))
    print("Test unique :", len(test_c))

    print(
        "Train ∩ Test:",
        len(train_c & test_c)
    )

    print(
        "Train ∩ Val:",
        len(train_c & val_c)
    )

# ------------------------------------------------
# 13. Save split indices
# ------------------------------------------------

SAVE_DIR = (
    "/content/drive/MyDrive/"
    "CatPred_kcat_Project/"
    "corrected_evaluation"
)

os.makedirs(SAVE_DIR, exist_ok=True)

np.save(
    os.path.join(SAVE_DIR, "train_indices.npy"),
    train_idx
)

np.save(
    os.path.join(SAVE_DIR, "validation_indices.npy"),
    val_idx
)

np.save(
    os.path.join(SAVE_DIR, "test_indices.npy"),
    test_idx
)

# Also save metadata
train_meta.to_csv(
    os.path.join(SAVE_DIR, "train_metadata.csv"),
    index=True
)

val_meta.to_csv(
    os.path.join(SAVE_DIR, "validation_metadata.csv"),
    index=True
)

test_meta.to_csv(
    os.path.join(SAVE_DIR, "test_metadata.csv"),
    index=True
)

print("\n" + "=" * 70)
print("SPLIT SAVED")
print("=" * 70)

print(
    "Location:",
    SAVE_DIR
)

print("\n✓ train_indices.npy")
print("✓ validation_indices.npy")
print("✓ test_indices.npy")
print("✓ train_metadata.csv")
print("✓ validation_metadata.csv")
print("✓ test_metadata.csv")

print("\n" + "=" * 70)
print("CORRECT SPLIT CREATED SUCCESSFULLY")
print("=" * 70)
```

```python
# ============================================================
# CATPRED kcat — CORRECTED BASELINE MODEL COMPARISON
# ============================================================

import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.linear_model import ElasticNet
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

from xgboost import XGBRegressor

print("=" * 70)
print("CATPRED kcat — CORRECTED BASELINE MODELS")
print("=" * 70)

# ------------------------------------------------------------
# Evaluation function
# ------------------------------------------------------------

def evaluate_model(model, X, y, name):

    pred = model.predict(X)

    r2 = r2_score(y, pred)
    rmse = np.sqrt(mean_squared_error(y, pred))
    mae = mean_absolute_error(y, pred)
    r, p = pearsonr(y, pred)

    print(f"\n{name}")
    print("-" * 60)
    print(f"R²          : {r2:.4f}")
    print(f"RMSE        : {rmse:.4f}")
    print(f"MAE         : {mae:.4f}")
    print(f"Pearson r   : {r:.4f}")
    print(f"Pearson p   : {p:.4e}")

    return {
        "Model": name,
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    }


results = []

# ============================================================
# 1. ElasticNet
# ============================================================

print("\n" + "=" * 60)
print("Training ElasticNet")
print("=" * 60)

elastic_model = ElasticNet(
    alpha=0.01,
    l1_ratio=0.5,
    max_iter=10000,
    random_state=42
)

elastic_model.fit(X_train, y_train)

elastic_result = evaluate_model(
    elastic_model,
    X_val,
    y_val,
    "ElasticNet"
)

results.append(elastic_result)


# ============================================================
# 2. ExtraTrees
# ============================================================

print("\n" + "=" * 60)
print("Training ExtraTrees")
print("=" * 60)

extra_model = ExtraTreesRegressor(
    n_estimators=300,
    max_features="sqrt",
    min_samples_leaf=2,
    random_state=42,
    n_jobs=-1
)

extra_model.fit(X_train, y_train)

extra_result = evaluate_model(
    extra_model,
    X_val,
    y_val,
    "ExtraTrees"
)

results.append(extra_result)


# ============================================================
# 3. XGBoost
# ============================================================

print("\n" + "=" * 60)
print("Training XGBoost")
print("=" * 60)

xgb_model = XGBRegressor(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train, y_train)

xgb_result = evaluate_model(
    xgb_model,
    X_val,
    y_val,
    "XGBoost"
)

results.append(xgb_result)


# ============================================================
# Results table
# ============================================================

baseline_results = pd.DataFrame(results)

baseline_results = baseline_results.sort_values(
    "R2",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("CORRECTED VALIDATION RESULTS")
print("=" * 70)

display(baseline_results)

# ============================================================
# Save
# ============================================================

SAVE_DIR = (
    "/content/drive/MyDrive/"
    "CatPred_kcat_Project/"
    "corrected_evaluation"
)

import os
os.makedirs(SAVE_DIR, exist_ok=True)

baseline_results.to_csv(
    os.path.join(
        SAVE_DIR,
        "corrected_baseline_validation_results.csv"
    ),
    index=False
)

print("\n✓ Saved:")
print(
    os.path.join(
        SAVE_DIR,
        "corrected_baseline_validation_results.csv"
    )
)
```

```python
# ============================================================
# CATPRED kcat — LIGHT EXTRATREES TUNING
# ============================================================

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr
import numpy as np
import pandas as pd
import os

print("=" * 70)
print("LIGHT EXTRATREES HYPERPARAMETER TUNING")
print("=" * 70)

param_grid = {
    "n_estimators": [200, 300, 400],
    "max_depth": [None, 10, 20, 30],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": [0.7, 1.0, "sqrt"]
}

base_model = ExtraTreesRegressor(
    random_state=42,
    n_jobs=-1
)

search = RandomizedSearchCV(
    estimator=base_model,
    param_distributions=param_grid,
    n_iter=15,
    scoring="r2",
    cv=3,
    random_state=42,
    n_jobs=-1,
    verbose=1
)

search.fit(X_train, y_train)

print("\n" + "=" * 70)
print("BEST PARAMETERS")
print("=" * 70)

print(search.best_params_)
print("\nBest CV R²:", search.best_score_)

# ------------------------------------------------------------
# Evaluate best model on VALIDATION
# ------------------------------------------------------------

tuned_extra_model = search.best_estimator_

val_pred = tuned_extra_model.predict(X_val)

val_r2 = r2_score(y_val, val_pred)
val_rmse = np.sqrt(mean_squared_error(y_val, val_pred))
val_mae = mean_absolute_error(y_val, val_pred)
val_r, val_p = pearsonr(y_val, val_pred)

print("\n" + "=" * 70)
print("TUNED EXTRATREES — VALIDATION PERFORMANCE")
print("=" * 70)

print(f"R²          : {val_r2:.4f}")
print(f"RMSE        : {val_rmse:.4f}")
print(f"MAE         : {val_mae:.4f}")
print(f"Pearson r   : {val_r:.4f}")
print(f"Pearson p   : {val_p:.4e}")

# ------------------------------------------------------------
# Save tuning results
# ------------------------------------------------------------

SAVE_DIR = (
    "/content/drive/MyDrive/"
    "CatPred_kcat_Project/"
    "corrected_evaluation"
)

os.makedirs(SAVE_DIR, exist_ok=True)

pd.DataFrame([{
    "R2": val_r2,
    "RMSE": val_rmse,
    "MAE": val_mae,
    "Pearson_r": val_r,
    "Pearson_p": val_p,
    "CV_R2": search.best_score_,
    **search.best_params_
}]).to_csv(
    os.path.join(
        SAVE_DIR,
        "tuned_extratrees_validation.csv"
    ),
    index=False
)

print("\n✓ Tuning results saved.")
```

```python
import os
import pandas as pd

BASE = "/content/drive/MyDrive/CatPred_kcat_Project"

print("="*70)
print("SEARCHING FOR ORIGINAL CATPRED DATASET")
print("="*70)

required_cols = {
    "sequence",
    "uniprot",
    "value",
    "temperature",
    "ph",
    "log10_value"
}

candidates = []

for root, dirs, files in os.walk(BASE):
    for file in files:
        path = os.path.join(root, file)

        if not file.lower().endswith((
            ".csv", ".tsv", ".pkl", ".pickle", ".parquet"
        )):
            continue

        try:
            if file.lower().endswith(".csv"):
                tmp = pd.read_csv(path, nrows=5)
            elif file.lower().endswith(".tsv"):
                tmp = pd.read_csv(path, sep="\t", nrows=5)
            elif file.lower().endswith((".pkl", ".pickle")):
                tmp = pd.read_pickle(path)

                if not isinstance(tmp, pd.DataFrame):
                    continue

                tmp = tmp.head(5)

            elif file.lower().endswith(".parquet"):
                tmp = pd.read_parquet(path).head(5)

            cols = set(tmp.columns)

            if required_cols.issubset(cols):
                candidates.append((path, cols))

        except Exception:
            pass

print("\nPossible original datasets:\n")

for i, (path, cols) in enumerate(candidates, 1):
    print(f"{i}. {path}")

if not candidates:
    print("❌ Original dataset was not found automatically.")
else:
    print(f"\n✓ Found {len(candidates)} candidate dataset(s)")
```

```python
import pandas as pd
import numpy as np
import os

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/corrected_evaluation"

print("=" * 70)
print("RECONSTRUCTING CORRECT CATPRED DATASET + 38 FEATURES")
print("=" * 70)

# ============================================================
# 1. LOAD THE THREE RECOVERED SPLITS
# ============================================================

train_meta = pd.read_csv(
    os.path.join(BASE, "train_metadata.csv")
)

val_meta = pd.read_csv(
    os.path.join(BASE, "validation_metadata.csv")
)

test_meta = pd.read_csv(
    os.path.join(BASE, "test_metadata.csv")
)

# ============================================================
# 2. COMBINE THEM
# ============================================================

df_all = pd.concat(
    [train_meta, val_meta, test_meta],
    axis=0,
    ignore_index=True
)

print("\nCombined metadata shape:", df_all.shape)

# ============================================================
# 3. RESTORE ORIGINAL ROW ORDER
# ============================================================

df_all = df_all.sort_values("Unnamed: 0").reset_index(drop=True)

# Verify original index
original_index = df_all["Unnamed: 0"].astype(int).values

assert np.array_equal(
    original_index,
    np.arange(18751)
), "Original index reconstruction failed!"

print("✓ Original row order restored")

# Remove artificial CSV index
df_all = df_all.drop(columns=["Unnamed: 0"])

print("Recovered original dataset:", df_all.shape)

# ============================================================
# 4. VERIFY TARGET
# ============================================================

assert "log10_value" in df_all.columns

y = df_all["log10_value"].astype(float).copy()

print("\nTarget:")
print("log10_value")
print("Shape:", y.shape)

# ============================================================
# 5. AMINO-ACID FEATURES
# ============================================================

amino_acids = list("ACDEFGHIKLMNPQRSTVWY")

feature_data = pd.DataFrame(index=df_all.index)

# Sequence length
feature_data["seq_length"] = (
    df_all["sequence"]
    .fillna("")
    .astype(str)
    .str.len()
)

# Amino-acid fractions
for aa in amino_acids:
    feature_data[f"aa_{aa}"] = (
        df_all["sequence"]
        .fillna("")
        .astype(str)
        .str.upper()
        .apply(
            lambda seq, aa=aa:
            seq.count(aa) / len(seq)
            if len(seq) > 0 else 0
        )
    )

# ============================================================
# 6. SEQUENCE COMPOSITION FEATURES
# ============================================================

hydrophobic = set("AILMFWVY")
charged = set("DEKR")
positive = set("KR")
negative = set("DE")
aromatic = set("FWY")
polar = set("STNQ")
small = set("AGST")

def fraction_from_set(seq, residues):
    seq = str(seq).upper()
    if len(seq) == 0:
        return 0
    return sum(aa in residues for aa in seq) / len(seq)

seq_series = df_all["sequence"].fillna("").astype(str)

feature_data["hydrophobic_fraction"] = seq_series.apply(
    lambda x: fraction_from_set(x, hydrophobic)
)

feature_data["charged_fraction"] = seq_series.apply(
    lambda x: fraction_from_set(x, charged)
)

feature_data["positive_fraction"] = seq_series.apply(
    lambda x: fraction_from_set(x, positive)
)

feature_data["negative_fraction"] = seq_series.apply(
    lambda x: fraction_from_set(x, negative)
)

feature_data["aromatic_fraction"] = seq_series.apply(
    lambda x: fraction_from_set(x, aromatic)
)

feature_data["polar_fraction"] = seq_series.apply(
    lambda x: fraction_from_set(x, polar)
)

feature_data["small_residue_fraction"] = seq_series.apply(
    lambda x: fraction_from_set(x, small)
)

# ============================================================
# 7. SUBSTRATE FEATURES
# ============================================================
#
# These must already exist in the feature-generation workflow.
# We calculate them from reactant_smiles using RDKit.
# ============================================================

from rdkit import Chem
from rdkit.Chem import Descriptors, Crippen, Lipinski

def substrate_features(smiles):

    try:
        mol = Chem.MolFromSmiles(smiles)

        if mol is None:
            return [np.nan] * 8

        return [
            Descriptors.MolWt(mol),
            Crippen.MolLogP(mol),
            Descriptors.TPSA(mol),
            Lipinski.NumHDonors(mol),
            Lipinski.NumHAcceptors(mol),
            Lipinski.NumRotatableBonds(mol),
            Lipinski.RingCount(mol),
            Descriptors.HeavyAtomCount(mol)
        ]

    except Exception:
        return [np.nan] * 8


substrate_names = [
    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms"
]

substrate_matrix = df_all["reactant_smiles"].apply(
    substrate_features
)

substrate_df = pd.DataFrame(
    substrate_matrix.tolist(),
    columns=substrate_names,
    index=df_all.index
)

feature_data = pd.concat(
    [feature_data, substrate_df],
    axis=1
)

# ============================================================
# 8. CONTEXT FEATURES
# ============================================================

feature_data["temperature"] = pd.to_numeric(
    df_all["temperature"],
    errors="coerce"
)

feature_data["ph"] = pd.to_numeric(
    df_all["ph"],
    errors="coerce"
)

# ============================================================
# 9. FINAL FEATURE CHECK
# ============================================================

expected_features = [
    "seq_length",
    "aa_A", "aa_C", "aa_D", "aa_E", "aa_F", "aa_G",
    "aa_H", "aa_I", "aa_K", "aa_L", "aa_M", "aa_N",
    "aa_P", "aa_Q", "aa_R", "aa_S", "aa_T", "aa_V",
    "aa_W", "aa_Y",
    "hydrophobic_fraction",
    "charged_fraction",
    "positive_fraction",
    "negative_fraction",
    "aromatic_fraction",
    "polar_fraction",
    "small_residue_fraction",
    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms",
    "temperature",
    "ph"
]

assert list(feature_data.columns) == expected_features

print("\n" + "=" * 70)
print("FEATURE MATRIX")
print("=" * 70)

print("Shape:", feature_data.shape)
print("Number of features:", len(feature_data.columns))

print("\nMissing values:")
print(feature_data.isna().sum().sum())

print("\nFeatures:")
print(feature_data.columns.tolist())
```

```python
print("=" * 70)
print("DIAGNOSING MISSING VALUES IN RECONSTRUCTED 38-FEATURE MATRIX")
print("=" * 70)

# ------------------------------------------------------------
# 1. Missing values by feature
# ------------------------------------------------------------

missing_by_feature = feature_data.isna().sum()

print("\nMissing values by feature:")
print(
    missing_by_feature[
        missing_by_feature > 0
    ].sort_values(ascending=False)
)

# ------------------------------------------------------------
# 2. How many rows have missing values?
# ------------------------------------------------------------

rows_with_missing = feature_data.isna().any(axis=1)

print("\nRows with at least one missing feature:")
print(rows_with_missing.sum())

print(
    "Percentage:",
    round(rows_with_missing.mean() * 100, 2),
    "%"
)

# ------------------------------------------------------------
# 3. Check which feature groups are affected
# ------------------------------------------------------------

print("\nSequence features missing:")
print(
    feature_data.iloc[:, :28]
    .isna()
    .sum()
    .sum()
)

print("\nSubstrate features missing:")
print(
    feature_data[
        [
            "mol_weight",
            "logp",
            "tpsa",
            "hbd",
            "hba",
            "rotatable_bonds",
            "rings",
            "heavy_atoms"
        ]
    ]
    .isna()
    .sum()
)

print("\nContext features missing:")
print(
    feature_data[
        ["temperature", "ph"]
    ]
    .isna()
)

# ------------------------------------------------------------
# 4. Check source columns
# ------------------------------------------------------------

print("\nOriginal source missingness:")

for col in [
    "sequence",
    "reactant_smiles",
    "temperature",
    "ph",
    "log10_value"
]:
    if col in df_all.columns:
        print(
            f"{col:20s}:",
            df_all[col].isna().sum()
        )

# ------------------------------------------------------------
# 5. Inspect problematic rows
# ------------------------------------------------------------

problem_rows = df_all.loc[
    rows_with_missing,
    [
        "sequence",
        "reactant_smiles",
        "temperature",
        "ph",
        "log10_value"
    ]
].copy()

print("\nFirst 10 problematic rows:")
display(problem_rows.head(10))
```

```python
import numpy as np
import os

print("=" * 70)
print("BUILDING CORRECT CATPRED FEATURE SPLITS")
print("=" * 70)

# Load the verified split indices
BASE = "/content/drive/MyDrive/CatPred_kcat_Project/corrected_evaluation"

train_indices = np.load(
    os.path.join(BASE, "train_indices.npy")
)

val_indices = np.load(
    os.path.join(BASE, "validation_indices.npy")
)

test_indices = np.load(
    os.path.join(BASE, "test_indices.npy")
)

# ------------------------------------------------------------
# Build feature splits
# ------------------------------------------------------------

X_train = feature_data.iloc[train_indices].copy()
X_val   = feature_data.iloc[val_indices].copy()
X_test  = feature_data.iloc[test_indices].copy()

y_train = y.iloc[train_indices].copy()
y_val   = y.iloc[val_indices].copy()
y_test  = y.iloc[test_indices].copy()

# ------------------------------------------------------------
# Verify dimensions
# ------------------------------------------------------------

print("\nX_train:", X_train.shape)
print("X_val  :", X_val.shape)
print("X_test :", X_test.shape)

print("\ny_train:", y_train.shape)
print("y_val  :", y_val.shape)
print("y_test :", y_test.shape)

assert X_train.shape == (13125, 38)
assert X_val.shape == (1875, 38)
assert X_test.shape == (3751, 38)

assert y_train.shape == (13125,)
assert y_val.shape == (1875,)
assert y_test.shape == (3751,)

# ------------------------------------------------------------
# Check feature order
# ------------------------------------------------------------

expected_features = [
    'seq_length',
    'aa_A', 'aa_C', 'aa_D', 'aa_E', 'aa_F', 'aa_G',
    'aa_H', 'aa_I', 'aa_K', 'aa_L', 'aa_M', 'aa_N',
    'aa_P', 'aa_Q', 'aa_R', 'aa_S', 'aa_T', 'aa_V',
    'aa_W', 'aa_Y',
    'hydrophobic_fraction',
    'charged_fraction',
    'positive_fraction',
    'negative_fraction',
    'aromatic_fraction',
    'polar_fraction',
    'small_residue_fraction',
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms',
    'temperature',
    'ph'
]

assert list(X_train.columns) == expected_features
assert list(X_val.columns) == expected_features
assert list(X_test.columns) == expected_features

print("\n✓ All three splits have identical 38 features")

# ------------------------------------------------------------
# Check row overlap
# ------------------------------------------------------------

train_set = set(train_indices)
val_set   = set(val_indices)
test_set  = set(test_indices)

print("\nROW OVERLAP")
print("Train ∩ Val :", len(train_set & val_set))
print("Train ∩ Test:", len(train_set & test_set))
print("Val ∩ Test  :", len(val_set & test_set))

assert len(train_set & val_set) == 0
assert len(train_set & test_set) == 0
assert len(val_set & test_set) == 0

print("\n✓ ZERO ROW OVERLAP")

# ------------------------------------------------------------
# Check missing values in each split
# ------------------------------------------------------------

print("\nMISSING VALUES")

print("Train:", X_train.isna().sum().sum())
print("Val  :", X_val.isna().sum().sum())
print("Test :", X_test.isna().sum().sum())

print("\nMissing by context feature:")

print(
    pd.DataFrame({
        "Train": X_train[["temperature", "ph"]].isna().sum(),
        "Validation": X_val[["temperature", "ph"]].isna().sum(),
        "Test": X_test[["temperature", "ph"]].isna().sum()
    })
)

print("\n" + "=" * 70)
print("✓ CORRECTED SPLITS READY")
print("=" * 70)
```

```python
from sklearn.ensemble import ExtraTreesRegressor

print("=" * 70)
print("RECREATING TUNED EXTRATREES — CORRECTED PIPELINE")
print("=" * 70)

final_model = ExtraTreesRegressor(
    n_estimators=200,
    min_samples_split=2,
    min_samples_leaf=2,
    max_features=0.7,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

print("\nModel:")
print(final_model)

print("\nTraining data:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nTraining model...")

final_model.fit(X_train, y_train)

print("\n✓ Tuned ExtraTrees trained successfully")
```

```python
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr
import numpy as np
import pandas as pd

print("=" * 70)
print("FINAL HELD-OUT TEST PERFORMANCE — CORRECTED CATPRED")
print("=" * 70)

# ============================================================
# PREDICT ON NEVER-SEEN TEST SET
# ============================================================

y_test_pred = final_model.predict(X_test)

# ============================================================
# METRICS
# ============================================================

r2 = r2_score(y_test, y_test_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_test_pred)
)

mae = mean_absolute_error(
    y_test,
    y_test_pred
)

pearson_r, pearson_p = pearsonr(
    y_test,
    y_test_pred
)

print("\n============================================================")
print("FINAL HELD-OUT TEST RESULTS")
print("============================================================")

print(f"R²          : {r2:.4f}")
print(f"RMSE        : {rmse:.4f}")
print(f"MAE         : {mae:.4f}")
print(f"Pearson r   : {pearson_r:.4f}")
print(f"Pearson p   : {pearson_p:.4e}")

# ============================================================
# SAVE RESULTS
# ============================================================

RESULT_DIR = "/content/drive/MyDrive/CatPred_kcat_Project/corrected_evaluation"

os.makedirs(RESULT_DIR, exist_ok=True)

final_test_performance = pd.DataFrame([{
    "Model": "Tuned ExtraTrees",
    "N_test": len(y_test),
    "R2": r2,
    "RMSE": rmse,
    "MAE": mae,
    "Pearson_r": pearson_r,
    "Pearson_p": pearson_p
}])

final_test_performance.to_csv(
    os.path.join(
        RESULT_DIR,
        "final_heldout_test_performance.csv"
    ),
    index=False
)

print("\n✓ Saved:")
print(
    os.path.join(
        RESULT_DIR,
        "final_heldout_test_performance.csv"
    )
)
```

```python
# ================================================================
# CATPRED — FEATURE GROUP IMPORTANCE ON CORRECTED HELD-OUT TEST
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

print("=" * 70)
print("CATPRED — FEATURE GROUP IMPORTANCE")
print("=" * 70)

# ------------------------------------------------------------
# Load corrected permutation importance
# ------------------------------------------------------------

importance_path = (
    "/content/drive/MyDrive/CatPred_kcat_Project/"
    "corrected_evaluation/corrected_test_permutation_importance.csv"
)

importance_df = pd.read_csv(importance_path)

print("\nLoaded feature importance:")
print(importance_df.shape)

# ------------------------------------------------------------
# Define biological feature groups
# ------------------------------------------------------------

sequence_features = [
    'seq_length',
    'aa_A', 'aa_C', 'aa_D', 'aa_E', 'aa_F', 'aa_G',
    'aa_H', 'aa_I', 'aa_K', 'aa_L', 'aa_M', 'aa_N',
    'aa_P', 'aa_Q', 'aa_R', 'aa_S', 'aa_T', 'aa_V',
    'aa_W', 'aa_Y',
    'hydrophobic_fraction',
    'charged_fraction',
    'positive_fraction',
    'negative_fraction',
    'aromatic_fraction',
    'polar_fraction',
    'small_residue_fraction'
]

substrate_features = [
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms'
]

context_features = [
    'temperature',
    'ph'
]

# ------------------------------------------------------------
# Assign groups
# ------------------------------------------------------------

def assign_group(feature):

    if feature in sequence_features:
        return "Sequence"

    elif feature in substrate_features:
        return "Substrate"

    elif feature in context_features:
        return "Context"

    else:
        return "Other"


importance_df["FeatureGroup"] = (
    importance_df["Feature"].apply(assign_group)
)

# ------------------------------------------------------------
# Summarize group importance
# ------------------------------------------------------------

group_importance = (
    importance_df
    .groupby("FeatureGroup")
    .agg(
        Importance_Mean=("Importance_Mean", "sum"),
        Importance_STD=("Importance_STD", "mean"),
        Number_of_Features=("Feature", "count")
    )
    .reset_index()
)

# Sort by importance
group_importance = group_importance.sort_values(
    "Importance_Mean",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("FEATURE GROUP IMPORTANCE")
print("=" * 70)

print(group_importance.to_string(index=False))

# ------------------------------------------------------------
# Save results
# ------------------------------------------------------------

SAVE_DIR = (
    "/content/drive/MyDrive/CatPred_kcat_Project/"
    "corrected_evaluation"
)

group_path = os.path.join(
    SAVE_DIR,
    "corrected_feature_group_importance.csv"
)

group_importance.to_csv(
    group_path,
    index=False
)

print("\n✓ Saved:")
print(group_path)

# ------------------------------------------------------------
# Figure
# ------------------------------------------------------------

plt.figure(figsize=(8, 5))

plt.bar(
    group_importance["FeatureGroup"],
    group_importance["Importance_Mean"]
)

plt.ylabel("Summed Permutation Importance")
plt.xlabel("Feature Group")
plt.title(
    "Feature Group Importance — Corrected Held-Out Test Set"
)

plt.tight_layout()

fig_path = os.path.join(
    SAVE_DIR,
    "corrected_feature_group_importance.png"
)

plt.savefig(
    fig_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("\n✓ Figure saved:")
print(fig_path)

print("\n" + "=" * 70)
print("FEATURE GROUP ANALYSIS COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# CORRECTED CATPRED — FEATURE GROUP ABLATION
# VALIDATION-BASED ANALYSIS
# ============================================================

import numpy as np
import pandas as pd
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

print("=" * 70)
print("CORRECTED CATPRED — FEATURE GROUP ABLATION")
print("=" * 70)

# ------------------------------------------------------------
# Feature groups
# ------------------------------------------------------------

sequence_features = [
    'seq_length',
    'aa_A', 'aa_C', 'aa_D', 'aa_E', 'aa_F', 'aa_G',
    'aa_H', 'aa_I', 'aa_K', 'aa_L', 'aa_M', 'aa_N',
    'aa_P', 'aa_Q', 'aa_R', 'aa_S', 'aa_T', 'aa_V',
    'aa_W', 'aa_Y',
    'hydrophobic_fraction',
    'charged_fraction',
    'positive_fraction',
    'negative_fraction',
    'aromatic_fraction',
    'polar_fraction',
    'small_residue_fraction'
]

substrate_features = [
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms'
]

context_features = [
    'temperature',
    'ph'
]

feature_groups = {
    "Full model": list(X_train.columns),
    "Without Sequence": [
        f for f in X_train.columns
        if f not in sequence_features
    ],
    "Without Substrate": [
        f for f in X_train.columns
        if f not in substrate_features
    ],
    "Without Context": [
        f for f in X_train.columns
        if f not in context_features
    ]
}

# ------------------------------------------------------------
# Evaluation function
# ------------------------------------------------------------

def evaluate_model(model, X_eval, y_eval):

    pred = model.predict(X_eval)

    r2 = r2_score(y_eval, pred)
    rmse = np.sqrt(mean_squared_error(y_eval, pred))
    mae = mean_absolute_error(y_eval, pred)
    r, p = pearsonr(y_eval, pred)

    return r2, rmse, mae, r, p


# ------------------------------------------------------------
# Run ablation
# ------------------------------------------------------------

results = []

for name, features in feature_groups.items():

    print("\n" + "=" * 60)
    print(f"Training: {name}")
    print("=" * 60)

    Xtr = X_train[features]
    Xv = X_val[features]

    model = ExtraTreesRegressor(
        n_estimators=200,
        max_features=0.7,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    )

    model.fit(Xtr, y_train)

    r2, rmse, mae, r, p = evaluate_model(
        model,
        Xv,
        y_val
    )

    print(f"Features    : {len(features)}")
    print(f"R²          : {r2:.4f}")
    print(f"RMSE        : {rmse:.4f}")
    print(f"MAE         : {mae:.4f}")
    print(f"Pearson r   : {r:.4f}")
    print(f"Pearson p   : {p:.4e}")

    results.append({
        "Model": name,
        "Num_Features": len(features),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    })

# ------------------------------------------------------------
# Results table
# ------------------------------------------------------------

ablation_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("FEATURE GROUP ABLATION RESULTS")
print("=" * 70)

print(ablation_df.to_string(index=False))

# ------------------------------------------------------------
# Calculate performance change relative to full model
# ------------------------------------------------------------

full_r2 = ablation_df.loc[
    ablation_df["Model"] == "Full model", "R2"
].iloc[0]

full_rmse = ablation_df.loc[
    ablation_df["Model"] == "Full model", "RMSE"
].iloc[0]

ablation_df["Delta_R2"] = ablation_df["R2"] - full_r2
ablation_df["Delta_RMSE"] = ablation_df["RMSE"] - full_rmse

print("\n" + "=" * 70)
print("PERFORMANCE CHANGE RELATIVE TO FULL MODEL")
print("=" * 70)

print(
    ablation_df[
        ["Model", "Delta_R2", "Delta_RMSE"]
    ].to_string(index=False)
)

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

ablation_path = (
    "/content/drive/MyDrive/"
    "CatPred_kcat_Project/"
    "corrected_evaluation/"
    "corrected_feature_group_ablation.csv"
)

ablation_df.to_csv(
    ablation_path,
    index=False
)

print("\n✓ Saved:")
print(ablation_path)
```

```python
# ============================================================
# CORRECTED CATPRED — FEATURE GROUP ABLATION
# ============================================================

import os
import pandas as pd
import numpy as np

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

# ------------------------------------------------------------
# CORRECTED SAVE DIRECTORY
# ------------------------------------------------------------

SAVE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project/corrected_evaluation"
os.makedirs(SAVE_DIR, exist_ok=True)

# ------------------------------------------------------------
# VERIFY CORRECT SPLITS
# ------------------------------------------------------------

print("=" * 70)
print("CORRECTED CATPRED — FEATURE GROUP ABLATION")
print("=" * 70)

print("\nData used:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_val  :", X_val.shape)
print("y_val  :", y_val.shape)
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)

# ------------------------------------------------------------
# FEATURE GROUPS
# ------------------------------------------------------------

sequence_features = [
    'seq_length',
    'aa_A', 'aa_C', 'aa_D', 'aa_E', 'aa_F', 'aa_G',
    'aa_H', 'aa_I', 'aa_K', 'aa_L', 'aa_M', 'aa_N',
    'aa_P', 'aa_Q', 'aa_R', 'aa_S', 'aa_T', 'aa_V',
    'aa_W', 'aa_Y',
    'hydrophobic_fraction',
    'charged_fraction',
    'positive_fraction',
    'negative_fraction',
    'aromatic_fraction',
    'polar_fraction',
    'small_residue_fraction'
]

substrate_features = [
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms'
]

context_features = [
    'temperature',
    'ph'
]

# Verify
all_features = (
    sequence_features +
    substrate_features +
    context_features
)

assert set(all_features) == set(X_train.columns)
assert len(all_features) == 38

print("\n✓ All 38 features accounted for")

# ------------------------------------------------------------
# FUNCTION TO TRAIN + EVALUATE
# ------------------------------------------------------------

def train_evaluate(feature_list, model_name):

    print("\n" + "=" * 60)
    print(f"Training: {model_name}")
    print("=" * 60)

    Xtr = X_train[feature_list]
    Xt = X_test[feature_list]

    model = ExtraTreesRegressor(
        n_estimators=200,
        max_features=0.7,
        min_samples_leaf=2,
        random_state=42,
        n_jobs=-1
    )

    model.fit(Xtr, y_train)

    predictions = model.predict(Xt)

    r2 = r2_score(y_test, predictions)
    rmse = np.sqrt(mean_squared_error(y_test, predictions))
    mae = mean_absolute_error(y_test, predictions)
    r, p = pearsonr(y_test, predictions)

    print(f"Features    : {len(feature_list)}")
    print(f"R²          : {r2:.4f}")
    print(f"RMSE        : {rmse:.4f}")
    print(f"MAE         : {mae:.4f}")
    print(f"Pearson r   : {r:.4f}")
    print(f"Pearson p   : {p:.4e}")

    return {
        "Model": model_name,
        "Num_Features": len(feature_list),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    }

# ------------------------------------------------------------
# RUN ABLATION ON THE CORRECTED HELD-OUT TEST SET
# ------------------------------------------------------------

results = []

# Full model
results.append(
    train_evaluate(
        all_features,
        "Full model"
    )
)

# Without sequence
results.append(
    train_evaluate(
        substrate_features + context_features,
        "Without Sequence"
    )
)

# Without substrate
results.append(
    train_evaluate(
        sequence_features + context_features,
        "Without Substrate"
    )
)

# Without context
results.append(
    train_evaluate(
        sequence_features + substrate_features,
        "Without Context"
    )
)

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

ablation_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("CORRECTED FEATURE GROUP ABLATION RESULTS")
print("=" * 70)

print(ablation_df.to_string(index=False))

# ------------------------------------------------------------
# PERFORMANCE CHANGE
# ------------------------------------------------------------

full_r2 = ablation_df.loc[
    ablation_df["Model"] == "Full model",
    "R2"
].iloc[0]

full_rmse = ablation_df.loc[
    ablation_df["Model"] == "Full model",
    "RMSE"
].iloc[0]

ablation_df["Delta_R2"] = ablation_df["R2"] - full_r2
ablation_df["Delta_RMSE"] = ablation_df["RMSE"] - full_rmse

print("\n" + "=" * 70)
print("PERFORMANCE CHANGE RELATIVE TO FULL MODEL")
print("=" * 70)

print(
    ablation_df[
        ["Model", "Delta_R2", "Delta_RMSE"]
    ].to_string(index=False)
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

output_path = os.path.join(
    SAVE_DIR,
    "corrected_feature_group_ablation_test.csv"
)

ablation_df.to_csv(
    output_path,
    index=False
)

print("\n✓ Saved:")
print(output_path)
```

```python
# ============================================================
# CORRECTED CATPRED — ERROR PERFORMANCE ACROSS kcat RANGES
# ============================================================

import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from scipy.stats import pearsonr

print("=" * 70)
print("ERROR PERFORMANCE ACROSS kcat RANGES")
print("=" * 70)

# Make sure these are NumPy arrays
actual = np.asarray(y_test)
predicted = np.asarray(full_predictions)

assert len(actual) == len(predicted)

# ------------------------------------------------------------
# Define kcat ranges using observed/test values
# ------------------------------------------------------------

q1 = np.quantile(actual, 1/3)
q2 = np.quantile(actual, 2/3)

print(f"\nObserved log10(kcat) tertile boundaries:")
print(f"Low / Medium  : {q1:.4f}")
print(f"Medium / High : {q2:.4f}")

# ------------------------------------------------------------
# Assign groups
# ------------------------------------------------------------

groups = np.where(
    actual <= q1,
    "Low kcat",
    np.where(
        actual <= q2,
        "Medium kcat",
        "High kcat"
    )
)

# ------------------------------------------------------------
# Calculate performance within each range
# ------------------------------------------------------------

results = []

for group_name in ["Low kcat", "Medium kcat", "High kcat"]:

    mask = groups == group_name

    y_true_group = actual[mask]
    y_pred_group = predicted[mask]

    r2 = r2_score(y_true_group, y_pred_group)
    rmse = np.sqrt(
        mean_squared_error(y_true_group, y_pred_group)
    )
    mae = mean_absolute_error(
        y_true_group, y_pred_group
    )

    if len(y_true_group) > 1:
        r, p = pearsonr(
            y_true_group,
            y_pred_group
        )
    else:
        r, p = np.nan, np.nan

    results.append({
        "kcat_group": group_name,
        "N": len(y_true_group),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    })

# ------------------------------------------------------------
# Results table
# ------------------------------------------------------------

range_results = pd.DataFrame(results)

print("\n" + "=" * 70)
print("ERROR PERFORMANCE ACROSS kcat RANGES")
print("=" * 70)

print(
    range_results.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

output_path = os.path.join(
    SAVE_DIR,
    "corrected_test_performance_by_kcat_range.csv"
)

range_results.to_csv(
    output_path,
    index=False
)

print("\n✓ Saved:")
print(output_path)
```

```python
# ============================================================
# CORRECTED CATPRED — FINAL HELD-OUT TEST RESIDUAL ANALYSIS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)

from scipy.stats import pearsonr

print("=" * 70)
print("CORRECTED CATPRED — HELD-OUT TEST RESIDUAL ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# OUTPUT DIRECTORY
# ------------------------------------------------------------

SAVE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project/corrected_evaluation"
os.makedirs(SAVE_DIR, exist_ok=True)

# ------------------------------------------------------------
# VERIFY CORRECT TEST DATA
# ------------------------------------------------------------

actual = np.asarray(y_test)
predicted = np.asarray(full_predictions)

print("\nTest data:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
print("Predictions:", predicted.shape)

assert len(actual) == 3751
assert len(predicted) == 3751
assert len(actual) == len(predicted)

print("✓ Correct 3,751-row held-out test set confirmed")

# ------------------------------------------------------------
# RESIDUALS
# ------------------------------------------------------------

residuals = actual - predicted
absolute_errors = np.abs(residuals)

error_df = pd.DataFrame({
    "Observed_log10_kcat": actual,
    "Predicted_log10_kcat": predicted,
    "Residual": residuals,
    "Absolute_Error": absolute_errors
})

# ------------------------------------------------------------
# OVERALL METRICS
# ------------------------------------------------------------

r2 = r2_score(actual, predicted)
rmse = np.sqrt(mean_squared_error(actual, predicted))
mae = mean_absolute_error(actual, predicted)
pearson_r, pearson_p = pearsonr(actual, predicted)

print("\n" + "=" * 70)
print("OVERALL HELD-OUT TEST PERFORMANCE")
print("=" * 70)

print(f"R²        : {r2:.4f}")
print(f"RMSE      : {rmse:.4f}")
print(f"MAE       : {mae:.4f}")
print(f"Pearson r : {pearson_r:.4f}")
print(f"Pearson p : {pearson_p:.4e}")

# ------------------------------------------------------------
# 1. OBSERVED VS PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

plt.scatter(
    actual,
    predicted,
    alpha=0.35,
    s=18
)

min_val = min(actual.min(), predicted.min())
max_val = max(actual.max(), predicted.max())

plt.plot(
    [min_val, max_val],
    [min_val, max_val],
    linestyle="--"
)

plt.xlabel("Observed log10(kcat)")
plt.ylabel("Predicted log10(kcat)")
plt.title("Observed vs Predicted kcat — Held-Out Test Set")

plt.text(
    0.05,
    0.95,
    f"$R^2$ = {r2:.3f}\n"
    f"RMSE = {rmse:.3f}\n"
    f"MAE = {mae:.3f}",
    transform=plt.gca().transAxes,
    verticalalignment="top"
)

plt.tight_layout()

path = os.path.join(
    SAVE_DIR,
    "corrected_observed_vs_predicted_kcat.png"
)

plt.savefig(path, dpi=300, bbox_inches="tight")
plt.show()

print("✓ Saved:", path)

# ------------------------------------------------------------
# 2. RESIDUALS VS PREDICTED
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

plt.scatter(
    predicted,
    residuals,
    alpha=0.35,
    s=18
)

plt.axhline(
    0,
    linestyle="--"
)

plt.xlabel("Predicted log10(kcat)")
plt.ylabel("Residual (Observed − Predicted)")
plt.title("Residuals vs Predicted kcat — Held-Out Test Set")

plt.tight_layout()

path = os.path.join(
    SAVE_DIR,
    "corrected_residuals_vs_predicted_kcat.png"
)

plt.savefig(path, dpi=300, bbox_inches="tight")
plt.show()

print("✓ Saved:", path)

# ------------------------------------------------------------
# 3. RESIDUAL DISTRIBUTION
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

plt.hist(
    residuals,
    bins=50,
    edgecolor="black"
)

plt.axvline(
    0,
    linestyle="--"
)

plt.xlabel("Residual (Observed − Predicted)")
plt.ylabel("Frequency")
plt.title("Residual Distribution — Held-Out Test Set")

plt.tight_layout()

path = os.path.join(
    SAVE_DIR,
    "corrected_residual_distribution.png"
)

plt.savefig(path, dpi=300, bbox_inches="tight")
plt.show()

print("✓ Saved:", path)

# ------------------------------------------------------------
# 4. ABSOLUTE ERROR VS OBSERVED kcat
# ------------------------------------------------------------

plt.figure(figsize=(7, 6))

plt.scatter(
    actual,
    absolute_errors,
    alpha=0.35,
    s=18
)

plt.xlabel("Observed log10(kcat)")
plt.ylabel("Absolute Error")
plt.title("Absolute Prediction Error vs Observed kcat")

plt.tight_layout()

path = os.path.join(
    SAVE_DIR,
    "corrected_absolute_error_vs_observed_kcat.png"
)

plt.savefig(path, dpi=300, bbox_inches="tight")
plt.show()

print("✓ Saved:", path)

# ------------------------------------------------------------
# 5. TOP 20 LARGEST ERRORS
# ------------------------------------------------------------

top_errors = (
    error_df
    .sort_values("Absolute_Error", ascending=False)
    .head(20)
    .copy()
)

top_errors.insert(
    0,
    "Rank",
    range(1, len(top_errors) + 1)
)

path = os.path.join(
    SAVE_DIR,
    "corrected_top_20_prediction_errors.csv"
)

top_errors.to_csv(
    path,
    index=False
)

print("\n✓ Saved:", path)

# ------------------------------------------------------------
# 6. TOP 10 UNDER-PREDICTIONS
# ------------------------------------------------------------

under_predictions = (
    error_df
    .sort_values("Residual", ascending=False)
    .head(10)
    .copy()
)

under_predictions.insert(
    0,
    "Rank",
    range(1, len(under_predictions) + 1)
)

path = os.path.join(
    SAVE_DIR,
    "corrected_top_10_under_predictions.csv"
)

under_predictions.to_csv(
    path,
    index=False
)

print("✓ Saved:", path)

# ------------------------------------------------------------
# 7. TOP 10 OVER-PREDICTIONS
# ------------------------------------------------------------

over_predictions = (
    error_df
    .sort_values("Residual", ascending=True)
    .head(10)
    .copy()
)

over_predictions.insert(
    0,
    "Rank",
    range(1, len(over_predictions) + 1)
)

path = os.path.join(
    SAVE_DIR,
    "corrected_top_10_over_predictions.csv"
)

over_predictions.to_csv(
    path,
    index=False
)

print("✓ Saved:", path)

# ------------------------------------------------------------
# 8. RESIDUAL SUMMARY
# ------------------------------------------------------------

summary = pd.DataFrame({
    "Metric": [
        "Test_N",
        "R2",
        "RMSE",
        "MAE",
        "Pearson_r",
        "Pearson_p",
        "Mean_Residual",
        "Median_Residual",
        "Residual_STD",
        "Mean_Absolute_Error",
        "Max_Absolute_Error"
    ],
    "Value": [
        len(actual),
        r2,
        rmse,
        mae,
        pearson_r,
        pearson_p,
        np.mean(residuals),
        np.median(residuals),
        np.std(residuals),
        np.mean(absolute_errors),
        np.max(absolute_errors)
    ]
})

path = os.path.join(
    SAVE_DIR,
    "corrected_residual_analysis_summary.csv"
)

summary.to_csv(
    path,
    index=False
)

print("✓ Saved:", path)

# ------------------------------------------------------------
# 9. FULL ERROR TABLE
# ------------------------------------------------------------

path = os.path.join(
    SAVE_DIR,
    "corrected_residual_error_analysis.csv"
)

error_df.to_csv(
    path,
    index=False
)

print("✓ Saved:", path)

# ------------------------------------------------------------
# FINAL
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("CORRECTED RESIDUAL ANALYSIS COMPLETE")
print("=" * 70)

print("\nAll analyses use:")
print("✓ Correct 13,125-row training set")
print("✓ Correct 1,875-row validation set")
print("✓ Correct 3,751-row held-out test set")
print("✓ Corrected final ExtraTrees model")
print("✓ No old 2,316-row test data")
```

```python
import pandas as pd
import os

print("=" * 70)
print("CORRECTED CATPRED — PROTEIN SEQUENCE OVERLAP AUDIT")
print("=" * 70)

BASE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project"
CORRECTED_DIR = os.path.join(BASE_DIR, "corrected_evaluation")

train_meta = pd.read_csv(
    os.path.join(CORRECTED_DIR, "train_metadata.csv")
)

val_meta = pd.read_csv(
    os.path.join(CORRECTED_DIR, "validation_metadata.csv")
)

test_meta = pd.read_csv(
    os.path.join(CORRECTED_DIR, "test_metadata.csv")
)

print("\nDataset sizes:")
print("Train:", len(train_meta))
print("Validation:", len(val_meta))
print("Test:", len(test_meta))


# ============================================================
# 1. EXACT SEQUENCE OVERLAP
# ============================================================

print("\n" + "=" * 70)
print("EXACT PROTEIN SEQUENCE OVERLAP")
print("=" * 70)

train_sequences = set(train_meta["sequence"].dropna())
val_sequences = set(val_meta["sequence"].dropna())
test_sequences = set(test_meta["sequence"].dropna())

train_val_overlap = train_sequences & val_sequences
train_test_overlap = train_sequences & test_sequences
val_test_overlap = val_sequences & test_sequences

print(f"\nTrain ∩ Validation : {len(train_val_overlap)}")
print(f"Train ∩ Test       : {len(train_test_overlap)}")
print(f"Validation ∩ Test  : {len(val_test_overlap)}")


# ============================================================
# 2. CLUSTER-LEVEL OVERLAP
# ============================================================

cluster_columns = [
    "sequence_20cluster",
    "sequence_40cluster",
    "sequence_60cluster",
    "sequence_80cluster",
    "sequence_99cluster"
]

results = []

print("\n" + "=" * 70)
print("SEQUENCE CLUSTER OVERLAP")
print("=" * 70)

for col in cluster_columns:

    train_clusters = set(train_meta[col].dropna())
    val_clusters = set(val_meta[col].dropna())
    test_clusters = set(test_meta[col].dropna())

    tv = train_clusters & val_clusters
    tt = train_clusters & test_clusters
    vt = val_clusters & test_clusters

    results.append({
        "Cluster_Level": col,
        "Train_Val_Overlap": len(tv),
        "Train_Test_Overlap": len(tt),
        "Val_Test_Overlap": len(vt),
        "Train_Clusters": len(train_clusters),
        "Val_Clusters": len(val_clusters),
        "Test_Clusters": len(test_clusters)
    })

    print(f"\n{col}")
    print(f"  Train ∩ Val  : {len(tv)}")
    print(f"  Train ∩ Test : {len(tt)}")
    print(f"  Val ∩ Test   : {len(vt)}")


# ============================================================
# 3. SAVE AUDIT
# ============================================================

overlap_df = pd.DataFrame(results)

audit_path = os.path.join(
    CORRECTED_DIR,
    "corrected_protein_sequence_overlap_audit.csv"
)

overlap_df.to_csv(
    audit_path,
    index=False
)

print("\n" + "=" * 70)
print("AUDIT SAVED")
print("=" * 70)

print(audit_path)

print("\n✓ Sequence leakage audit complete")
```

```python
# ============================================================
# CATPRED — PROTEIN-LEVEL / SEQUENCE-CLUSTER-AWARE SPLIT
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupShuffleSplit

print("=" * 70)
print("CATPRED — SEQUENCE-CLUSTER-AWARE PROTEIN-LEVEL SPLIT")
print("=" * 70)

# ------------------------------------------------------------
# PATHS
# ------------------------------------------------------------

BASE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project"
CORRECTED_DIR = os.path.join(BASE_DIR, "corrected_evaluation")

PROTEIN_DIR = os.path.join(
    BASE_DIR,
    "protein_level_evaluation"
)

os.makedirs(PROTEIN_DIR, exist_ok=True)

# ------------------------------------------------------------
# LOAD THE COMPLETE 18,751-ROW DATASET
# ------------------------------------------------------------

print("\nLoading saved metadata splits...")

train_meta = pd.read_csv(
    os.path.join(CORRECTED_DIR, "train_metadata.csv")
)

val_meta = pd.read_csv(
    os.path.join(CORRECTED_DIR, "validation_metadata.csv")
)

test_meta = pd.read_csv(
    os.path.join(CORRECTED_DIR, "test_metadata.csv")
)

# Combine them back into the complete dataset
df = pd.concat(
    [train_meta, val_meta, test_meta],
    ignore_index=True
)

# Original row index stored in Unnamed: 0
df["original_index"] = df["Unnamed: 0"].astype(int)

# Sort by original dataset index
df = df.sort_values(
    "original_index"
).reset_index(drop=True)

print("\nComplete dataset:")
print(df.shape)

assert len(df) == 18751

assert df["original_index"].nunique() == 18751

print("✓ Complete 18,751-row dataset recovered")

# ------------------------------------------------------------
# CHECK SEQUENCE CLUSTER
# ------------------------------------------------------------

GROUP_COLUMN = "sequence_20cluster"

assert GROUP_COLUMN in df.columns

missing_groups = df[GROUP_COLUMN].isna().sum()

print("\nSequence-cluster information:")
print("Grouping variable:", GROUP_COLUMN)
print("Missing clusters:", missing_groups)
print("Unique clusters:", df[GROUP_COLUMN].nunique())

if missing_groups > 0:
    raise ValueError(
        "Missing sequence_20cluster values detected. "
        "Do not create the protein-level split until these are handled."
    )

# ------------------------------------------------------------
# FIRST SPLIT:
# 80% DEVELOPMENT + 20% HELD-OUT TEST
#
# IMPORTANT:
# Splitting is performed on GROUPS, not individual rows.
# ------------------------------------------------------------

groups = df[GROUP_COLUMN].values

gss_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

development_idx, test_idx = next(
    gss_test.split(
        df,
        groups=groups
    )
)

# ------------------------------------------------------------
# SECOND SPLIT:
# DEVELOPMENT → 87.5% TRAIN + 12.5% VALIDATION
#
# This produces approximately:
# 70% TRAIN
# 10% VALIDATION
# 20% TEST
# ------------------------------------------------------------

development_df = df.iloc[development_idx].copy()

development_groups = development_df[
    GROUP_COLUMN
].values

gss_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.125,
    random_state=42
)

train_local_idx, val_local_idx = next(
    gss_val.split(
        development_df,
        groups=development_groups
    )
)

train_df = development_df.iloc[
    train_local_idx
].copy()

val_df = development_df.iloc[
    val_local_idx
].copy()

test_df = df.iloc[
    test_idx
].copy()

# ------------------------------------------------------------
# SORT BY ORIGINAL INDEX FOR EASY TRACKING
# ------------------------------------------------------------

train_df = train_df.sort_values(
    "original_index"
).reset_index(drop=True)

val_df = val_df.sort_values(
    "original_index"
).reset_index(drop=True)

test_df = test_df.sort_values(
    "original_index"
).reset_index(drop=True)

# ------------------------------------------------------------
# BASIC SPLIT INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PROTEIN-LEVEL SPLIT SIZES")
print("=" * 70)

print(f"Total      : {len(df)}")
print(f"Train      : {len(train_df)}")
print(f"Validation : {len(val_df)}")
print(f"Test       : {len(test_df)}")

print("\nPercentages:")

print(
    f"Train      : {len(train_df) / len(df) * 100:.2f}%"
)

print(
    f"Validation : {len(val_df) / len(df) * 100:.2f}%"
)

print(
    f"Test       : {len(test_df) / len(df) * 100:.2f}%"
)

# ------------------------------------------------------------
# ROW-LEVEL OVERLAP CHECK
# ------------------------------------------------------------

train_rows = set(train_df["original_index"])
val_rows = set(val_df["original_index"])
test_rows = set(test_df["original_index"])

print("\n" + "=" * 70)
print("ROW OVERLAP")
print("=" * 70)

print(
    "Train ∩ Validation:",
    len(train_rows & val_rows)
)

print(
    "Train ∩ Test:",
    len(train_rows & test_rows)
)

print(
    "Validation ∩ Test:",
    len(val_rows & test_rows)
)

assert len(train_rows & val_rows) == 0
assert len(train_rows & test_rows) == 0
assert len(val_rows & test_rows) == 0

print("✓ ZERO ROW OVERLAP")

# ------------------------------------------------------------
# SEQUENCE CLUSTER OVERLAP CHECK
# ------------------------------------------------------------

train_clusters = set(
    train_df[GROUP_COLUMN]
)

val_clusters = set(
    val_df[GROUP_COLUMN]
)

test_clusters = set(
    test_df[GROUP_COLUMN]
)

print("\n" + "=" * 70)
print("SEQUENCE-CLUSTER OVERLAP")
print("=" * 70)

train_val_cluster_overlap = (
    train_clusters & val_clusters
)

train_test_cluster_overlap = (
    train_clusters & test_clusters
)

val_test_cluster_overlap = (
    val_clusters & test_clusters
)

print(
    "Train ∩ Validation:",
    len(train_val_cluster_overlap)
)

print(
    "Train ∩ Test:",
    len(train_test_cluster_overlap)
)

print(
    "Validation ∩ Test:",
    len(val_test_cluster_overlap)
)

assert len(train_val_cluster_overlap) == 0
assert len(train_test_cluster_overlap) == 0
assert len(val_test_cluster_overlap) == 0

print("✓ ZERO SEQUENCE-CLUSTER OVERLAP")

# ------------------------------------------------------------
# EXACT SEQUENCE OVERLAP CHECK
# ------------------------------------------------------------

train_sequences = set(
    train_df["sequence"]
)

val_sequences = set(
    val_df["sequence"]
)

test_sequences = set(
    test_df["sequence"]
)

print("\n" + "=" * 70)
print("EXACT SEQUENCE OVERLAP")
print("=" * 70)

train_val_seq = (
    train_sequences & val_sequences
)

train_test_seq = (
    train_sequences & test_sequences
)

val_test_seq = (
    val_sequences & test_sequences
)

print(
    "Train ∩ Validation:",
    len(train_val_seq)
)

print(
    "Train ∩ Test:",
    len(train_test_seq)
)

print(
    "Validation ∩ Test:",
    len(val_test_seq)
)

assert len(train_val_seq) == 0
assert len(train_test_seq) == 0
assert len(val_test_seq) == 0

print("✓ ZERO EXACT-SEQUENCE OVERLAP")

# ------------------------------------------------------------
# CLUSTER COUNTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("UNIQUE SEQUENCE CLUSTERS")
print("=" * 70)

print(
    "Train clusters:",
    len(train_clusters)
)

print(
    "Validation clusters:",
    len(val_clusters)
)

print(
    "Test clusters:",
    len(test_clusters)
)

# ------------------------------------------------------------
# TARGET DISTRIBUTION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TARGET DISTRIBUTION")
print("=" * 70)

for name, subset in [
    ("Train", train_df),
    ("Validation", val_df),
    ("Test", test_df)
]:

    print(f"\n{name}")

    print(
        "N:",
        len(subset)
    )

    print(
        "Mean log10(kcat):",
        subset["log10_value"].mean()
    )

    print(
        "Std:",
        subset["log10_value"].std()
    )

    print(
        "Min:",
        subset["log10_value"].min()
    )

    print(
        "Max:",
        subset["log10_value"].max()
    )

# ------------------------------------------------------------
# SAVE INDICES
# ------------------------------------------------------------

np.save(
    os.path.join(
        PROTEIN_DIR,
        "protein_train_indices.npy"
    ),
    train_df["original_index"].values
)

np.save(
    os.path.join(
        PROTEIN_DIR,
        "protein_validation_indices.npy"
    ),
    val_df["original_index"].values
)

np.save(
    os.path.join(
        PROTEIN_DIR,
        "protein_test_indices.npy"
    ),
    test_df["original_index"].values
)

# ------------------------------------------------------------
# SAVE METADATA
# ------------------------------------------------------------

train_df.to_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_train_metadata.csv"
    ),
    index=False
)

val_df.to_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_validation_metadata.csv"
    ),
    index=False
)

test_df.to_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_test_metadata.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# SAVE SPLIT SUMMARY
# ------------------------------------------------------------

summary = pd.DataFrame({
    "Split": [
        "Train",
        "Validation",
        "Test"
    ],
    "Rows": [
        len(train_df),
        len(val_df),
        len(test_df)
    ],
    "Percentage": [
        len(train_df) / len(df) * 100,
        len(val_df) / len(df) * 100,
        len(test_df) / len(df) * 100
    ],
    "Unique_20Clusters": [
        len(train_clusters),
        len(val_clusters),
        len(test_clusters)
    ]
})

summary.to_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_level_split_summary.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# FINAL CHECK
# ------------------------------------------------------------

combined_indices = (
    train_rows |
    val_rows |
    test_rows
)

assert len(combined_indices) == 18751

print("\n" + "=" * 70)
print("PROTEIN-LEVEL SPLIT COMPLETE")
print("=" * 70)

print("\n✓ 18,751 observations accounted for")
print("✓ Zero row overlap")
print("✓ Zero sequence-cluster overlap")
print("✓ Zero exact-sequence overlap")

print("\nSaved to:")
print(PROTEIN_DIR)

print("\nFiles:")
print("✓ protein_train_indices.npy")
print("✓ protein_validation_indices.npy")
print("✓ protein_test_indices.npy")
print("✓ protein_train_metadata.csv")
print("✓ protein_validation_metadata.csv")
print("✓ protein_test_metadata.csv")
print("✓ protein_level_split_summary.csv")

print("\nDO NOT TRAIN THE MODEL YET.")
print("First inspect the output above.")
```

```python
# ============================================================
# CATPRED — RECONSTRUCT 38 FEATURES FOR PROTEIN-DISJOINT SPLIT
# ============================================================

import os
import numpy as np
import pandas as pd

from rdkit import Chem
from rdkit.Chem import Descriptors, Crippen, Lipinski

print("=" * 70)
print("RECONSTRUCTING 38 FEATURES")
print("=" * 70)

BASE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project"
PROTEIN_DIR = os.path.join(
    BASE_DIR,
    "protein_level_evaluation"
)

# ------------------------------------------------------------
# LOAD METADATA
# ------------------------------------------------------------

train_meta = pd.read_csv(
    os.path.join(PROTEIN_DIR, "protein_train_metadata.csv")
)

val_meta = pd.read_csv(
    os.path.join(PROTEIN_DIR, "protein_validation_metadata.csv")
)

test_meta = pd.read_csv(
    os.path.join(PROTEIN_DIR, "protein_test_metadata.csv")
)

# ------------------------------------------------------------
# FEATURE DEFINITIONS
# ------------------------------------------------------------

AA = list("ACDEFGHIKLMNPQRSTVWY")

hydrophobic = set("AVILMFWY")
charged = set("DEKR")
positive = set("KR")
negative = set("DE")
aromatic = set("FWY")
polar = set("STNQ")
small = set("AGST")

# ------------------------------------------------------------
# SEQUENCE FEATURES
# ------------------------------------------------------------

def sequence_features(sequence):

    sequence = str(sequence).strip().upper()

    length = len(sequence)

    if length == 0:
        raise ValueError("Empty protein sequence encountered.")

    counts = {
        aa: sequence.count(aa)
        for aa in AA
    }

    features = {
        "seq_length": length
    }

    # Amino-acid composition
    for aa in AA:
        features[f"aa_{aa}"] = (
            counts[aa] / length
        )

    # Physicochemical fractions
    features["hydrophobic_fraction"] = (
        sum(sequence.count(aa) for aa in hydrophobic)
        / length
    )

    features["charged_fraction"] = (
        sum(sequence.count(aa) for aa in charged)
        / length
    )

    features["positive_fraction"] = (
        sum(sequence.count(aa) for aa in positive)
        / length
    )

    features["negative_fraction"] = (
        sum(sequence.count(aa) for aa in negative)
        / length
    )

    features["aromatic_fraction"] = (
        sum(sequence.count(aa) for aa in aromatic)
        / length
    )

    features["polar_fraction"] = (
        sum(sequence.count(aa) for aa in polar)
        / length
    )

    features["small_residue_fraction"] = (
        sum(sequence.count(aa) for aa in small)
        / length
    )

    return features


# ------------------------------------------------------------
# SUBSTRATE FEATURES
# ------------------------------------------------------------

def substrate_features(smiles):

    mol = Chem.MolFromSmiles(str(smiles))

    if mol is None:
        return {
            "mol_weight": np.nan,
            "logp": np.nan,
            "tpsa": np.nan,
            "hbd": np.nan,
            "hba": np.nan,
            "rotatable_bonds": np.nan,
            "rings": np.nan,
            "heavy_atoms": np.nan
        }

    return {
        "mol_weight": Descriptors.MolWt(mol),
        "logp": Crippen.MolLogP(mol),
        "tpsa": Descriptors.TPSA(mol),
        "hbd": Lipinski.NumHDonors(mol),
        "hba": Lipinski.NumHAcceptors(mol),
        "rotatable_bonds": Lipinski.NumRotatableBonds(mol),
        "rings": Lipinski.RingCount(mol),
        "heavy_atoms": Descriptors.HeavyAtomCount(mol)
    }


# ------------------------------------------------------------
# BUILD FEATURES FOR ONE DATASET
# ------------------------------------------------------------

def build_features(meta):

    rows = []

    for _, row in meta.iterrows():

        seq_part = sequence_features(
            row["sequence"]
        )

        substrate_part = substrate_features(
            row["reactant_smiles"]
        )

        combined = {
            **seq_part,
            **substrate_part,
            "temperature": row["temperature"],
            "ph": row["ph"]
        }

        rows.append(combined)

    return pd.DataFrame(rows)


# ------------------------------------------------------------
# RECONSTRUCT
# ------------------------------------------------------------

print("\nBuilding TRAIN features...")

X_protein_train = build_features(
    train_meta
)

print("Building VALIDATION features...")

X_protein_val = build_features(
    val_meta
)

print("Building TEST features...")

X_protein_test = build_features(
    test_meta
)

# ------------------------------------------------------------
# EXACT FEATURE ORDER
# ------------------------------------------------------------

FEATURES = [
    "seq_length",

    "aa_A", "aa_C", "aa_D", "aa_E", "aa_F", "aa_G",
    "aa_H", "aa_I", "aa_K", "aa_L", "aa_M", "aa_N",
    "aa_P", "aa_Q", "aa_R", "aa_S", "aa_T", "aa_V",
    "aa_W", "aa_Y",

    "hydrophobic_fraction",
    "charged_fraction",
    "positive_fraction",
    "negative_fraction",
    "aromatic_fraction",
    "polar_fraction",
    "small_residue_fraction",

    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms",

    "temperature",
    "ph"
]

X_protein_train = X_protein_train[FEATURES]
X_protein_val = X_protein_val[FEATURES]
X_protein_test = X_protein_test[FEATURES]

# ------------------------------------------------------------
# TARGETS
# ------------------------------------------------------------

y_protein_train = train_meta["log10_value"].copy()
y_protein_val = val_meta["log10_value"].copy()
y_protein_test = test_meta["log10_value"].copy()

# ------------------------------------------------------------
# VERIFY SHAPES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE MATRIX RESULTS")
print("=" * 70)

print(
    "X_protein_train:",
    X_protein_train.shape
)

print(
    "X_protein_val  :",
    X_protein_val.shape
)

print(
    "X_protein_test :",
    X_protein_test.shape
)

print(
    "y_protein_train:",
    y_protein_train.shape
)

print(
    "y_protein_val  :",
    y_protein_val.shape
)

print(
    "y_protein_test :",
    y_protein_test.shape
)

assert X_protein_train.shape == (
    len(train_meta), 38
)

assert X_protein_val.shape == (
    len(val_meta), 38
)

assert X_protein_test.shape == (
    len(test_meta), 38
)

assert list(X_protein_train.columns) == FEATURES
assert list(X_protein_val.columns) == FEATURES
assert list(X_protein_test.columns) == FEATURES

print("\n✓ All matrices contain exactly 38 features")
print("✓ Feature order is identical across all splits")

# ------------------------------------------------------------
# MISSING VALUE AUDIT
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUE AUDIT")
print("=" * 70)

print("\nTrain:")
print(
    X_protein_train.isna().sum()[
        X_protein_train.isna().sum() > 0
    ]
)

print("\nValidation:")
print(
    X_protein_val.isna().sum()[
        X_protein_val.isna().sum() > 0
    ]
)

print("\nTest:")
print(
    X_protein_test.isna().sum()[
        X_protein_test.isna().sum() > 0
    ]
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

X_protein_train.to_pickle(
    os.path.join(
        PROTEIN_DIR,
        "protein_X_train.pkl"
    )
)

X_protein_val.to_pickle(
    os.path.join(
        PROTEIN_DIR,
        "protein_X_val.pkl"
    )
)

X_protein_test.to_pickle(
    os.path.join(
        PROTEIN_DIR,
        "protein_X_test.pkl"
    )
)

y_protein_train.to_pickle(
    os.path.join(
        PROTEIN_DIR,
        "protein_y_train.pkl"
    )
)

y_protein_val.to_pickle(
    os.path.join(
        PROTEIN_DIR,
        "protein_y_val.pkl"
    )
)

y_protein_test.to_pickle(
    os.path.join(
        PROTEIN_DIR,
        "protein_y_test.pkl"
    )
)

pd.to_pickle(
    FEATURES,
    os.path.join(
        PROTEIN_DIR,
        "protein_feature_names.pkl"
    )
)

print("\n" + "=" * 70)
print("FEATURE RECONSTRUCTION COMPLETE")
print("=" * 70)

print("\nSaved to:")
print(PROTEIN_DIR)

print("\n✓ protein_X_train.pkl")
print("✓ protein_X_val.pkl")
print("✓ protein_X_test.pkl")
print("✓ protein_y_train.pkl")
print("✓ protein_y_val.pkl")
print("✓ protein_y_test.pkl")
print("✓ protein_feature_names.pkl")

print("\nDO NOT TRAIN THE MODEL YET.")
```

```python
# ============================================================
# CATPRED — PROTEIN-DISJOINT TUNED EXTRATREES
# TRAIN + VALIDATION
# ============================================================

import os
import numpy as np
import pandas as pd

from sklearn.impute import SimpleImputer
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

print("=" * 70)
print("CATPRED — PROTEIN-DISJOINT TUNED EXTRATREES")
print("=" * 70)

BASE_DIR = "/content/drive/MyDrive/CatPred_kcat_Project"
PROTEIN_DIR = os.path.join(
    BASE_DIR,
    "protein_level_evaluation"
)

# ------------------------------------------------------------
# LOAD PROTEIN-DISJOINT DATA
# ------------------------------------------------------------

X_train = pd.read_pickle(
    os.path.join(PROTEIN_DIR, "protein_X_train.pkl")
)

X_val = pd.read_pickle(
    os.path.join(PROTEIN_DIR, "protein_X_val.pkl")
)

X_test = pd.read_pickle(
    os.path.join(PROTEIN_DIR, "protein_X_test.pkl")
)

y_train = pd.read_pickle(
    os.path.join(PROTEIN_DIR, "protein_y_train.pkl")
)

y_val = pd.read_pickle(
    os.path.join(PROTEIN_DIR, "protein_y_val.pkl")
)

y_test = pd.read_pickle(
    os.path.join(PROTEIN_DIR, "protein_y_test.pkl")
)

print("\nData:")
print("X_train:", X_train.shape)
print("X_val  :", X_val.shape)
print("X_test :", X_test.shape)

# ------------------------------------------------------------
# VERIFY FEATURE ORDER
# ------------------------------------------------------------

assert list(X_train.columns) == list(X_val.columns)
assert list(X_train.columns) == list(X_test.columns)

print("\n✓ All splits have identical 38 features")

# ------------------------------------------------------------
# MISSING VALUE CHECK
# ------------------------------------------------------------

print("\nMissing values before imputation:")

print(
    "Train:",
    X_train.isna().sum().sum()
)

print(
    "Validation:",
    X_val.isna().sum().sum()
)

print(
    "Test:",
    X_test.isna().sum().sum()
)

# ------------------------------------------------------------
# MEDIAN IMPUTATION
#
# IMPORTANT:
# The imputer is FIT ONLY on the TRAINING DATA.
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FITTING TRAIN-ONLY MEDIAN IMPUTER")
print("=" * 70)

imputer = SimpleImputer(
    strategy="median"
)

X_train_imp = imputer.fit_transform(
    X_train
)

X_val_imp = imputer.transform(
    X_val
)

X_test_imp = imputer.transform(
    X_test
)

print("\n✓ Imputer fitted only on protein-disjoint training data")
print("✓ Same imputation values applied to validation/test")

# ------------------------------------------------------------
# VERIFY NO MISSING VALUES REMAIN
# ------------------------------------------------------------

print("\nMissing values after imputation:")

print(
    "Train:",
    np.isnan(X_train_imp).sum()
)

print(
    "Validation:",
    np.isnan(X_val_imp).sum()
)

print(
    "Test:",
    np.isnan(X_test_imp).sum()
)

assert np.isnan(X_train_imp).sum() == 0
assert np.isnan(X_val_imp).sum() == 0
assert np.isnan(X_test_imp).sum() == 0

# ------------------------------------------------------------
# TUNED EXTRATREES
# SAME CONFIGURATION AS CORRECTED MODEL
# ------------------------------------------------------------

model = ExtraTreesRegressor(
    n_estimators=200,
    max_features=0.7,
    min_samples_leaf=2,
    min_samples_split=2,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

print("\n" + "=" * 70)
print("TRAINING PROTEIN-DISJOINT TUNED EXTRATREES")
print("=" * 70)

print("\nModel:")
print(model)

print("\nTraining:")
print("X_train:", X_train_imp.shape)
print("y_train:", y_train.shape)

model.fit(
    X_train_imp,
    y_train
)

print("\n✓ Model trained")

# ------------------------------------------------------------
# VALIDATION PREDICTIONS
# ------------------------------------------------------------

val_predictions = model.predict(
    X_val_imp
)

# ------------------------------------------------------------
# VALIDATION METRICS
# ------------------------------------------------------------

val_r2 = r2_score(
    y_val,
    val_predictions
)

val_rmse = np.sqrt(
    mean_squared_error(
        y_val,
        val_predictions
    )
)

val_mae = mean_absolute_error(
    y_val,
    val_predictions
)

val_pearson_r, val_pearson_p = pearsonr(
    y_val,
    val_predictions
)

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("PROTEIN-DISJOINT VALIDATION PERFORMANCE")
print("=" * 70)

print(f"R²          : {val_r2:.4f}")
print(f"RMSE        : {val_rmse:.4f}")
print(f"MAE         : {val_mae:.4f}")
print(f"Pearson r   : {val_pearson_r:.4f}")

if val_pearson_p == 0:
    print(
        "Pearson p   : < 1e-300 "
        "(floating-point underflow)"
    )
else:
    print(
        f"Pearson p   : {val_pearson_p:.4e}"
    )

# ------------------------------------------------------------
# SAVE VALIDATION RESULTS
# ------------------------------------------------------------

validation_results = pd.DataFrame({
    "Model": ["Protein-disjoint ExtraTrees"],
    "R2": [val_r2],
    "RMSE": [val_rmse],
    "MAE": [val_mae],
    "Pearson_r": [val_pearson_r],
    "Pearson_p": [val_pearson_p]
})

validation_results.to_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_validation_results.csv"
    ),
    index=False
)

# ------------------------------------------------------------
# SAVE MODEL + IMPUTER
# ------------------------------------------------------------

import joblib

joblib.dump(
    model,
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_extratrees_validation_model.pkl"
    )
)

joblib.dump(
    imputer,
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_imputer.pkl"
    )
)

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_validation_results.csv"
    )
)

print(
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_extratrees_validation_model.pkl"
    )
)

print(
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_imputer.pkl"
    )
)

print("\n⚠️ DO NOT EVALUATE THE PROTEIN-DISJOINT TEST SET YET.")
print("Send me the validation results first.")
```

```python
# ============================================================
# CATPRED — PROTEIN-DISJOINT FEATURE GROUP ABLATION
# VALIDATION ONLY
# ============================================================

import os
import joblib
import pickle
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

# ------------------------------------------------------------
# LOAD PROTEIN-DISJOINT DATA
# ------------------------------------------------------------

with open(f"{BASE}/protein_X_train.pkl", "rb") as f:
    X_train = pickle.load(f)

with open(f"{BASE}/protein_X_val.pkl", "rb") as f:
    X_val = pickle.load(f)

with open(f"{BASE}/protein_y_train.pkl", "rb") as f:
    y_train = pickle.load(f)

with open(f"{BASE}/protein_y_val.pkl", "rb") as f:
    y_val = pickle.load(f)

with open(f"{BASE}/protein_feature_names.pkl", "rb") as f:
    feature_names = pickle.load(f)

print("=" * 70)
print("PROTEIN-DISJOINT FEATURE GROUP ABLATION")
print("=" * 70)

print("\nTraining:", X_train.shape)
print("Validation:", X_val.shape)

# ------------------------------------------------------------
# FEATURE GROUPS
# ------------------------------------------------------------

sequence_features = [
    'seq_length',
    'aa_A', 'aa_C', 'aa_D', 'aa_E', 'aa_F', 'aa_G',
    'aa_H', 'aa_I', 'aa_K', 'aa_L', 'aa_M', 'aa_N',
    'aa_P', 'aa_Q', 'aa_R', 'aa_S', 'aa_T', 'aa_V',
    'aa_W', 'aa_Y',
    'hydrophobic_fraction',
    'charged_fraction',
    'positive_fraction',
    'negative_fraction',
    'aromatic_fraction',
    'polar_fraction',
    'small_residue_fraction'
]

substrate_features = [
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms'
]

context_features = [
    'temperature',
    'ph'
]

all_features = (
    sequence_features +
    substrate_features +
    context_features
)

assert len(all_features) == 38
assert set(all_features) == set(feature_names)

# ------------------------------------------------------------
# MODEL FUNCTION
# ------------------------------------------------------------

def evaluate_group(name, features):

    print("\n" + "=" * 60)
    print(f"Training: {name}")
    print("=" * 60)

    print("Features:", len(features))

    # Select features
    Xtr = X_train[features].copy()
    Xv = X_val[features].copy()

    # TRAIN-ONLY IMPUTATION
    imputer = SimpleImputer(strategy="median")

    Xtr_imp = imputer.fit_transform(Xtr)
    Xv_imp = imputer.transform(Xv)

    # SAME TUNED MODEL
    model = ExtraTreesRegressor(
        n_estimators=200,
        max_features=0.7,
        min_samples_split=2,
        min_samples_leaf=2,
        max_depth=None,
        n_jobs=-1,
        random_state=42
    )

    model.fit(Xtr_imp, y_train)

    # Validation prediction
    pred = model.predict(Xv_imp)

    # Metrics
    r2 = r2_score(y_val, pred)

    rmse = np.sqrt(
        mean_squared_error(y_val, pred)
    )

    mae = mean_absolute_error(
        y_val, pred
    )

    r, p = pearsonr(
        y_val,
        pred
    )

    print(f"R²          : {r2:.4f}")
    print(f"RMSE        : {rmse:.4f}")
    print(f"MAE         : {mae:.4f}")
    print(f"Pearson r   : {r:.4f}")

    if p == 0:
        print("Pearson p   : < 1e-300")
    else:
        print(f"Pearson p   : {p:.4e}")

    return {
        "Model": name,
        "Num_Features": len(features),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    }


# ------------------------------------------------------------
# RUN ABLATION
# ------------------------------------------------------------

results = []

# Full
results.append(
    evaluate_group(
        "Full model",
        all_features
    )
)

# Without sequence
results.append(
    evaluate_group(
        "Without Sequence",
        substrate_features + context_features
    )
)

# Without substrate
results.append(
    evaluate_group(
        "Without Substrate",
        sequence_features + context_features
    )
)

# Without context
results.append(
    evaluate_group(
        "Without Context",
        sequence_features + substrate_features
    )
)

# ------------------------------------------------------------
# RESULTS TABLE
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("PROTEIN-DISJOINT VALIDATION ABLATION RESULTS")
print("=" * 70)

print(
    results_df.to_string(index=False)
)

# ------------------------------------------------------------
# CHANGE RELATIVE TO FULL MODEL
# ------------------------------------------------------------

full_r2 = results_df.loc[
    results_df["Model"] == "Full model",
    "R2"
].iloc[0]

full_rmse = results_df.loc[
    results_df["Model"] == "Full model",
    "RMSE"
].iloc[0]

results_df["Delta_R2"] = (
    results_df["R2"] - full_r2
)

results_df["Delta_RMSE"] = (
    results_df["RMSE"] - full_rmse
)

print("\n" + "=" * 70)
print("PERFORMANCE CHANGE RELATIVE TO FULL MODEL")
print("=" * 70)

print(
    results_df[
        ["Model", "Delta_R2", "Delta_RMSE"]
    ].to_string(index=False)
)

# ------------------------------------------------------------
# SAVE
# ------------------------------------------------------------

OUTPUT = os.path.join(
    BASE,
    "protein_disjoint_feature_group_ablation_validation.csv"
)

results_df.to_csv(
    OUTPUT,
    index=False
)

print("\n✓ Saved:")
print(OUTPUT)
```

```python
# ============================================================
# CATPRED — PROTEIN-DISJOINT 10-FEATURE MODEL TUNING
# TRAIN + VALIDATION ONLY
# ============================================================

import os
import pickle
import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor, RandomForestRegressor, RandomForestRegressor, HistGradientBoostingRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

print("=" * 70)
print("CATPRED — PROTEIN-DISJOINT 10-FEATURE MODEL TUNING")
print("=" * 70)

# ------------------------------------------------------------
# LOAD TRAINING + VALIDATION ONLY
# ------------------------------------------------------------

with open(f"{BASE}/protein_X_train.pkl", "rb") as f:
    X_train = pickle.load(f)

with open(f"{BASE}/protein_X_val.pkl", "rb") as f:
    X_val = pickle.load(f)

with open(f"{BASE}/protein_y_train.pkl", "rb") as f:
    y_train = pickle.load(f)

with open(f"{BASE}/protein_y_val.pkl", "rb") as f:
    y_val = pickle.load(f)

print("\nTraining:", X_train.shape)
print("Validation:", X_val.shape)

# ------------------------------------------------------------
# 10 FEATURES
# ------------------------------------------------------------

selected_features = [
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms',
    'temperature',
    'ph'
]

X_train_10 = X_train[selected_features].copy()
X_val_10 = X_val[selected_features].copy()

print("\nSelected features:")
for f in selected_features:
    print(" -", f)

print("\nX_train_10:", X_train_10.shape)
print("X_val_10  :", X_val_10.shape)

# ------------------------------------------------------------
# TRAIN-ONLY MEDIAN IMPUTATION
# ------------------------------------------------------------

imputer = SimpleImputer(strategy="median")

X_train_imp = imputer.fit_transform(X_train_10)
X_val_imp = imputer.transform(X_val_10)

print("\n✓ Imputer fitted ONLY on training data")
print("Train missing after imputation:",
      np.isnan(X_train_imp).sum())
print("Validation missing after imputation:",
      np.isnan(X_val_imp).sum())

# ------------------------------------------------------------
# MODELS
# ------------------------------------------------------------

models = {

    "ExtraTrees_200": ExtraTreesRegressor(
        n_estimators=200,
        max_features=0.7,
        min_samples_leaf=2,
        min_samples_split=2,
        max_depth=None,
        n_jobs=-1,
        random_state=42
    ),

    "ExtraTrees_500": ExtraTreesRegressor(
        n_estimators=500,
        max_features=0.7,
        min_samples_leaf=2,
        min_samples_split=2,
        max_depth=None,
        n_jobs=-1,
        random_state=42
    ),

    "ExtraTrees_leaf4": ExtraTreesRegressor(
        n_estimators=300,
        max_features=0.7,
        min_samples_leaf=4,
        min_samples_split=2,
        max_depth=None,
        n_jobs=-1,
        random_state=42
    ),

    "ExtraTrees_leaf8": ExtraTreesRegressor(
        n_estimators=300,
        max_features=0.7,
        min_samples_leaf=8,
        min_samples_split=2,
        max_depth=None,
        n_jobs=-1,
        random_state=42
    ),

    "RandomForest": RandomForestRegressor(
        n_estimators=300,
        max_features=0.7,
        min_samples_leaf=2,
        min_samples_split=2,
        max_depth=None,
        n_jobs=-1,
        random_state=42
    ),

    "HistGradientBoosting": HistGradientBoostingRegressor(
        max_iter=300,
        learning_rate=0.05,
        max_leaf_nodes=15,
        l2_regularization=1.0,
        random_state=42
    )
}

# ------------------------------------------------------------
# TRAIN + VALIDATE
# ------------------------------------------------------------

results = {}
predictions = {}

for name, model in models.items():

    print("\n" + "=" * 60)
    print(f"TRAINING: {name}")
    print("=" * 60)

    model.fit(X_train_imp, y_train)

    pred = model.predict(X_val_imp)

    r2 = r2_score(y_val, pred)

    rmse = np.sqrt(
        mean_squared_error(y_val, pred)
    )

    mae = mean_absolute_error(
        y_val, pred
    )

    r, p = pearsonr(
        y_val,
        pred
    )

    results[name] = {
        "Model": name,
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    }

    predictions[name] = pred

    print(f"R²          : {r2:.4f}")
    print(f"RMSE        : {rmse:.4f}")
    print(f"MAE         : {mae:.4f}")
    print(f"Pearson r   : {r:.4f}")

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

results_df = pd.DataFrame(
    list(results.values())
).sort_values(
    "R2",
    ascending=False
)

print("\n" + "=" * 70)
print("PROTEIN-DISJOINT 10-FEATURE MODEL COMPARISON")
print("=" * 70)

print(results_df.to_string(index=False))

# ------------------------------------------------------------
# SELECT BEST BY VALIDATION R²
# ------------------------------------------------------------

best_name = results_df.iloc[0]["Model"]

best_model = models[best_name]
best_pred = predictions[best_name]

print("\n" + "=" * 70)
print("BEST VALIDATION MODEL")
print("=" * 70)

print("Model:", best_name)
print(f"R²  : {results_df.iloc[0]['R2']:.4f}")
print(f"RMSE: {results_df.iloc[0]['RMSE']:.4f}")
print(f"MAE : {results_df.iloc[0]['MAE']:.4f}")
print(f"r   : {results_df.iloc[0]['Pearson_r']:.4f}")

# ------------------------------------------------------------
# SAVE RESULTS
# ------------------------------------------------------------

RESULT_PATH = os.path.join(
    BASE,
    "protein_disjoint_10feature_model_comparison.csv"
)

results_df.to_csv(
    RESULT_PATH,
    index=False
)

print("\n✓ Results saved:")
print(RESULT_PATH)

# ------------------------------------------------------------
# SAVE BEST VALIDATION MODEL
# ------------------------------------------------------------

MODEL_PATH = os.path.join(
    BASE,
    "protein_disjoint_best_10feature_validation_model.pkl"
)

IMPUTER_PATH = os.path.join(
    BASE,
    "protein_disjoint_10feature_imputer.pkl"
)

FEATURE_PATH = os.path.join(
    BASE,
    "protein_disjoint_10feature_names.pkl"
)

joblib.dump(
    best_model,
    MODEL_PATH
)

joblib.dump(
    imputer,
    IMPUTER_PATH
)

with open(FEATURE_PATH, "wb") as f:
    pickle.dump(selected_features, f)

print("\n✓ Best model saved:")
print(MODEL_PATH)

print("\n✓ Imputer saved:")
print(IMPUTER_PATH)

print("\n✓ Feature list saved:")
print(FEATURE_PATH)

print("\n" + "=" * 70)
print("IMPORTANT")
print("=" * 70)
print("The 3,985-row protein-disjoint test set was NOT used.")
print("Model selection was performed using validation R² only.")
```

```python
# ============================================================
# CATPRED — PROTEIN-DISJOINT 10-FEATURE EXTRATREES SEARCH
# VALIDATION-ONLY MODEL SELECTION
# ============================================================

import os
import pickle
import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.model_selection import ParameterGrid
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

# ------------------------------------------------------------
# LOAD TRAIN + VALIDATION ONLY
# ------------------------------------------------------------

with open(f"{BASE}/protein_X_train.pkl", "rb") as f:
    X_train = pickle.load(f)

with open(f"{BASE}/protein_X_val.pkl", "rb") as f:
    X_val = pickle.load(f)

with open(f"{BASE}/protein_y_train.pkl", "rb") as f:
    y_train = pickle.load(f)

with open(f"{BASE}/protein_y_val.pkl", "rb") as f:
    y_val = pickle.load(f)

# ------------------------------------------------------------
# 10 FEATURES
# ------------------------------------------------------------

selected_features = [
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms',
    'temperature',
    'ph'
]

X_train_10 = X_train[selected_features].copy()
X_val_10 = X_val[selected_features].copy()

# ------------------------------------------------------------
# TRAIN-ONLY IMPUTATION
# ------------------------------------------------------------

imputer = SimpleImputer(strategy="median")

X_train_imp = imputer.fit_transform(X_train_10)
X_val_imp = imputer.transform(X_val_10)

print("=" * 70)
print("PROTEIN-DISJOINT 10-FEATURE EXTRATREES HYPERPARAMETER SEARCH")
print("=" * 70)

print("\nTraining:", X_train_imp.shape)
print("Validation:", X_val_imp.shape)

# ------------------------------------------------------------
# SEARCH SPACE
# ------------------------------------------------------------

param_grid = {
    "n_estimators": [200, 400],
    "min_samples_leaf": [2, 4, 6, 8, 12],
    "min_samples_split": [2, 5],
    "max_features": [0.5, 0.7, 1.0],
    "max_depth": [None, 15]
}

grid = list(ParameterGrid(param_grid))

print("\nTotal configurations:", len(grid))

# ------------------------------------------------------------
# SEARCH
# ------------------------------------------------------------

results = []

for i, params in enumerate(grid, 1):

    model = ExtraTreesRegressor(
        n_estimators=params["n_estimators"],
        min_samples_leaf=params["min_samples_leaf"],
        min_samples_split=params["min_samples_split"],
        max_features=params["max_features"],
        max_depth=params["max_depth"],
        n_jobs=-1,
        random_state=42
    )

    model.fit(X_train_imp, y_train)

    pred = model.predict(X_val_imp)

    r2 = r2_score(y_val, pred)

    rmse = np.sqrt(
        mean_squared_error(y_val, pred)
    )

    mae = mean_absolute_error(
        y_val, pred
    )

    r, p = pearsonr(
        y_val,
        pred
    )

    results.append({
        "n_estimators": params["n_estimators"],
        "min_samples_leaf": params["min_samples_leaf"],
        "min_samples_split": params["min_samples_split"],
        "max_features": params["max_features"],
        "max_depth": params["max_depth"],
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    })

    if i % 10 == 0 or i == len(grid):
        print(f"Completed {i}/{len(grid)}")

# ------------------------------------------------------------
# RESULTS
# ------------------------------------------------------------

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="R2",
    ascending=False
).reset_index(drop=True)

print("\n" + "=" * 70)
print("TOP 15 VALIDATION CONFIGURATIONS")
print("=" * 70)

print(
    results_df.head(15).to_string(index=False)
)

# ------------------------------------------------------------
# BEST MODEL
# ------------------------------------------------------------

best = results_df.iloc[0]

print("\n" + "=" * 70)
print("BEST PROTEIN-DISJOINT VALIDATION MODEL")
print("=" * 70)

print("n_estimators      :", best["n_estimators"])
print("min_samples_leaf  :", best["min_samples_leaf"])
print("min_samples_split :", best["min_samples_split"])
print("max_features      :", best["max_features"])
print("max_depth         :", best["max_depth"])

print(f"\nR²          : {best['R2']:.4f}")
print(f"RMSE        : {best['RMSE']:.4f}")
print(f"MAE         : {best['MAE']:.4f}")
print(f"Pearson r   : {best['Pearson_r']:.4f}")

# ------------------------------------------------------------
# SAVE SEARCH RESULTS
# ------------------------------------------------------------

RESULT_PATH = os.path.join(
    BASE,
    "protein_disjoint_10feature_extratrees_tuning.csv"
)

results_df.to_csv(
    RESULT_PATH,
    index=False
)

print("\n✓ Search results saved:")
print(RESULT_PATH)

# ------------------------------------------------------------
# SAVE BEST CONFIGURATION
# ------------------------------------------------------------

BEST_PARAMS_PATH = os.path.join(
    BASE,
    "protein_disjoint_best_10feature_params.pkl"
)

best_params = {
    "n_estimators": int(best["n_estimators"]),
    "min_samples_leaf": int(best["min_samples_leaf"]),
    "min_samples_split": int(best["min_samples_split"]),
    "max_features": best["max_features"],
    "max_depth": best["max_depth"]
}

with open(BEST_PARAMS_PATH, "wb") as f:
    pickle.dump(best_params, f)

print("✓ Best parameters saved:")
print(BEST_PARAMS_PATH)

print("\n" + "=" * 70)
print("IMPORTANT")
print("=" * 70)
print("✓ Model selection used TRAIN + VALIDATION only")
print("✓ 3,985-row protein-disjoint TEST remains untouched")
print("✓ Do NOT evaluate the final test yet")
```

```python
# ============================================================
# CATPRED — FINAL PROTEIN-DISJOINT HELD-OUT TEST EVALUATION
# LOCKED MODEL — DO NOT TUNE AFTER THIS
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

# ------------------------------------------------------------
# LOAD PROTEIN-DISJOINT DATA
# ------------------------------------------------------------

with open(f"{BASE}/protein_X_train.pkl", "rb") as f:
    X_train = pickle.load(f)

with open(f"{BASE}/protein_X_val.pkl", "rb") as f:
    X_val = pickle.load(f)

with open(f"{BASE}/protein_X_test.pkl", "rb") as f:
    X_test = pickle.load(f)

with open(f"{BASE}/protein_y_train.pkl", "rb") as f:
    y_train = pickle.load(f)

with open(f"{BASE}/protein_y_val.pkl", "rb") as f:
    y_val = pickle.load(f)

with open(f"{BASE}/protein_y_test.pkl", "rb") as f:
    y_test = pickle.load(f)

# ------------------------------------------------------------
# LOCKED 10-FEATURE SET
# ------------------------------------------------------------

selected_features = [
    'mol_weight',
    'logp',
    'tpsa',
    'hbd',
    'hba',
    'rotatable_bonds',
    'rings',
    'heavy_atoms',
    'temperature',
    'ph'
]

X_train_10 = X_train[selected_features].copy()
X_val_10   = X_val[selected_features].copy()
X_test_10  = X_test[selected_features].copy()

print("=" * 70)
print("CATPRED — FINAL PROTEIN-DISJOINT HELD-OUT TEST")
print("=" * 70)

print("\nData:")
print("Train:", X_train_10.shape)
print("Validation:", X_val_10.shape)
print("Test:", X_test_10.shape)

# ------------------------------------------------------------
# TRAIN-ONLY IMPUTER
# ------------------------------------------------------------

imputer = SimpleImputer(strategy="median")

X_train_imp = imputer.fit_transform(X_train_10)
X_val_imp   = imputer.transform(X_val_10)
X_test_imp  = imputer.transform(X_test_10)

print("\nMissing values after imputation:")
print("Train:", np.isnan(X_train_imp).sum())
print("Validation:", np.isnan(X_val_imp).sum())
print("Test:", np.isnan(X_test_imp).sum())

# ------------------------------------------------------------
# LOCKED MODEL
#
# Selected using TRAIN + VALIDATION ONLY
# ------------------------------------------------------------

model = ExtraTreesRegressor(
    n_estimators=200,
    min_samples_leaf=4,
    min_samples_split=2,
    max_features=1.0,
    max_depth=None,
    n_jobs=-1,
    random_state=42
)

print("\n" + "=" * 70)
print("TRAINING LOCKED MODEL")
print("=" * 70)

print("""
ExtraTreesRegressor(
    n_estimators=200,
    min_samples_leaf=4,
    min_samples_split=2,
    max_features=1.0,
    max_depth=None,
    random_state=42
)
""")

# IMPORTANT:
# The test set is NOT involved in fitting.
model.fit(X_train_imp, y_train)

print("✓ Model trained using training set only")

# ------------------------------------------------------------
# FINAL TEST PREDICTIONS
# ------------------------------------------------------------

test_predictions = model.predict(X_test_imp)

print("\nPredictions:", test_predictions.shape)

# ------------------------------------------------------------
# FINAL HELD-OUT TEST METRICS
# ------------------------------------------------------------

r2 = r2_score(y_test, test_predictions)

rmse = np.sqrt(
    mean_squared_error(y_test, test_predictions)
)

mae = mean_absolute_error(
    y_test,
    test_predictions
)

pearson_r, pearson_p = pearsonr(
    y_test,
    test_predictions
)

print("\n" + "=" * 70)
print("FINAL PROTEIN-DISJOINT HELD-OUT TEST PERFORMANCE")
print("=" * 70)

print(f"\nR²          : {r2:.4f}")
print(f"RMSE        : {rmse:.4f}")
print(f"MAE         : {mae:.4f}")
print(f"Pearson r   : {pearson_r:.4f}")

if pearson_p == 0:
    print("Pearson p   : < 1e-300 (floating-point underflow)")
else:
    print(f"Pearson p   : {pearson_p:.6e}")

# ------------------------------------------------------------
# TEST SET INTEGRITY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TEST SET INTEGRITY")
print("=" * 70)

print("Held-out test samples:", len(y_test))
print("Test used for training: NO")
print("Test used for imputer fitting: NO")
print("Test used for hyperparameter selection: NO")
print("Sequence-cluster overlap with training: ZERO")
print("Exact sequence overlap with training: ZERO")

# ------------------------------------------------------------
# SAVE FINAL PREDICTIONS
# ------------------------------------------------------------

prediction_df = pd.DataFrame({
    "Observed_log10_kcat": np.asarray(y_test),
    "Predicted_log10_kcat": test_predictions,
    "Residual": np.asarray(y_test) - test_predictions
})

prediction_path = os.path.join(
    BASE,
    "protein_disjoint_final_heldout_test_predictions.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

# ------------------------------------------------------------
# SAVE FINAL PERFORMANCE
# ------------------------------------------------------------

performance_df = pd.DataFrame([{
    "Model": "ExtraTrees_10Feature_ProteinDisjoint",
    "N_Test": len(y_test),
    "R2": r2,
    "RMSE": rmse,
    "MAE": mae,
    "Pearson_r": pearson_r,
    "Pearson_p": pearson_p,
    "n_estimators": 200,
    "min_samples_leaf": 4,
    "min_samples_split": 2,
    "max_features": 1.0,
    "max_depth": None
}])

performance_path = os.path.join(
    BASE,
    "protein_disjoint_final_heldout_test_performance.csv"
)

performance_df.to_csv(
    performance_path,
    index=False
)

# ------------------------------------------------------------
# SAVE FINAL MODEL + IMPUTER
# ------------------------------------------------------------

model_path = os.path.join(
    BASE,
    "protein_disjoint_final_extratrees_10feature.pkl"
)

imputer_path = os.path.join(
    BASE,
    "protein_disjoint_final_imputer.pkl"
)

with open(model_path, "wb") as f:
    pickle.dump(model, f)

with open(imputer_path, "wb") as f:
    pickle.dump(imputer, f)

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print(prediction_path)
print(performance_path)
print(model_path)
print(imputer_path)

print("\n" + "=" * 70)
print("FINAL PROTEIN-DISJOINT TEST EVALUATION COMPLETE")
print("=" * 70)
```

```python
import os
import pickle
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

# ============================================================
# PATHS
# ============================================================

MODEL_PATH = os.path.join(
    BASE, "protein_disjoint_final_extratrees_10feature.pkl"
)

FEATURE_NAMES_PATH = os.path.join(
    BASE, "protein_disjoint_10feature_names.pkl"
)

# ============================================================
# LOAD TRAIN / TEST MATRICES
# ============================================================

with open(os.path.join(BASE, "protein_X_train.pkl"), "rb") as f:
    X_train = pickle.load(f)

with open(os.path.join(BASE, "protein_X_test.pkl"), "rb") as f:
    X_test = pickle.load(f)

with open(os.path.join(BASE, "protein_y_test.pkl"), "rb") as f:
    y_test = pickle.load(f)

# Also load train target for completeness
with open(os.path.join(BASE, "protein_y_train.pkl"), "rb") as f:
    y_train = pickle.load(f)

print("=" * 70)
print("CATPRED — FINAL PROTEIN-DISJOINT TEST PREDICTIONS")
print("=" * 70)

print("\nOriginal matrices:")
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_test :", y_test.shape)


# ============================================================
# LOAD FINAL MODEL
# ============================================================

with open(MODEL_PATH, "rb") as f:
    model = pickle.load(f)

print("\n✓ Final model loaded")
print(type(model))


# ============================================================
# LOAD THE 10 FEATURE NAMES
# ============================================================

with open(FEATURE_NAMES_PATH, "rb") as f:
    feature_names = pickle.load(f)

feature_names = list(feature_names)

print("\n10 features used by final model:")
for i, feature in enumerate(feature_names, 1):
    print(f"{i:2d}. {feature}")

print("\nNumber of selected features:", len(feature_names))

if len(feature_names) != 10:
    raise ValueError(
        f"Expected exactly 10 features, but found {len(feature_names)}"
    )


# ============================================================
# CONVERT MATRICES TO DATAFRAMES IF NECESSARY
# ============================================================

# The original feature order is stored separately.
with open(os.path.join(BASE, "protein_feature_names.pkl"), "rb") as f:
    all_feature_names = pickle.load(f)

all_feature_names = list(all_feature_names)

print("\nTotal available features:", len(all_feature_names))

if len(all_feature_names) != X_train.shape[1]:
    raise ValueError(
        "Number of feature names does not match X_train columns."
    )

if len(all_feature_names) != X_test.shape[1]:
    raise ValueError(
        "Number of feature names does not match X_test columns."
    )

X_train_df = pd.DataFrame(
    X_train,
    columns=all_feature_names
)

X_test_df = pd.DataFrame(
    X_test,
    columns=all_feature_names
)


# ============================================================
# EXTRACT EXACT 10 FEATURES
# ============================================================

missing_train = [
    f for f in feature_names
    if f not in X_train_df.columns
]

missing_test = [
    f for f in feature_names
    if f not in X_test_df.columns
]

if missing_train or missing_test:
    raise ValueError(
        f"Missing features.\n"
        f"Train: {missing_train}\n"
        f"Test: {missing_test}"
    )

X_train_10 = X_train_df[feature_names].copy()
X_test_10 = X_test_df[feature_names].copy()

print("\n" + "=" * 70)
print("10-FEATURE MATRICES")
print("=" * 70)

print("X_train_10:", X_train_10.shape)
print("X_test_10 :", X_test_10.shape)


# ============================================================
# RECONSTRUCT TRAIN-ONLY IMPUTER
# ============================================================

print("\n" + "=" * 70)
print("RECONSTRUCTING TRAIN-ONLY MEDIAN IMPUTER")
print("=" * 70)

print("\nMissing values BEFORE imputation:")
print("Train:", X_train_10.isna().sum().sum())
print("Test :", X_test_10.isna().sum().sum())

# IMPORTANT:
# Fit ONLY on the protein-disjoint training data.
# The test set is never used to calculate medians.

imputer = SimpleImputer(strategy="median")

X_train_10_imp = imputer.fit_transform(X_train_10)

# Apply the TRAIN-FITTED imputer to test
X_test_10_imp = imputer.transform(X_test_10)

print("\n✓ Imputer reconstructed")
print("✓ Imputer fitted ONLY on training data")
print("✓ Test data was NOT used for fitting")

print("\nMissing values AFTER imputation:")
print("Train:", np.isnan(X_train_10_imp).sum())
print("Test :", np.isnan(X_test_10_imp).sum())


# ============================================================
# VERIFY MODEL FEATURE COUNT
# ============================================================

print("\n" + "=" * 70)
print("MODEL INPUT VERIFICATION")
print("=" * 70)

print("Model expects:", model.n_features_in_, "features")
print("Provided     :", X_test_10_imp.shape[1], "features")

if model.n_features_in_ != X_test_10_imp.shape[1]:
    raise ValueError(
        f"Feature mismatch: model expects "
        f"{model.n_features_in_}, but test has "
        f"{X_test_10_imp.shape[1]}"
    )

print("✓ Feature count matches")


# ============================================================
# FINAL HELD-OUT TEST PREDICTION
# ============================================================

print("\n" + "=" * 70)
print("GENERATING FINAL HELD-OUT TEST PREDICTIONS")
print("=" * 70)

y_pred_test = model.predict(X_test_10_imp)

print("Predictions:", y_pred_test.shape)


# ============================================================
# TEST PERFORMANCE
# ============================================================

r2 = r2_score(y_test, y_pred_test)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred_test)
)

mae = mean_absolute_error(
    y_test, y_pred_test
)

pearson_r, pearson_p = pearsonr(
    y_test, y_pred_test
)

print("\n" + "=" * 70)
print("FINAL PROTEIN-DISJOINT HELD-OUT TEST PERFORMANCE")
print("=" * 70)

print(f"\nR²          : {r2:.4f}")
print(f"RMSE        : {rmse:.4f}")
print(f"MAE         : {mae:.4f}")
print(f"Pearson r   : {pearson_r:.4f}")
print(f"Pearson p   : {pearson_p:.6e}")


# ============================================================
# SAVE TEST PREDICTIONS
# ============================================================

pred_df = pd.DataFrame({
    "y_true": np.asarray(y_test),
    "y_pred": y_pred_test
})

PRED_PATH = os.path.join(
    BASE,
    "protein_disjoint_test_predictions.csv"
)

pred_df.to_csv(PRED_PATH, index=False)

print("\n✓ Test predictions saved:")
print(PRED_PATH)


# ============================================================
# SAVE RECONSTRUCTED TRAIN-ONLY IMPUTER
# ============================================================

IMPUTER_FIXED_PATH = os.path.join(
    BASE,
    "protein_disjoint_10feature_imputer_reconstructed.pkl"
)

with open(IMPUTER_FIXED_PATH, "wb") as f:
    pickle.dump(imputer, f)

print("\n✓ Reconstructed train-only imputer saved:")
print(IMPUTER_FIXED_PATH)


# ============================================================
# FINAL INTEGRITY STATEMENT
# ============================================================

print("\n" + "=" * 70)
print("TEST SET INTEGRITY")
print("=" * 70)

print("Held-out test samples       :", len(y_test))
print("Test used for training      : NO")
print("Test used for imputer fit   : NO")
print("Test used for model tuning  : NO")
print("Sequence-cluster overlap    : ZERO")
print("Exact sequence overlap      : ZERO")
print("Feature count               : 10")
print("=" * 70)
```

```python
# ============================================================
# CATPRED — FINAL PROTEIN-DISJOINT TEST POST-HOC ANALYSIS
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd

from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

PRED_PATH = os.path.join(
    BASE,
    "protein_disjoint_test_predictions.csv"
)

MODEL_PATH = os.path.join(
    BASE,
    "protein_disjoint_final_extratrees_10feature.pkl"
)

FEATURE_NAMES_PATH = os.path.join(
    BASE,
    "protein_disjoint_10feature_names.pkl"
)

# ============================================================
# LOAD FINAL TEST PREDICTIONS
# ============================================================

df = pd.read_csv(PRED_PATH)

actual = df["y_true"].to_numpy(dtype=float)
predicted = df["y_pred"].to_numpy(dtype=float)

print("=" * 70)
print("CATPRED — FINAL PROTEIN-DISJOINT TEST POST-HOC ANALYSIS")
print("=" * 70)

print("\nTest samples:", len(df))

# ============================================================
# BASIC TEST METRICS
# ============================================================

r2 = r2_score(actual, predicted)

rmse = np.sqrt(
    mean_squared_error(actual, predicted)
)

mae = mean_absolute_error(
    actual, predicted
)

pearson_r, pearson_p = pearsonr(
    actual, predicted
)

print("\n" + "=" * 70)
print("OVERALL TEST PERFORMANCE")
print("=" * 70)

print(f"R²        : {r2:.4f}")
print(f"RMSE      : {rmse:.4f}")
print(f"MAE       : {mae:.4f}")
print(f"Pearson r : {pearson_r:.4f}")
print(f"p-value   : {pearson_p:.6e}")


# ============================================================
# MEAN-PREDICTION BASELINE
# ============================================================

baseline_pred = np.full_like(
    actual,
    np.mean(actual)
)

baseline_r2 = r2_score(
    actual,
    baseline_pred
)

baseline_rmse = np.sqrt(
    mean_squared_error(actual, baseline_pred)
)

baseline_mae = mean_absolute_error(
    actual,
    baseline_pred
)

print("\n" + "=" * 70)
print("MEAN-PREDICTION BASELINE")
print("=" * 70)

print(f"Baseline R²   : {baseline_r2:.4f}")
print(f"Baseline RMSE : {baseline_rmse:.4f}")
print(f"Baseline MAE  : {baseline_mae:.4f}")

print(
    f"\nRMSE improvement over baseline: "
    f"{100 * (baseline_rmse - rmse) / baseline_rmse:.2f}%"
)


# ============================================================
# TEST kcat TERTILES
# ============================================================

q1 = np.quantile(actual, 1/3)
q2 = np.quantile(actual, 2/3)

print("\n" + "=" * 70)
print("kcat TERTILE BOUNDARIES")
print("=" * 70)

print(f"Low / Medium  : {q1:.4f}")
print(f"Medium / High : {q2:.4f}")


# IMPORTANT:
# Use strings consistently to avoid np.select dtype errors.

groups = np.select(
    [
        actual <= q1,
        actual <= q2
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

df["kcat_range"] = groups


# ============================================================
# RANGE-WISE PERFORMANCE
# ============================================================

range_results = []

for group in ["Low", "Medium", "High"]:

    mask = df["kcat_range"].to_numpy() == group

    y_true_group = actual[mask]
    y_pred_group = predicted[mask]

    group_r2 = r2_score(
        y_true_group,
        y_pred_group
    )

    group_rmse = np.sqrt(
        mean_squared_error(
            y_true_group,
            y_pred_group
        )
    )

    group_mae = mean_absolute_error(
        y_true_group,
        y_pred_group
    )

    if len(y_true_group) > 1:
        group_r, group_p = pearsonr(
            y_true_group,
            y_pred_group
        )
    else:
        group_r = np.nan
        group_p = np.nan

    range_results.append({
        "kcat_range": group,
        "N": len(y_true_group),
        "Observed_mean": np.mean(y_true_group),
        "Predicted_mean": np.mean(y_pred_group),
        "Observed_std": np.std(y_true_group),
        "Predicted_std": np.std(y_pred_group),
        "R2": group_r2,
        "RMSE": group_rmse,
        "MAE": group_mae,
        "Pearson_r": group_r,
        "Pearson_p": group_p
    })


range_df = pd.DataFrame(range_results)

print("\n" + "=" * 70)
print("PERFORMANCE ACROSS kcat RANGES")
print("=" * 70)

print(
    range_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# PREDICTION BIAS / CALIBRATION BY RANGE
# ============================================================

range_df["Mean_Bias"] = (
    range_df["Predicted_mean"]
    - range_df["Observed_mean"]
)

range_df["Bias_%"] = (
    100 *
    range_df["Mean_Bias"]
    /
    range_df["Observed_mean"].replace(0, np.nan)
)

print("\n" + "=" * 70)
print("PREDICTION BIAS BY kcat RANGE")
print("=" * 70)

print(
    range_df[
        [
            "kcat_range",
            "N",
            "Observed_mean",
            "Predicted_mean",
            "Mean_Bias",
            "Bias_%"
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# PREDICTION COMPRESSION
# ============================================================

observed_std = np.std(actual)
predicted_std = np.std(predicted)

print("\n" + "=" * 70)
print("PREDICTION DISPERSION")
print("=" * 70)

print(f"Observed SD  : {observed_std:.4f}")
print(f"Predicted SD : {predicted_std:.4f}")

print(
    f"Prediction/Observed SD ratio: "
    f"{predicted_std / observed_std:.4f}"
)

print(
    "\nInterpretation:"
    "\nA ratio substantially below 1 indicates that predictions "
    "are compressed toward the central range."
)


# ============================================================
# RESIDUAL ANALYSIS
# ============================================================

residuals = actual - predicted

df["residual"] = residuals
df["absolute_error"] = np.abs(residuals)

print("\n" + "=" * 70)
print("RESIDUAL SUMMARY")
print("=" * 70)

print(f"Mean residual       : {np.mean(residuals):.4f}")
print(f"Median residual     : {np.median(residuals):.4f}")
print(f"Residual SD         : {np.std(residuals):.4f}")
print(f"Mean absolute error : {np.mean(np.abs(residuals)):.4f}")


# ============================================================
# FINAL MODEL FEATURE IMPORTANCE
# ============================================================

print("\n" + "=" * 70)
print("FINAL MODEL FEATURE IMPORTANCE")
print("=" * 70)

with open(MODEL_PATH, "rb") as f:
    model = pickle.load(f)

with open(FEATURE_NAMES_PATH, "rb") as f:
    feature_names = list(pickle.load(f))

if len(feature_names) != len(model.feature_importances_):
    raise ValueError(
        "Feature-name count does not match model feature importance count."
    )

importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
).reset_index(drop=True)

importance_df["Importance_%"] = (
    100 * importance_df["Importance"]
)

print(
    importance_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)


# ============================================================
# SAVE RESULTS
# ============================================================

range_path = os.path.join(
    BASE,
    "protein_disjoint_test_kcat_range_performance.csv"
)

importance_path = os.path.join(
    BASE,
    "protein_disjoint_final_feature_importance.csv"
)

pred_analysis_path = os.path.join(
    BASE,
    "protein_disjoint_test_predictions_with_residuals.csv"
)

range_df.to_csv(
    range_path,
    index=False
)

importance_df.to_csv(
    importance_path,
    index=False
)

df.to_csv(
    pred_analysis_path,
    index=False
)


# ============================================================
# FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("POST-HOC ANALYSIS SAVED")
print("=" * 70)

print("\n✓", range_path)
print("✓", importance_path)
print("✓", pred_analysis_path)

print("\n" + "=" * 70)
print("TEST SET REMAINS LOCKED")
print("=" * 70)

print("No retraining performed")
print("No hyperparameter tuning performed")
print("No imputer refitting performed")
print("No model selection performed")
print("Analysis is post-hoc only")
```

```python
# ============================================================
# CATPRED — 5-FOLD PROTEIN-DISJOINT CROSS-VALIDATION
# DEVELOPMENT DATA ONLY
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupKFold
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

# ============================================================
# PATHS
# ============================================================

BASE = "/content/drive/MyDrive/CatPred_kcat_Project"
PROTEIN_DIR = os.path.join(BASE, "protein_level_evaluation")

# ============================================================
# LOAD PROTEIN-DISJOINT DEVELOPMENT DATA
# ============================================================

with open(
    os.path.join(PROTEIN_DIR, "protein_X_train.pkl"), "rb"
) as f:
    X_train = pickle.load(f)

with open(
    os.path.join(PROTEIN_DIR, "protein_y_train.pkl"), "rb"
) as f:
    y_train = pickle.load(f)

with open(
    os.path.join(PROTEIN_DIR, "protein_X_val.pkl"), "rb"
) as f:
    X_val = pickle.load(f)

with open(
    os.path.join(PROTEIN_DIR, "protein_y_val.pkl"), "rb"
) as f:
    y_val = pickle.load(f)

# ============================================================
# LOAD DEVELOPMENT METADATA
# ============================================================

train_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_train_metadata.csv"
    )
)

val_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_validation_metadata.csv"
    )
)

# ============================================================
# COMBINE TRAIN + VALIDATION ONLY
# ============================================================

X_train_df = pd.DataFrame(X_train)
X_val_df = pd.DataFrame(X_val)

# Recover the original 38 feature names
with open(
    os.path.join(PROTEIN_DIR, "protein_feature_names.pkl"), "rb"
) as f:
    all_feature_names = list(pickle.load(f))

X_train_df.columns = all_feature_names
X_val_df.columns = all_feature_names

X_dev = pd.concat(
    [X_train_df, X_val_df],
    axis=0,
    ignore_index=True
)

y_dev = np.concatenate([
    np.asarray(y_train),
    np.asarray(y_val)
])

metadata_dev = pd.concat(
    [train_meta, val_meta],
    axis=0,
    ignore_index=True
)

# ============================================================
# VERIFY DEVELOPMENT DATA
# ============================================================

print("=" * 70)
print("CATPRED — PROTEIN-DISJOINT 5-FOLD CROSS-VALIDATION")
print("=" * 70)

print("\nDevelopment data:")
print("X_dev:", X_dev.shape)
print("y_dev:", y_dev.shape)
print("Metadata:", metadata_dev.shape)

assert len(X_dev) == len(y_dev)
assert len(X_dev) == len(metadata_dev)

print("\n✓ Development set = original train + validation")
print("✓ Held-out 3,985-row test is NOT included")

# ============================================================
# LOAD EXACT 10 FEATURES
# ============================================================

with open(
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_10feature_names.pkl"
    ), "rb"
) as f:
    selected_features = list(pickle.load(f))

print("\nSelected 10 features:")

for i, feature in enumerate(selected_features, 1):
    print(f"{i:2d}. {feature}")

assert len(selected_features) == 10

X_dev_10 = X_dev[selected_features].copy()

print("\nX_dev_10:", X_dev_10.shape)

# ============================================================
# GROUP VARIABLE
# ============================================================

GROUP_COLUMN = "sequence_20cluster"

if GROUP_COLUMN not in metadata_dev.columns:
    raise ValueError(
        f"{GROUP_COLUMN} not found in metadata columns:\n"
        f"{metadata_dev.columns.tolist()}"
    )

groups = metadata_dev[GROUP_COLUMN].to_numpy()

print("\nGrouping variable:", GROUP_COLUMN)
print("Unique protein clusters:", len(np.unique(groups)))

# ============================================================
# VERIFY NO MISSING GROUPS
# ============================================================

if pd.isna(groups).any():
    raise ValueError(
        "Missing sequence clusters detected."
    )

print("Missing clusters:", pd.isna(groups).sum())

# ============================================================
# GROUP K-FOLD
# ============================================================

N_SPLITS = 5

gkf = GroupKFold(
    n_splits=N_SPLITS
)

# ============================================================
# MODEL
# ============================================================

MODEL_PARAMS = {
    "n_estimators": 200,
    "min_samples_leaf": 4,
    "min_samples_split": 2,
    "max_features": 1.0,
    "n_jobs": -1,
    "random_state": 42
}

# ============================================================
# CROSS-VALIDATION
# ============================================================

results = []

print("\n" + "=" * 70)
print("STARTING 5-FOLD PROTEIN-DISJOINT CV")
print("=" * 70)

for fold, (train_idx, val_idx) in enumerate(
    gkf.split(
        X_dev_10,
        y_dev,
        groups=groups
    ),
    start=1
):

    print("\n" + "-" * 70)
    print(f"FOLD {fold}")
    print("-" * 70)

    X_fold_train = X_dev_10.iloc[train_idx]
    X_fold_val = X_dev_10.iloc[val_idx]

    y_fold_train = y_dev[train_idx]
    y_fold_val = y_dev[val_idx]

    groups_train = groups[train_idx]
    groups_val = groups[val_idx]

    # --------------------------------------------------------
    # VERIFY ZERO CLUSTER OVERLAP
    # --------------------------------------------------------

    overlap = set(groups_train).intersection(
        set(groups_val)
    )

    print("Training samples:", len(train_idx))
    print("Validation samples:", len(val_idx))
    print("Training clusters:", len(np.unique(groups_train)))
    print("Validation clusters:", len(np.unique(groups_val)))
    print("Cluster overlap:", len(overlap))

    assert len(overlap) == 0

    # --------------------------------------------------------
    # TRAIN-ONLY IMPUTER
    # --------------------------------------------------------

    imputer = SimpleImputer(
        strategy="median"
    )

    X_fold_train_imp = imputer.fit_transform(
        X_fold_train
    )

    X_fold_val_imp = imputer.transform(
        X_fold_val
    )

    # --------------------------------------------------------
    # TRAIN MODEL
    # --------------------------------------------------------

    model = ExtraTreesRegressor(
        **MODEL_PARAMS
    )

    model.fit(
        X_fold_train_imp,
        y_fold_train
    )

    # --------------------------------------------------------
    # PREDICT VALIDATION FOLD
    # --------------------------------------------------------

    y_pred = model.predict(
        X_fold_val_imp
    )

    # --------------------------------------------------------
    # METRICS
    # --------------------------------------------------------

    r2 = r2_score(
        y_fold_val,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_fold_val,
            y_pred
        )
    )

    mae = mean_absolute_error(
        y_fold_val,
        y_pred
    )

    pearson_r, pearson_p = pearsonr(
        y_fold_val,
        y_pred
    )

    results.append({
        "Fold": fold,
        "N_train": len(train_idx),
        "N_validation": len(val_idx),
        "N_train_clusters": len(
            np.unique(groups_train)
        ),
        "N_validation_clusters": len(
            np.unique(groups_val)
        ),
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": pearson_r,
        "Pearson_p": pearson_p
    })

    print("\nFold performance:")
    print(f"R²          : {r2:.4f}")
    print(f"RMSE        : {rmse:.4f}")
    print(f"MAE         : {mae:.4f}")
    print(f"Pearson r   : {pearson_r:.4f}")
    print(f"Pearson p   : {pearson_p:.4e}")

# ============================================================
# RESULTS TABLE
# ============================================================

cv_df = pd.DataFrame(results)

print("\n" + "=" * 70)
print("PROTEIN-DISJOINT 5-FOLD CV RESULTS")
print("=" * 70)

print(
    cv_df[
        [
            "Fold",
            "N_train",
            "N_validation",
            "R2",
            "RMSE",
            "MAE",
            "Pearson_r",
            "Pearson_p"
        ]
    ].to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ============================================================
# MEAN ± SD
# ============================================================

summary = pd.DataFrame({
    "Metric": [
        "R2",
        "RMSE",
        "MAE",
        "Pearson_r"
    ],
    "Mean": [
        cv_df["R2"].mean(),
        cv_df["RMSE"].mean(),
        cv_df["MAE"].mean(),
        cv_df["Pearson_r"].mean()
    ],
    "SD": [
        cv_df["R2"].std(ddof=1),
        cv_df["RMSE"].std(ddof=1),
        cv_df["MAE"].std(ddof=1),
        cv_df["Pearson_r"].std(ddof=1)
    ]
})

print("\n" + "=" * 70)
print("5-FOLD CV SUMMARY")
print("=" * 70)

for _, row in summary.iterrows():
    print(
        f"{row['Metric']:12s}: "
        f"{row['Mean']:.4f} ± {row['SD']:.4f}"
    )

# ============================================================
# SAVE
# ============================================================

CV_PATH = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_5fold_cv_results.csv"
)

SUMMARY_PATH = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_5fold_cv_summary.csv"
)

cv_df.to_csv(
    CV_PATH,
    index=False
)

summary.to_csv(
    SUMMARY_PATH,
    index=False
)

print("\n" + "=" * 70)
print("SAVED")
print("=" * 70)

print("✓ Fold results:")
print(CV_PATH)

print("\n✓ CV summary:")
print(SUMMARY_PATH)

# ============================================================
# TEST SET INTEGRITY
# ============================================================

print("\n" + "=" * 70)
print("TEST SET INTEGRITY")
print("=" * 70)

print("3,985-row test set:")
print("✓ NOT loaded")
print("✓ NOT used")
print("✓ NOT used for imputation")
print("✓ NOT used for model fitting")
print("✓ NOT used for hyperparameter selection")

print("\n✓ 5-fold protein-disjoint CV COMPLETE")
```

```python
# ============================================================
# CATPRED — 5-FOLD PROTEIN-DISJOINT PERMUTATION IMPORTANCE
# DEVELOPMENT DATA ONLY
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd

from sklearn.model_selection import GroupKFold
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.inspection import permutation_importance
from sklearn.metrics import r2_score

# ============================================================
# PATHS
# ============================================================

BASE = "/content/drive/MyDrive/CatPred_kcat_Project"
PROTEIN_DIR = os.path.join(BASE, "protein_level_evaluation")

# ============================================================
# LOAD TRAIN + VALIDATION ONLY
# ============================================================

with open(
    os.path.join(PROTEIN_DIR, "protein_X_train.pkl"), "rb"
) as f:
    X_train = pickle.load(f)

with open(
    os.path.join(PROTEIN_DIR, "protein_y_train.pkl"), "rb"
) as f:
    y_train = pickle.load(f)

with open(
    os.path.join(PROTEIN_DIR, "protein_X_val.pkl"), "rb"
) as f:
    X_val = pickle.load(f)

with open(
    os.path.join(PROTEIN_DIR, "protein_y_val.pkl"), "rb"
) as f:
    y_val = pickle.load(f)

with open(
    os.path.join(PROTEIN_DIR, "protein_feature_names.pkl"), "rb"
) as f:
    feature_names = list(pickle.load(f))

train_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_train_metadata.csv"
    )
)

val_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_validation_metadata.csv"
    )
)

# ============================================================
# RECONSTRUCT DEVELOPMENT SET
# ============================================================

X_train_df = pd.DataFrame(
    X_train,
    columns=feature_names
)

X_val_df = pd.DataFrame(
    X_val,
    columns=feature_names
)

X_dev = pd.concat(
    [X_train_df, X_val_df],
    ignore_index=True
)

y_dev = np.concatenate([
    np.asarray(y_train),
    np.asarray(y_val)
])

metadata_dev = pd.concat(
    [train_meta, val_meta],
    ignore_index=True
)

print("=" * 70)
print("CATPRED — 5-FOLD PROTEIN-DISJOINT PERMUTATION IMPORTANCE")
print("=" * 70)

print("\nDevelopment data:")
print("X_dev:", X_dev.shape)
print("y_dev:", y_dev.shape)
print("Metadata:", metadata_dev.shape)

assert len(X_dev) == len(y_dev)
assert len(X_dev) == len(metadata_dev)

print("\n✓ Train + validation only")
print("✓ Held-out test NOT loaded")

# ============================================================
# LOAD EXACT 10 FEATURES
# ============================================================

with open(
    os.path.join(
        PROTEIN_DIR,
        "protein_disjoint_10feature_names.pkl"
    ),
    "rb"
) as f:
    selected_features = list(pickle.load(f))

print("\nSelected features:")

for i, feature in enumerate(selected_features, 1):
    print(f"{i:2d}. {feature}")

assert len(selected_features) == 10
assert all(
    feature in X_dev.columns
    for feature in selected_features
)

X_dev_10 = X_dev[selected_features].copy()

# ============================================================
# GROUPS
# ============================================================

GROUP_COLUMN = "sequence_20cluster"

groups = metadata_dev[
    GROUP_COLUMN
].to_numpy()

assert not pd.isna(groups).any()

print("\nGrouping variable:", GROUP_COLUMN)
print(
    "Unique clusters:",
    len(np.unique(groups))
)

# ============================================================
# MODEL PARAMETERS
# ============================================================

MODEL_PARAMS = {
    "n_estimators": 200,
    "min_samples_leaf": 4,
    "min_samples_split": 2,
    "max_features": 1.0,
    "n_jobs": -1,
    "random_state": 42
}

# ============================================================
# GROUP K-FOLD
# ============================================================

gkf = GroupKFold(
    n_splits=5
)

# ============================================================
# PERMUTATION SETTINGS
# ============================================================

N_REPEATS = 20

fold_results = []

# Store every individual permutation score
all_importances = []

# ============================================================
# RUN 5-FOLD ANALYSIS
# ============================================================

for fold, (train_idx, val_idx) in enumerate(
    gkf.split(
        X_dev_10,
        y_dev,
        groups=groups
    ),
    start=1
):

    print("\n" + "=" * 70)
    print(f"FOLD {fold}")
    print("=" * 70)

    # --------------------------------------------------------
    # SPLIT
    # --------------------------------------------------------

    X_fold_train = X_dev_10.iloc[train_idx]
    X_fold_val = X_dev_10.iloc[val_idx]

    y_fold_train = y_dev[train_idx]
    y_fold_val = y_dev[val_idx]

    train_groups = set(
        groups[train_idx]
    )

    val_groups = set(
        groups[val_idx]
    )

    overlap = train_groups.intersection(
        val_groups
    )

    print(
        "Training samples   :",
        len(train_idx)
    )

    print(
        "Validation samples :",
        len(val_idx)
    )

    print(
        "Cluster overlap    :",
        len(overlap)
    )

    assert len(overlap) == 0

    # --------------------------------------------------------
    # TRAIN-ONLY IMPUTATION
    # --------------------------------------------------------

    imputer = SimpleImputer(
        strategy="median"
    )

    X_fold_train_imp = imputer.fit_transform(
        X_fold_train
    )

    X_fold_val_imp = imputer.transform(
        X_fold_val
    )

    # --------------------------------------------------------
    # TRAIN MODEL
    # --------------------------------------------------------

    model = ExtraTreesRegressor(
        **MODEL_PARAMS
    )

    model.fit(
        X_fold_train_imp,
        y_fold_train
    )

    # --------------------------------------------------------
    # BASELINE VALIDATION R²
    # --------------------------------------------------------

    baseline_pred = model.predict(
        X_fold_val_imp
    )

    baseline_r2 = r2_score(
        y_fold_val,
        baseline_pred
    )

    print(
        f"Baseline validation R²: "
        f"{baseline_r2:.4f}"
    )

    # --------------------------------------------------------
    # PERMUTATION IMPORTANCE
    # --------------------------------------------------------
    #
    # Important:
    # Permutation is performed ONLY on the validation fold.
    #
    # The model remains fixed.
    # --------------------------------------------------------

    perm = permutation_importance(
        model,
        X_fold_val_imp,
        y_fold_val,
        scoring="r2",
        n_repeats=N_REPEATS,
        random_state=42 + fold,
        n_jobs=-1
    )

    # --------------------------------------------------------
    # STORE FOLD RESULTS
    # --------------------------------------------------------

    for i, feature in enumerate(
        selected_features
    ):

        mean_importance = perm.importances_mean[i]
        std_importance = perm.importances_std[i]

        fold_results.append({
            "Fold": fold,
            "Feature": feature,
            "Importance_Mean": mean_importance,
            "Importance_SD": std_importance,
            "Baseline_R2": baseline_r2
        })

        for repeat_value in perm.importances[i]:

            all_importances.append({
                "Fold": fold,
                "Feature": feature,
                "Permutation_Importance": repeat_value
            })

    # --------------------------------------------------------
    # PRINT FOLD RANKING
    # --------------------------------------------------------

    fold_df = pd.DataFrame([
        {
            "Feature": selected_features[i],
            "Importance": perm.importances_mean[i]
        }
        for i in range(len(selected_features))
    ])

    fold_df = fold_df.sort_values(
        "Importance",
        ascending=False
    )

    print("\nFeature ranking:")

    print(
        fold_df.to_string(
            index=False,
            float_format=lambda x: f"{x:.6f}"
        )
    )

# ============================================================
# COMBINE RESULTS
# ============================================================

fold_df = pd.DataFrame(
    fold_results
)

repeat_df = pd.DataFrame(
    all_importances
)

# ============================================================
# AGGREGATE ACROSS FOLDS
# ============================================================

summary = (
    fold_df
    .groupby("Feature")
    .agg(
        Mean_Permutation_Importance=(
            "Importance_Mean",
            "mean"
        ),
        SD_Across_Folds=(
            "Importance_Mean",
            "std"
        ),
        Mean_Within_Fold_SD=(
            "Importance_SD",
            "mean"
        ),
        Mean_Baseline_R2=(
            "Baseline_R2",
            "mean"
        )
    )
    .reset_index()
)

summary = summary.sort_values(
    "Mean_Permutation_Importance",
    ascending=False
).reset_index(drop=True)

# ============================================================
# RANK
# ============================================================

summary.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(summary) + 1
    )
)

# ============================================================
# RELATIVE CONTRIBUTION
# ============================================================

# Use positive importance only for relative contribution.
positive_total = summary[
    "Mean_Permutation_Importance"
].clip(
    lower=0
).sum()

if positive_total > 0:

    summary[
        "Relative_Positive_Importance_%"
    ] = (
        summary[
            "Mean_Permutation_Importance"
        ].clip(lower=0)
        / positive_total
        * 100
    )

else:

    summary[
        "Relative_Positive_Importance_%"
    ] = 0

# ============================================================
# PRINT FINAL RESULTS
# ============================================================

print("\n" + "=" * 70)
print("5-FOLD PROTEIN-DISJOINT PERMUTATION IMPORTANCE")
print("=" * 70)

print(
    summary.to_string(
        index=False,
        float_format=lambda x: f"{x:.6f}"
    )
)

# ============================================================
# SAVE
# ============================================================

FOLD_PATH = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_5fold_permutation_importance_by_fold.csv"
)

SUMMARY_PATH = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_5fold_permutation_importance_summary.csv"
)

REPEAT_PATH = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_5fold_permutation_importance_all_repeats.csv"
)

fold_df.to_csv(
    FOLD_PATH,
    index=False
)

summary.to_csv(
    SUMMARY_PATH,
    index=False
)

repeat_df.to_csv(
    REPEAT_PATH,
    index=False
)

# ============================================================
# FINAL CHECK
# ============================================================

print("\n" + "=" * 70)
print("PERMUTATION IMPORTANCE COMPLETE")
print("=" * 70)

print("\n✓ Fold-level results:")
print(FOLD_PATH)

print("\n✓ Aggregated results:")
print(SUMMARY_PATH)

print("\n✓ All permutation repeats:")
print(REPEAT_PATH)

print("\n" + "=" * 70)
print("TEST SET INTEGRITY")
print("=" * 70)

print("✓ 3,985-row test set NOT loaded")
print("✓ Test set NOT used")
print("✓ Test predictions NOT used")
print("✓ No test-based feature selection")
print("✓ No test-based tuning")
print("✓ Imputation performed separately within each fold")
print("✓ Zero sequence-cluster overlap within every fold")
```

```python
# ============================================================
# CATPRED — PROTEIN-DISJOINT PREDICTION CALIBRATION /
# RANGE-COMPRESSION ANALYSIS
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

from sklearn.model_selection import GroupKFold

# ============================================================
# PATHS
# ============================================================

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

# ============================================================
# LOAD DEVELOPMENT DATA ONLY
# ============================================================

X_dev = pd.read_pickle(
    os.path.join(BASE, "protein_X_train.pkl")
)

X_val = pd.read_pickle(
    os.path.join(BASE, "protein_X_val.pkl")
)

y_dev = pd.read_pickle(
    os.path.join(BASE, "protein_y_train.pkl")
)

y_val = pd.read_pickle(
    os.path.join(BASE, "protein_y_val.pkl")
)

train_meta = pd.read_csv(
    os.path.join(BASE, "protein_train_metadata.csv")
)

val_meta = pd.read_csv(
    os.path.join(BASE, "protein_validation_metadata.csv")
)

# Combine original protein-disjoint train + validation
X_all = pd.concat([X_dev, X_val], axis=0).reset_index(drop=True)
y_all = pd.concat([y_dev, y_val], axis=0).reset_index(drop=True)

meta_all = pd.concat(
    [train_meta, val_meta],
    axis=0,
    ignore_index=True
)

print("=" * 70)
print("CATPRED — PROTEIN-DISJOINT RANGE-COMPRESSION ANALYSIS")
print("=" * 70)

print("\nDevelopment data:")
print("X:", X_all.shape)
print("y:", y_all.shape)
print("Metadata:", meta_all.shape)

# ============================================================
# SELECT FINAL 10 FEATURES
# ============================================================

FEATURES = [
    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms",
    "temperature",
    "ph"
]

X_all = X_all[FEATURES].copy()

groups = meta_all["sequence_20cluster"].to_numpy()

print("\nSelected features:")
for i, f in enumerate(FEATURES, 1):
    print(f"{i:2d}. {f}")

print("\nUnique protein clusters:",
      len(np.unique(groups)))

# ============================================================
# 5-FOLD PROTEIN-DISJOINT CV
# ============================================================

gkf = GroupKFold(n_splits=5)

fold_results = []
all_predictions = []

for fold, (train_idx, val_idx) in enumerate(
    gkf.split(X_all, y_all, groups=groups), 1
):

    print("\n" + "-" * 70)
    print(f"FOLD {fold}")
    print("-" * 70)

    X_train_fold = X_all.iloc[train_idx]
    X_val_fold = X_all.iloc[val_idx]

    y_train_fold = y_all.iloc[train_idx].to_numpy()
    y_val_fold = y_all.iloc[val_idx].to_numpy()

    groups_train = groups[train_idx]
    groups_val = groups[val_idx]

    # --------------------------------------------------------
    # TRAIN-ONLY IMPUTATION
    # --------------------------------------------------------

    imputer = SimpleImputer(strategy="median")

    X_train_imp = imputer.fit_transform(X_train_fold)
    X_val_imp = imputer.transform(X_val_fold)

    # --------------------------------------------------------
    # FINAL 10-FEATURE MODEL
    # --------------------------------------------------------

    model = ExtraTreesRegressor(
        n_estimators=200,
        min_samples_leaf=4,
        min_samples_split=2,
        max_features=1.0,
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train_imp, y_train_fold)

    predictions = model.predict(X_val_imp)

    # --------------------------------------------------------
    # BASIC PERFORMANCE
    # --------------------------------------------------------

    r2 = r2_score(y_val_fold, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_val_fold, predictions)
    )

    mae = mean_absolute_error(
        y_val_fold,
        predictions
    )

    r, p = pearsonr(
        y_val_fold,
        predictions
    )

    # --------------------------------------------------------
    # RANGE COMPRESSION
    # --------------------------------------------------------

    observed_sd = np.std(y_val_fold, ddof=1)
    predicted_sd = np.std(predictions, ddof=1)

    sd_ratio = predicted_sd / observed_sd

    observed_mean = np.mean(y_val_fold)
    predicted_mean = np.mean(predictions)

    mean_bias = predicted_mean - observed_mean

    print(f"R²                  : {r2:.4f}")
    print(f"RMSE                : {rmse:.4f}")
    print(f"MAE                 : {mae:.4f}")
    print(f"Pearson r           : {r:.4f}")
    print(f"Observed SD         : {observed_sd:.4f}")
    print(f"Predicted SD        : {predicted_sd:.4f}")
    print(f"Prediction/Observed : {sd_ratio:.4f}")
    print(f"Mean bias           : {mean_bias:.4f}")

    # --------------------------------------------------------
    # TERTILES WITHIN EACH VALIDATION FOLD
    # --------------------------------------------------------

    q1 = np.quantile(y_val_fold, 1/3)
    q2 = np.quantile(y_val_fold, 2/3)

    fold_groups = np.where(
        y_val_fold <= q1,
        "Low",
        np.where(
            y_val_fold <= q2,
            "Medium",
            "High"
        )
    )

    for group_name in ["Low", "Medium", "High"]:

        mask = fold_groups == group_name

        actual = y_val_fold[mask]
        pred = predictions[mask]

        group_observed_mean = np.mean(actual)
        group_predicted_mean = np.mean(pred)

        group_bias = (
            group_predicted_mean -
            group_observed_mean
        )

        fold_results.append({
            "Fold": fold,
            "N": len(actual),
            "R2": r2_score(actual, pred),
            "RMSE": np.sqrt(
                mean_squared_error(actual, pred)
            ),
            "MAE": mean_absolute_error(actual, pred),
            "Pearson_r": pearsonr(actual, pred)[0],
            "kcat_range": group_name,
            "Observed_mean": group_observed_mean,
            "Predicted_mean": group_predicted_mean,
            "Bias": group_bias
        })

    # --------------------------------------------------------
    # SAVE PREDICTIONS FOR THIS FOLD
    # --------------------------------------------------------

    for i, idx in enumerate(val_idx):

        all_predictions.append({
            "Fold": fold,
            "Original_index": idx,
            "y_true": y_val_fold[i],
            "y_pred": predictions[i]
        })

# ============================================================
# RESULTS DATAFRAME
# ============================================================

results_df = pd.DataFrame(fold_results)

pred_df = pd.DataFrame(all_predictions)

# ============================================================
# AGGREGATED RANGE-COMPRESSION RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("5-FOLD RANGE-COMPRESSION SUMMARY")
print("=" * 70)

compression_summary = []

for fold in sorted(pred_df["Fold"].unique()):

    subset = pred_df[
        pred_df["Fold"] == fold
    ]

    actual = subset["y_true"].to_numpy()
    pred = subset["y_pred"].to_numpy()

    observed_sd = np.std(actual, ddof=1)
    predicted_sd = np.std(pred, ddof=1)

    compression_summary.append({
        "Fold": fold,
        "Observed_SD": observed_sd,
        "Predicted_SD": predicted_sd,
        "SD_Ratio": predicted_sd / observed_sd,
        "Observed_Mean": np.mean(actual),
        "Predicted_Mean": np.mean(pred),
        "Mean_Bias": np.mean(pred) - np.mean(actual)
    })

compression_df = pd.DataFrame(
    compression_summary
)

print(compression_df.to_string(index=False))

print("\n")
print("=" * 70)
print("AGGREGATED COMPRESSION")
print("=" * 70)

print(
    f"Mean SD ratio : "
    f"{compression_df['SD_Ratio'].mean():.4f} "
    f"± "
    f"{compression_df['SD_Ratio'].std():.4f}"
)

print(
    f"Mean observed SD : "
    f"{compression_df['Observed_SD'].mean():.4f}"
)

print(
    f"Mean predicted SD : "
    f"{compression_df['Predicted_SD'].mean():.4f}"
)

# ============================================================
# AGGREGATED TERTILE BIAS
# ============================================================

print("\n")
print("=" * 70)
print("PREDICTION BIAS ACROSS kcat RANGES")
print("=" * 70)

tertile_summary = (
    results_df
    .groupby("kcat_range")
    .agg(
        N=("N", "sum"),
        Observed_mean=("Observed_mean", "mean"),
        Predicted_mean=("Predicted_mean", "mean"),
        Mean_Bias=("Bias", "mean"),
        SD_Bias=("Bias", "std"),
        Mean_RMSE=("RMSE", "mean"),
        Mean_MAE=("MAE", "mean"),
        Mean_Pearson_r=("Pearson_r", "mean")
    )
    .reset_index()
)

print(
    tertile_summary.to_string(index=False)
)

# ============================================================
# OVERALL OOF PERFORMANCE
# ============================================================

actual_all = pred_df["y_true"].to_numpy()
pred_all = pred_df["y_pred"].to_numpy()

overall_r2 = r2_score(
    actual_all,
    pred_all
)

overall_rmse = np.sqrt(
    mean_squared_error(
        actual_all,
        pred_all
    )
)

overall_mae = mean_absolute_error(
    actual_all,
    pred_all
)

overall_r, overall_p = pearsonr(
    actual_all,
    pred_all
)

overall_observed_sd = np.std(
    actual_all,
    ddof=1
)

overall_predicted_sd = np.std(
    pred_all,
    ddof=1
)

print("\n")
print("=" * 70)
print("OVERALL OUT-OF-FOLD DEVELOPMENT PERFORMANCE")
print("=" * 70)

print(f"R²                  : {overall_r2:.4f}")
print(f"RMSE                : {overall_rmse:.4f}")
print(f"MAE                 : {overall_mae:.4f}")
print(f"Pearson r           : {overall_r:.4f}")
print(f"Pearson p           : {overall_p:.4e}")

print(
    f"Observed SD         : "
    f"{overall_observed_sd:.4f}"
)

print(
    f"Predicted SD        : "
    f"{overall_predicted_sd:.4f}"
)

print(
    f"Prediction/Observed : "
    f"{overall_predicted_sd / overall_observed_sd:.4f}"
)

# ============================================================
# SAVE
# ============================================================

compression_path = os.path.join(
    BASE,
    "protein_disjoint_5fold_range_compression.csv"
)

tertile_path = os.path.join(
    BASE,
    "protein_disjoint_5fold_kcat_range_bias.csv"
)

prediction_path = os.path.join(
    BASE,
    "protein_disjoint_5fold_oof_predictions.csv"
)

compression_df.to_csv(
    compression_path,
    index=False
)

tertile_summary.to_csv(
    tertile_path,
    index=False
)

pred_df.to_csv(
    prediction_path,
    index=False
)

print("\n")
print("=" * 70)
print("RANGE-COMPRESSION ANALYSIS COMPLETE")
print("=" * 70)

print("\nSaved:")
print(compression_path)
print(tertile_path)
print(prediction_path)

print("\n✓ Held-out 3,985-row test was NOT loaded")
print("✓ No test predictions used")
print("✓ No test tuning performed")
print("✓ Imputer fitted independently within each fold")
print("✓ Protein clusters remain disjoint within every fold")
```

```python
# ============================================================
# CATPRED — SEQUENCE NOVELTY / GENERALIZATION ANALYSIS
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

from sklearn.model_selection import GroupKFold

# ============================================================
# PATHS
# ============================================================

BASE = "/content/drive/MyDrive/CatPred_kcat_Project"
PROTEIN_DIR = os.path.join(BASE, "protein_level_evaluation")

# ============================================================
# LOAD DEVELOPMENT DATA
# ============================================================

X_dev = np.load(
    os.path.join(PROTEIN_DIR, "protein_X_train.npy")
) if os.path.exists(
    os.path.join(PROTEIN_DIR, "protein_X_train.npy")
) else None

# ------------------------------------------------------------
# Safer loading from the saved protein-level matrices
# ------------------------------------------------------------

def load_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)

X_train = load_pickle(
    os.path.join(PROTEIN_DIR, "protein_X_train.pkl")
)

X_val = load_pickle(
    os.path.join(PROTEIN_DIR, "protein_X_val.pkl")
)

y_train = load_pickle(
    os.path.join(PROTEIN_DIR, "protein_y_train.pkl")
)

y_val = load_pickle(
    os.path.join(PROTEIN_DIR, "protein_y_val.pkl")
)

feature_names = load_pickle(
    os.path.join(PROTEIN_DIR, "protein_feature_names.pkl")
)

# Convert to arrays
X_train = np.asarray(X_train)
X_val = np.asarray(X_val)
y_train = np.asarray(y_train)
y_val = np.asarray(y_val)

# ============================================================
# LOAD METADATA
# ============================================================

train_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_train_metadata.csv"
    )
)

val_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_validation_metadata.csv"
    )
)

dev_meta = pd.concat(
    [train_meta, val_meta],
    axis=0,
    ignore_index=True
)

X_dev = np.vstack([X_train, X_val])
y_dev = np.concatenate([y_train, y_val])

print("=" * 70)
print("CATPRED — SEQUENCE NOVELTY / GENERALIZATION ANALYSIS")
print("=" * 70)

print("\nDevelopment data:")
print("X_dev:", X_dev.shape)
print("y_dev:", y_dev.shape)
print("Metadata:", dev_meta.shape)

# ============================================================
# SELECT THE 10-FEATURE MODEL
# ============================================================

selected_features = [
    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms",
    "temperature",
    "ph"
]

feature_indices = [
    feature_names.index(f)
    for f in selected_features
]

X_dev_10 = X_dev[:, feature_indices]

print("\nSelected features:")
for i, f in enumerate(selected_features, 1):
    print(f"{i:2d}. {f}")

print("\nX_dev_10:", X_dev_10.shape)

# ============================================================
# CHECK CLUSTER COLUMNS
# ============================================================

cluster_columns = [
    "sequence_20cluster",
    "sequence_40cluster",
    "sequence_60cluster",
    "sequence_80cluster",
    "sequence_99cluster"
]

available_clusters = [
    c for c in cluster_columns
    if c in dev_meta.columns
]

print("\nAvailable sequence-cluster levels:")
for c in available_clusters:
    print(
        f"{c}: "
        f"{dev_meta[c].nunique()} unique clusters"
    )

# ============================================================
# MODEL PARAMETERS
# Same architecture as final selected model
# ============================================================

MODEL_PARAMS = {
    "n_estimators": 200,
    "min_samples_leaf": 4,
    "min_samples_split": 2,
    "max_features": 1.0,
    "random_state": 42,
    "n_jobs": -1
}

# ============================================================
# EVALUATION FUNCTION
# ============================================================

def evaluate_model(y_true, y_pred):

    r2 = r2_score(y_true, y_pred)

    rmse = np.sqrt(
        mean_squared_error(y_true, y_pred)
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    r, p = pearsonr(
        y_true,
        y_pred
    )

    return {
        "R2": r2,
        "RMSE": rmse,
        "MAE": mae,
        "Pearson_r": r,
        "Pearson_p": p
    }

# ============================================================
# RUN PROTEIN-DISJOINT CV AT EACH CLUSTER THRESHOLD
# ============================================================

all_results = []

for cluster_col in available_clusters:

    print("\n")
    print("=" * 70)
    print(f"SEQUENCE CLUSTER LEVEL: {cluster_col}")
    print("=" * 70)

    groups = dev_meta[cluster_col].to_numpy()

    print(
        "Unique clusters:",
        len(np.unique(groups))
    )

    print(
        "Samples:",
        len(groups)
    )

    # --------------------------------------------------------
    # 5-fold group-disjoint CV
    # --------------------------------------------------------

    gkf = GroupKFold(
        n_splits=5
    )

    fold_results = []

    for fold, (train_idx, val_idx) in enumerate(
        gkf.split(
            X_dev_10,
            y_dev,
            groups=groups
        ),
        start=1
    ):

        print(f"\nFold {fold}")

        train_groups = set(
            groups[train_idx]
        )

        val_groups = set(
            groups[val_idx]
        )

        overlap = (
            train_groups &
            val_groups
        )

        print(
            "Training samples:",
            len(train_idx)
        )

        print(
            "Validation samples:",
            len(val_idx)
        )

        print(
            "Cluster overlap:",
            len(overlap)
        )

        # ----------------------------------------------------
        # Train-only imputation
        # ----------------------------------------------------

        imputer = SimpleImputer(
            strategy="median"
        )

        X_tr = imputer.fit_transform(
            X_dev_10[train_idx]
        )

        X_va = imputer.transform(
            X_dev_10[val_idx]
        )

        # ----------------------------------------------------
        # Model
        # ----------------------------------------------------

        model = ExtraTreesRegressor(
            **MODEL_PARAMS
        )

        model.fit(
            X_tr,
            y_dev[train_idx]
        )

        pred = model.predict(
            X_va
        )

        metrics = evaluate_model(
            y_dev[val_idx],
            pred
        )

        metrics["Fold"] = fold
        metrics["Cluster_Level"] = cluster_col
        metrics["N_train"] = len(train_idx)
        metrics["N_validation"] = len(val_idx)

        fold_results.append(
            metrics
        )

        print(
            f"R²          : {metrics['R2']:.4f}"
        )

        print(
            f"RMSE        : {metrics['RMSE']:.4f}"
        )

        print(
            f"MAE         : {metrics['MAE']:.4f}"
        )

        print(
            f"Pearson r   : {metrics['Pearson_r']:.4f}"
        )

        print(
            "Cluster overlap:",
            len(overlap)
        )

    # --------------------------------------------------------
    # Store fold results
    # --------------------------------------------------------

    all_results.extend(
        fold_results
    )

# ============================================================
# RESULTS TABLE
# ============================================================

results_df = pd.DataFrame(
    all_results
)

# ============================================================
# AGGREGATE RESULTS
# ============================================================

summary_df = (
    results_df
    .groupby("Cluster_Level")
    .agg(
        R2_mean=("R2", "mean"),
        R2_SD=("R2", "std"),

        RMSE_mean=("RMSE", "mean"),
        RMSE_SD=("RMSE", "std"),

        MAE_mean=("MAE", "mean"),
        MAE_SD=("MAE", "std"),

        Pearson_r_mean=("Pearson_r", "mean"),
        Pearson_r_SD=("Pearson_r", "std")
    )
    .reset_index()
)

# ============================================================
# PRINT SUMMARY
# ============================================================

print("\n")
print("=" * 70)
print("SEQUENCE NOVELTY / GENERALIZATION SUMMARY")
print("=" * 70)

print(
    summary_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ============================================================
# SAVE
# ============================================================

fold_path = os.path.join(
    PROTEIN_DIR,
    "sequence_novelty_5fold_results.csv"
)

summary_path = os.path.join(
    PROTEIN_DIR,
    "sequence_novelty_summary.csv"
)

results_df.to_csv(
    fold_path,
    index=False
)

summary_df.to_csv(
    summary_path,
    index=False
)

print("\n")
print("=" * 70)
print("ANALYSIS COMPLETE")
print("=" * 70)

print("\nSaved:")
print(f"✓ {fold_path}")
print(f"✓ {summary_path}")

print("\n✓ Held-out 3,985-row test was NOT loaded")
print("✓ No test-based model selection")
print("✓ No test-based tuning")
print("✓ Imputation performed independently within each fold")
print("✓ Sequence clusters remain disjoint within every fold")
```

```python
# ============================================================
# CATPRED — PROTEIN-DISJOINT APPLICABILITY DOMAIN ANALYSIS
# ============================================================
#
# Purpose:
# Determine whether prediction reliability decreases as a
# validation protein becomes more distant from the training
# feature distribution.
#
# IMPORTANT:
# - Uses development data ONLY (train + validation)
# - Held-out 3,985-row test is NEVER loaded
# - Distance reference is calculated from TRAINING data
#   separately within every CV fold
# - No test-based tuning or model selection
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd

from sklearn.ensemble import ExtraTreesRegressor
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr, spearmanr
from sklearn.model_selection import GroupKFold
from sklearn.neighbors import NearestNeighbors

# ============================================================
# PATHS
# ============================================================

BASE = "/content/drive/MyDrive/CatPred_kcat_Project"
PROTEIN_DIR = os.path.join(
    BASE,
    "protein_level_evaluation"
)

# ============================================================
# HELPER
# ============================================================

def load_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)

# ============================================================
# LOAD DEVELOPMENT DATA
# ============================================================

X_train = np.asarray(
    load_pickle(
        os.path.join(
            PROTEIN_DIR,
            "protein_X_train.pkl"
        )
    )
)

X_val = np.asarray(
    load_pickle(
        os.path.join(
            PROTEIN_DIR,
            "protein_X_val.pkl"
        )
    )
)

y_train = np.asarray(
    load_pickle(
        os.path.join(
            PROTEIN_DIR,
            "protein_y_train.pkl"
        )
    )
)

y_val = np.asarray(
    load_pickle(
        os.path.join(
            PROTEIN_DIR,
            "protein_y_val.pkl"
        )
    )
)

feature_names = list(
    load_pickle(
        os.path.join(
            PROTEIN_DIR,
            "protein_feature_names.pkl"
        )
    )
)

train_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_train_metadata.csv"
    )
)

val_meta = pd.read_csv(
    os.path.join(
        PROTEIN_DIR,
        "protein_validation_metadata.csv"
    )
)

# ============================================================
# COMBINE DEVELOPMENT DATA
# ============================================================

X_dev = np.vstack([
    X_train,
    X_val
])

y_dev = np.concatenate([
    y_train,
    y_val
])

dev_meta = pd.concat(
    [
        train_meta,
        val_meta
    ],
    ignore_index=True
)

print("=" * 70)
print("CATPRED — PROTEIN-DISJOINT APPLICABILITY DOMAIN ANALYSIS")
print("=" * 70)

print("\nDevelopment data:")
print("X_dev:", X_dev.shape)
print("y_dev:", y_dev.shape)
print("Metadata:", dev_meta.shape)

# ============================================================
# SELECT FINAL 10 FEATURES
# ============================================================

selected_features = [
    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms",
    "temperature",
    "ph"
]

feature_indices = [
    feature_names.index(f)
    for f in selected_features
]

X_dev_10 = X_dev[:, feature_indices]

print("\nSelected features:")
for i, f in enumerate(selected_features, 1):
    print(f"{i:2d}. {f}")

print("\nX_dev_10:", X_dev_10.shape)

# ============================================================
# GROUPING VARIABLE
# ============================================================

GROUP_COLUMN = "sequence_20cluster"

groups = dev_meta[
    GROUP_COLUMN
].to_numpy()

print("\nGrouping variable:", GROUP_COLUMN)
print(
    "Unique protein clusters:",
    len(np.unique(groups))
)

print(
    "Missing clusters:",
    pd.isna(groups).sum()
)

# ============================================================
# MODEL PARAMETERS
# FINAL VALIDATED 10-FEATURE MODEL
# ============================================================

MODEL_PARAMS = {
    "n_estimators": 200,
    "min_samples_leaf": 4,
    "min_samples_split": 2,
    "max_features": 1.0,
    "random_state": 42,
    "n_jobs": -1
}

# ============================================================
# 5-FOLD PROTEIN-DISJOINT CV
# ============================================================

gkf = GroupKFold(
    n_splits=5
)

all_predictions = []
fold_summaries = []

for fold, (train_idx, val_idx) in enumerate(
    gkf.split(
        X_dev_10,
        y_dev,
        groups=groups
    ),
    start=1
):

    print("\n")
    print("-" * 70)
    print(f"FOLD {fold}")
    print("-" * 70)

    # --------------------------------------------------------
    # Verify protein-disjoint split
    # --------------------------------------------------------

    train_groups = set(
        groups[train_idx]
    )

    val_groups = set(
        groups[val_idx]
    )

    overlap = (
        train_groups &
        val_groups
    )

    print(
        "Training samples:",
        len(train_idx)
    )

    print(
        "Validation samples:",
        len(val_idx)
    )

    print(
        "Cluster overlap:",
        len(overlap)
    )

    assert len(overlap) == 0

    # --------------------------------------------------------
    # TRAIN-ONLY IMPUTATION
    # --------------------------------------------------------

    imputer = SimpleImputer(
        strategy="median"
    )

    X_tr = imputer.fit_transform(
        X_dev_10[train_idx]
    )

    X_va = imputer.transform(
        X_dev_10[val_idx]
    )

    # --------------------------------------------------------
    # TRAIN-ONLY STANDARDIZATION FOR DISTANCE
    #
    # IMPORTANT:
    # The model itself does NOT use these standardized values.
    #
    # Standardization is ONLY for calculating feature-space
    # distance so that variables with different scales do not
    # dominate the distance calculation.
    # --------------------------------------------------------

    scaler = StandardScaler()

    X_tr_scaled = scaler.fit_transform(
        X_tr
    )

    X_va_scaled = scaler.transform(
        X_va
    )

    # --------------------------------------------------------
    # TRAIN FINAL MODEL FOR THIS FOLD
    # --------------------------------------------------------

    model = ExtraTreesRegressor(
        **MODEL_PARAMS
    )

    model.fit(
        X_tr,
        y_dev[train_idx]
    )

    predictions = model.predict(
        X_va
    )

    # --------------------------------------------------------
    # OVERALL FOLD PERFORMANCE
    # --------------------------------------------------------

    r2 = r2_score(
        y_dev[val_idx],
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_dev[val_idx],
            predictions
        )
    )

    mae = mean_absolute_error(
        y_dev[val_idx],
        predictions
    )

    r, p = pearsonr(
        y_dev[val_idx],
        predictions
    )

    print(
        f"R²        : {r2:.4f}"
    )

    print(
        f"RMSE      : {rmse:.4f}"
    )

    print(
        f"MAE       : {mae:.4f}"
    )

    print(
        f"Pearson r : {r:.4f}"
    )

    # --------------------------------------------------------
    # NEAREST-NEIGHBOR DISTANCE
    #
    # For every validation protein, calculate its distance to
    # the nearest TRAINING protein in standardized feature
    # space.
    # --------------------------------------------------------

    nn = NearestNeighbors(
        n_neighbors=1,
        metric="euclidean",
        n_jobs=-1
    )

    nn.fit(
        X_tr_scaled
    )

    distances, indices = nn.kneighbors(
        X_va_scaled
    )

    nearest_distance = distances[:, 0]

    # --------------------------------------------------------
    # ABSOLUTE PREDICTION ERROR
    # --------------------------------------------------------

    actual = y_dev[val_idx]

    residual = actual - predictions

    abs_error = np.abs(
        residual
    )

    squared_error = (
        residual ** 2
    )

    # --------------------------------------------------------
    # STORE SAMPLE-LEVEL RESULTS
    # --------------------------------------------------------

    fold_df = pd.DataFrame({

        "Fold": fold,

        "sample_index": val_idx,

        "y_true": actual,

        "y_pred": predictions,

        "residual": residual,

        "absolute_error": abs_error,

        "squared_error": squared_error,

        "nearest_train_distance":
            nearest_distance,

        "kcat_range": np.select(
            [
                actual <= np.quantile(
                    actual,
                    1/3
                ),

                actual <= np.quantile(
                    actual,
                    2/3
                )
            ],
            [
                "Low",
                "Medium"
            ],
            default="High"
        )
    })

    all_predictions.append(
        fold_df
    )

    fold_summaries.append({

        "Fold": fold,

        "N_validation": len(val_idx),

        "R2": r2,

        "RMSE": rmse,

        "MAE": mae,

        "Pearson_r": r,

        "Mean_nearest_distance":
            nearest_distance.mean(),

        "Median_nearest_distance":
            np.median(nearest_distance),

        "Max_nearest_distance":
            nearest_distance.max()
    })

# ============================================================
# COMBINE ALL OUT-OF-FOLD PREDICTIONS
# ============================================================

oof_df = pd.concat(
    all_predictions,
    ignore_index=True
)

print("\n")
print("=" * 70)
print("OUT-OF-FOLD APPLICABILITY DOMAIN DATASET")
print("=" * 70)

print(
    "Total OOF predictions:",
    len(oof_df)
)

# ============================================================
# CORRELATION BETWEEN DISTANCE AND ERROR
# ============================================================

distance = oof_df[
    "nearest_train_distance"
].to_numpy()

absolute_error = oof_df[
    "absolute_error"
].to_numpy()

# Pearson
pearson_distance_error = pearsonr(
    distance,
    absolute_error
)

# Spearman
spearman_distance_error = spearmanr(
    distance,
    absolute_error
)

print("\n")
print("=" * 70)
print("DISTANCE vs PREDICTION ERROR")
print("=" * 70)

print(
    f"Pearson r  : "
    f"{pearson_distance_error.statistic:.4f}"
)

print(
    f"Pearson p  : "
    f"{pearson_distance_error.pvalue:.4e}"
)

print(
    f"Spearman ρ : "
    f"{spearman_distance_error.statistic:.4f}"
)

print(
    f"Spearman p : "
    f"{spearman_distance_error.pvalue:.4e}"
)

# ============================================================
# CREATE DISTANCE QUARTILES
# ============================================================

oof_df["AD_quartile"] = pd.qcut(
    oof_df[
        "nearest_train_distance"
    ],
    q=4,
    labels=[
        "Q1 — Closest",
        "Q2",
        "Q3",
        "Q4 — Most Distant"
    ],
    duplicates="drop"
)

# ============================================================
# PERFORMANCE ACROSS APPLICABILITY DOMAIN
# ============================================================

ad_results = []

for group_name, subset in oof_df.groupby(
    "AD_quartile",
    observed=False
):

    y_true = subset[
        "y_true"
    ].to_numpy()

    y_pred = subset[
        "y_pred"
    ].to_numpy()

    r2 = r2_score(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    r, p = pearsonr(
        y_true,
        y_pred
    )

    ad_results.append({

        "AD_quartile": str(group_name),

        "N": len(subset),

        "Mean_distance":
            subset[
                "nearest_train_distance"
            ].mean(),

        "Median_distance":
            subset[
                "nearest_train_distance"
            ].median(),

        "R2": r2,

        "RMSE": rmse,

        "MAE": mae,

        "Pearson_r": r,

        "Pearson_p": p,

        "Mean_absolute_error":
            subset[
                "absolute_error"
            ].mean()
    })

ad_results_df = pd.DataFrame(
    ad_results
)

# ============================================================
# PRINT AD RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("PERFORMANCE ACROSS APPLICABILITY-DOMAIN QUARTILES")
print("=" * 70)

print(
    ad_results_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ============================================================
# DISTANCE vs ERROR BY FOLD
# ============================================================

fold_distance_results = []

for fold, subset in oof_df.groupby(
    "Fold"
):

    r, p = pearsonr(
        subset[
            "nearest_train_distance"
        ],
        subset[
            "absolute_error"
        ]
    )

    rho, sp = spearmanr(
        subset[
            "nearest_train_distance"
        ],
        subset[
            "absolute_error"
        ]
    )

    fold_distance_results.append({

        "Fold": fold,

        "N": len(subset),

        "Pearson_distance_error":
            r,

        "Pearson_p":
            p,

        "Spearman_distance_error":
            rho,

        "Spearman_p":
            sp,

        "Mean_distance":
            subset[
                "nearest_train_distance"
            ].mean(),

        "Mean_absolute_error":
            subset[
                "absolute_error"
            ].mean()
    })

fold_distance_df = pd.DataFrame(
    fold_distance_results
)

# ============================================================
# PRINT FOLD DISTANCE RESULTS
# ============================================================

print("\n")
print("=" * 70)
print("DISTANCE–ERROR ASSOCIATION BY FOLD")
print("=" * 70)

print(
    fold_distance_df.to_string(
        index=False,
        float_format=lambda x: f"{x:.4f}"
    )
)

# ============================================================
# OVERALL OOF PERFORMANCE
# ============================================================

overall_y = oof_df[
    "y_true"
].to_numpy()

overall_pred = oof_df[
    "y_pred"
].to_numpy()

overall_r2 = r2_score(
    overall_y,
    overall_pred
)

overall_rmse = np.sqrt(
    mean_squared_error(
        overall_y,
        overall_pred
    )
)

overall_mae = mean_absolute_error(
    overall_y,
    overall_pred
)

overall_r, overall_p = pearsonr(
    overall_y,
    overall_pred
)

print("\n")
print("=" * 70)
print("OVERALL OUT-OF-FOLD PERFORMANCE")
print("=" * 70)

print(
    f"R²        : {overall_r2:.4f}"
)

print(
    f"RMSE      : {overall_rmse:.4f}"
)

print(
    f"MAE       : {overall_mae:.4f}"
)

print(
    f"Pearson r : {overall_r:.4f}"
)

print(
    f"Pearson p : {overall_p:.4e}"
)

# ============================================================
# SAVE RESULTS
# ============================================================

oof_path = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_applicability_domain_oof.csv"
)

ad_path = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_applicability_domain_quartiles.csv"
)

fold_path = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_distance_error_by_fold.csv"
)

summary_path = os.path.join(
    PROTEIN_DIR,
    "protein_disjoint_applicability_domain_summary.csv"
)

oof_df.to_csv(
    oof_path,
    index=False
)

ad_results_df.to_csv(
    ad_path,
    index=False
)

fold_distance_df.to_csv(
    fold_path,
    index=False
)

summary_df = pd.DataFrame([{

    "OOF_R2": overall_r2,

    "OOF_RMSE": overall_rmse,

    "OOF_MAE": overall_mae,

    "OOF_Pearson_r": overall_r,

    "OOF_Pearson_p": overall_p,

    "Distance_Error_Pearson_r":
        pearson_distance_error.statistic,

    "Distance_Error_Pearson_p":
        pearson_distance_error.pvalue,

    "Distance_Error_Spearman_rho":
        spearman_distance_error.statistic,

    "Distance_Error_Spearman_p":
        spearman_distance_error.pvalue,

    "Mean_Nearest_Train_Distance":
        distance.mean(),

    "Median_Nearest_Train_Distance":
        np.median(distance),

    "Q1_Mean_Absolute_Error":
        ad_results_df.iloc[0][
            "Mean_absolute_error"
        ],

    "Q4_Mean_Absolute_Error":
        ad_results_df.iloc[-1][
            "Mean_absolute_error"
        ]
}])

summary_df.to_csv(
    summary_path,
    index=False
)

# ============================================================
# FINAL INTEGRITY CHECK
# ============================================================

print("\n")
print("=" * 70)
print("APPLICABILITY DOMAIN ANALYSIS COMPLETE")
print("=" * 70)

print("\nSaved:")

print(
    f"✓ {oof_path}"
)

print(
    f"✓ {ad_path}"
)

print(
    f"✓ {fold_path}"
)

print(
    f"✓ {summary_path}"
)

print("\nTEST SET INTEGRITY")
print("✓ 3,985-row held-out test was NOT loaded")
print("✓ No test predictions used")
print("✓ No test-based tuning")
print("✓ No test-based feature selection")
print("✓ Imputer fitted separately within each fold")
print("✓ Distance scaler fitted separately within each fold")
print("✓ Nearest-neighbor reference uses TRAINING proteins only")
print("✓ Sequence clusters remain disjoint within every fold")
```

```python
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# Align development metadata to OOF sample_index
# ------------------------------------------------------------

meta_oof = development_df.reset_index(drop=True).copy()

assert len(meta_oof) == len(oof_df), (
    f"Length mismatch: metadata={len(meta_oof)}, OOF={len(oof_df)}"
)

print("Metadata rows:", len(meta_oof))
print("OOF rows:", len(oof_df))

# ------------------------------------------------------------
# Biological overlap function
# ------------------------------------------------------------

def calculate_overlap(train_meta, val_meta, column):
    train_values = set(
        train_meta[column]
        .dropna()
        .astype(str)
    )

    val_values = set(
        val_meta[column]
        .dropna()
        .astype(str)
    )

    overlap = train_values & val_values

    # Validation samples whose biological group is represented
    # somewhere in the training set
    val_overlap_mask = val_meta[column].astype(str).isin(train_values)

    overlapping_val_samples = val_overlap_mask.sum()

    percentage = (
        100 * overlapping_val_samples / len(val_meta)
        if len(val_meta) > 0 else np.nan
    )

    return {
        "train_unique_groups": len(train_values),
        "validation_unique_groups": len(val_values),
        "overlapping_groups": len(overlap),
        "validation_samples_in_train_group": overlapping_val_samples,
        "validation_sample_overlap_percent": percentage
    }


# ------------------------------------------------------------
# Analyze all five OOF folds
# ------------------------------------------------------------

cluster_columns = {
    "Exact sequence": "sequence",
    "20% sequence cluster": "sequence_20cluster",
    "40% sequence cluster": "sequence_40cluster",
    "60% sequence cluster": "sequence_60cluster",
    "80% sequence cluster": "sequence_80cluster",
    "99% sequence cluster": "sequence_99cluster"
}

results = []

for fold in sorted(oof_df["Fold"].unique()):

    # Validation samples for this fold
    val_indices = oof_df.loc[
        oof_df["Fold"] == fold,
        "sample_index"
    ].astype(int).values

    # Remaining samples = training set
    all_indices = np.arange(len(meta_oof))

    train_indices = np.setdiff1d(
        all_indices,
        val_indices
    )

    fold_train = meta_oof.iloc[train_indices]
    fold_val = meta_oof.iloc[val_indices]

    print(f"\nFold {fold}")
    print("Training:", len(fold_train))
    print("Validation:", len(fold_val))

    for level, column in cluster_columns.items():

        result = calculate_overlap(
            fold_train,
            fold_val,
            column
        )

        result["Fold"] = fold
        result["Level"] = level

        results.append(result)


overlap_df = pd.DataFrame(results)

# Organize columns
overlap_df = overlap_df[
    [
        "Fold",
        "Level",
        "train_unique_groups",
        "validation_unique_groups",
        "overlapping_groups",
        "validation_samples_in_train_group",
        "validation_sample_overlap_percent"
    ]
]

print("\n" + "=" * 80)
print("BIOLOGICAL OVERLAP ACROSS OOF FOLDS")
print("=" * 80)

display(overlap_df)
```

```python
print("=" * 75)
print("PERFORMANCE BY EXACT-SEQUENCE OVERLAP")
print("=" * 75)

results = []

for fold in sorted(oof["Fold"].unique()):

    train = oof[oof["Fold"] != fold]
    val = oof[oof["Fold"] == fold]

    # Exact sequences present in training
    train_sequences = set(train["sequence"].dropna())

    # Whether each validation sample has an identical sequence in training
    val = val.copy()

    val["exact_sequence_overlap"] = (
        val["sequence"].isin(train_sequences)
    )

    # Overlapping samples
    overlap = val[val["exact_sequence_overlap"] == True]

    # Non-overlapping samples
    no_overlap = val[val["exact_sequence_overlap"] == False]

    print(f"\nFold {fold}")
    print("-" * 75)

    print(
        f"Exact sequence overlap    : "
        f"{len(overlap)} / {len(val)} "
        f"({len(overlap)/len(val)*100:.2f}%)"
    )

    print(
        f"No exact sequence overlap : "
        f"{len(no_overlap)} / {len(val)} "
        f"({len(no_overlap)/len(val)*100:.2f}%)"
    )

    if len(overlap) > 0:
        print(
            f"Overlap MAE              : "
            f"{overlap['absolute_error'].mean():.4f}"
        )

    if len(no_overlap) > 0:
        print(
            f"Non-overlap MAE          : "
            f"{no_overlap['absolute_error'].mean():.4f}"
        )

    results.append({
        "Fold": fold,
        "Overlap_n": len(overlap),
        "NonOverlap_n": len(no_overlap),
        "Overlap_MAE": overlap["absolute_error"].mean(),
        "NonOverlap_MAE": no_overlap["absolute_error"].mean()
    })


overlap_performance = pd.DataFrame(results)

print("\n")
print("=" * 75)
print("OVERALL")
print("=" * 75)

print(
    f"Mean overlap MAE     : "
    f"{overlap_performance['Overlap_MAE'].mean():.4f}"
)

print(
    f"Mean non-overlap MAE : "
    f"{overlap_performance['NonOverlap_MAE'].mean():.4f}"
)
```

```python
print("=" * 80)
print("PERFORMANCE ACROSS BIOLOGICAL SIMILARITY LEVELS")
print("=" * 80)

cluster_levels = {
    "Exact sequence": "sequence",
    "20% sequence cluster": "sequence_20cluster",
    "40% sequence cluster": "sequence_40cluster",
    "60% sequence cluster": "sequence_60cluster",
    "80% sequence cluster": "sequence_80cluster",
    "99% sequence cluster": "sequence_99cluster"
}

all_results = []

for fold in sorted(oof["Fold"].unique()):

    train = oof[oof["Fold"] != fold]
    val = oof[oof["Fold"] == fold].copy()

    print(f"\n{'FOLD ' + str(fold):^80}")
    print("-" * 80)

    for level, column in cluster_levels.items():

        # Groups represented in training
        train_groups = set(train[column].dropna())

        # Does each validation sample belong to a group represented in training?
        val["overlap"] = val[column].isin(train_groups)

        overlap = val[val["overlap"]]
        novel = val[~val["overlap"]]

        overlap_mae = overlap["absolute_error"].mean()
        novel_mae = novel["absolute_error"].mean()

        difference = novel_mae - overlap_mae

        print(
            f"{level:<25} | "
            f"Overlap: {len(overlap):4d} | "
            f"MAE: {overlap_mae:.4f} | "
            f"Novel: {len(novel):4d} | "
            f"MAE: {novel_mae:.4f} | "
            f"ΔMAE: {difference:.4f}"
        )

        all_results.append({
            "Fold": fold,
            "Level": level,
            "Overlap_n": len(overlap),
            "Overlap_MAE": overlap_mae,
            "Novel_n": len(novel),
            "Novel_MAE": novel_mae,
            "MAE_difference": difference
        })


biological_performance = pd.DataFrame(all_results)

print("\n")
print("=" * 80)
print("MEAN PERFORMANCE ACROSS FIVE FOLDS")
print("=" * 80)

for level in cluster_levels.keys():

    subset = biological_performance[
        biological_performance["Level"] == level
    ]

    print(
        f"{level:<25} | "
        f"Overlap MAE: {subset['Overlap_MAE'].mean():.4f} | "
        f"Novel MAE: {subset['Novel_MAE'].mean():.4f} | "
        f"ΔMAE: {subset['MAE_difference'].mean():.4f}"
    )
```

```python
# ================================================================
# SUBSTRATE OVERLAP ANALYSIS
# ================================================================

import pandas as pd
import numpy as np
from rdkit import Chem
from rdkit.Chem import AllChem, DataStructs

print("=" * 70)
print("SUBSTRATE OVERLAP ACROSS OOF FOLDS")
print("=" * 70)

# ------------------------------------------------
# 1. Prepare substrate column
# ------------------------------------------------

substrate_col = "reactant_smiles"

assert substrate_col in development_df.columns, \
    f"{substrate_col} not found in development_df"

meta_sub = development_df[[substrate_col]].copy()

print("Total samples:", len(meta_sub))
print("Missing substrates:", meta_sub[substrate_col].isna().sum())


# ------------------------------------------------
# 2. Canonicalize substrate SMILES
# ------------------------------------------------

def canonicalize_smiles(smiles):
    if pd.isna(smiles):
        return None

    try:
        mol = Chem.MolFromSmiles(str(smiles))

        if mol is None:
            return None

        return Chem.MolToSmiles(mol, canonical=True)

    except:
        return None


meta_sub["canonical_substrate"] = (
    meta_sub[substrate_col]
    .apply(canonicalize_smiles)
)

print(
    "Valid substrates:",
    meta_sub["canonical_substrate"].notna().sum()
)

print(
    "Invalid substrates:",
    meta_sub["canonical_substrate"].isna().sum()
)

print(
    "Unique substrates:",
    meta_sub["canonical_substrate"].nunique()
)


# ------------------------------------------------
# 3. Exact substrate overlap
# ------------------------------------------------

substrate_overlap_results = []

for fold in sorted(oof_df["Fold"].unique()):

    # IMPORTANT:
    # sample_index is 0-based row position in development_df
    val_idx = oof_df.loc[
        oof_df["Fold"] == fold,
        "sample_index"
    ].astype(int).values

    all_idx = np.arange(len(development_df))

    train_idx = np.setdiff1d(all_idx, val_idx)

    train_substrates = set(
        meta_sub.iloc[train_idx]["canonical_substrate"]
        .dropna()
    )

    val_substrates = (
        meta_sub.iloc[val_idx]["canonical_substrate"]
        .dropna()
    )

    overlap_mask = val_substrates.isin(train_substrates)

    overlap_count = overlap_mask.sum()
    val_count = len(val_substrates)

    overlap_percent = (
        overlap_count / val_count * 100
        if val_count > 0 else np.nan
    )

    substrate_overlap_results.append({
        "Fold": fold,
        "Train unique substrates": len(train_substrates),
        "Validation unique substrates": val_substrates.nunique(),
        "Overlapping substrates":
            len(set(val_substrates) & train_substrates),
        "Validation samples with exact substrate overlap":
            overlap_count,
        "Validation substrate overlap %":
            overlap_percent
    })

    print(
        f"Fold {fold} | "
        f"Train substrates: {len(train_substrates):5d} | "
        f"Val substrates: {val_substrates.nunique():5d} | "
        f"Overlap: {len(set(val_substrates) & train_substrates):5d} | "
        f"Val overlap: {overlap_percent:6.2f}%"
    )


# ------------------------------------------------
# 4. Final table
# ------------------------------------------------

substrate_overlap_df = pd.DataFrame(
    substrate_overlap_results
)

print("\n" + "=" * 70)
print("AVERAGE SUBSTRATE OVERLAP")
print("=" * 70)

print(
    substrate_overlap_df[
        "Validation substrate overlap %"
    ].mean()
)

display(
    substrate_overlap_df.round(2)
)
```

```python
# ================================================================
# PERFORMANCE BY EXACT SUBSTRATE OVERLAP
# ================================================================

print("=" * 75)
print("PERFORMANCE BY EXACT-SUBSTRATE OVERLAP")
print("=" * 75)

substrate_mae_results = []

for fold in sorted(oof_df["Fold"].unique()):

    # Validation samples for this fold
    fold_oof = oof_df[oof_df["Fold"] == fold].copy()

    val_idx = fold_oof["sample_index"].astype(int).values

    # Training indices
    all_idx = np.arange(len(development_df))
    train_idx = np.setdiff1d(all_idx, val_idx)

    # Training substrates
    train_substrates = set(
        meta_sub.iloc[train_idx]["canonical_substrate"]
        .dropna()
    )

    # Validation substrates
    val_substrates = (
        meta_sub.iloc[val_idx]["canonical_substrate"]
        .values
    )

    # Identify overlap
    overlap_mask = np.array([
        s in train_substrates if pd.notna(s) else False
        for s in val_substrates
    ])

    # Metrics
    overlap_mae = fold_oof.loc[
        overlap_mask, "absolute_error"
    ].mean()

    novel_mae = fold_oof.loc[
        ~overlap_mask, "absolute_error"
    ].mean()

    overlap_n = overlap_mask.sum()
    novel_n = (~overlap_mask).sum()

    delta = novel_mae - overlap_mae

    substrate_mae_results.append({
        "Fold": fold,
        "Overlap N": overlap_n,
        "Novel N": novel_n,
        "Overlap MAE": overlap_mae,
        "Novel MAE": novel_mae,
        "Delta MAE": delta
    })

    print(f"\nFold {fold}")
    print("-" * 75)

    print(
        f"Substrate overlap    : "
        f"{overlap_n} / {len(fold_oof)} "
        f"({overlap_n / len(fold_oof) * 100:.2f}%)"
    )

    print(
        f"Novel substrate      : "
        f"{novel_n} / {len(fold_oof)} "
        f"({novel_n / len(fold_oof) * 100:.2f}%)"
    )

    print(
        f"Overlap MAE          : "
        f"{overlap_mae:.4f}"
    )

    print(
        f"Novel substrate MAE  : "
        f"{novel_mae:.4f}"
    )

    print(
        f"ΔMAE                : "
        f"{delta:.4f}"
    )


# ================================================================
# OVERALL
# ================================================================

substrate_mae_df = pd.DataFrame(substrate_mae_results)

print("\n")
print("=" * 75)
print("OVERALL")
print("=" * 75)

print(
    f"Mean overlap MAE     : "
    f"{substrate_mae_df['Overlap MAE'].mean():.4f}"
)

print(
    f"Mean novel MAE       : "
    f"{substrate_mae_df['Novel MAE'].mean():.4f}"
)

print(
    f"Mean ΔMAE            : "
    f"{substrate_mae_df['Delta MAE'].mean():.4f}"
)
```

```python
# ================================================================
# CHEMICAL SIMILARITY OF VALIDATION SUBSTRATES TO TRAINING
# ================================================================

from rdkit import Chem
from rdkit.Chem import AllChem, DataStructs
import numpy as np
import pandas as pd

print("=" * 75)
print("SUBSTRATE CHEMICAL SIMILARITY ANALYSIS")
print("=" * 75)

# ------------------------------------------------
# Create molecules and Morgan fingerprints
# ------------------------------------------------

meta_sub["mol"] = meta_sub["canonical_substrate"].apply(
    lambda x: Chem.MolFromSmiles(x) if pd.notna(x) else None
)

meta_sub["fingerprint"] = meta_sub["mol"].apply(
    lambda mol: AllChem.GetMorganFingerprintAsBitVect(
        mol, radius=2, nBits=2048
    ) if mol is not None else None
)

# ------------------------------------------------
# Calculate maximum similarity to training substrates
# ------------------------------------------------

def max_similarity_to_training(val_fp, train_fps):

    if val_fp is None or len(train_fps) == 0:
        return np.nan

    similarities = DataStructs.BulkTanimotoSimilarity(
        val_fp,
        train_fps
    )

    return max(similarities)


similarity_results = []

for fold in sorted(oof_df["Fold"].unique()):

    fold_oof = oof_df[
        oof_df["Fold"] == fold
    ].copy()

    val_idx = fold_oof["sample_index"].astype(int).values

    all_idx = np.arange(len(development_df))

    train_idx = np.setdiff1d(
        all_idx,
        val_idx
    )

    # Training fingerprints
    train_fps = [
        fp for fp in
        meta_sub.iloc[train_idx]["fingerprint"]
        if fp is not None
    ]

    # Validation fingerprints
    val_fps = meta_sub.iloc[
        val_idx
    ]["fingerprint"].values

    max_sims = []

    for fp in val_fps:

        max_sims.append(
            max_similarity_to_training(
                fp,
                train_fps
            )
        )

    max_sims = np.array(max_sims)

    fold_oof["max_substrate_similarity"] = max_sims

    # Remove missing values
    valid = fold_oof[
        fold_oof["max_substrate_similarity"].notna()
    ].copy()

    # ------------------------------------------------
    # Similarity categories
    # ------------------------------------------------

    valid["similarity_group"] = pd.cut(
        valid["max_substrate_similarity"],
        bins=[-np.inf, 0.7, 0.8, 0.9, np.inf],
        labels=[
            "<0.7",
            "0.7–0.8",
            "0.8–0.9",
            "≥0.9"
        ]
    )

    print(f"\nFold {fold}")
    print("-" * 75)

    for group in [
        "<0.7",
        "0.7–0.8",
        "0.8–0.9",
        "≥0.9"
    ]:

        subset = valid[
            valid["similarity_group"] == group
        ]

        if len(subset) == 0:
            print(
                f"{group:10s} | N = 0"
            )
            continue

        mae = subset["absolute_error"].mean()

        print(
            f"{group:10s} | "
            f"N = {len(subset):4d} | "
            f"MAE = {mae:.4f}"
        )

        similarity_results.append({
            "Fold": fold,
            "Similarity group": group,
            "N": len(subset),
            "MAE": mae
        })


# ------------------------------------------------
# Summary
# ------------------------------------------------

similarity_df = pd.DataFrame(
    similarity_results
)

print("\n")
print("=" * 75)
print("MEAN PERFORMANCE BY CHEMICAL SIMILARITY")
print("=" * 75)

summary = (
    similarity_df
    .groupby("Similarity group", observed=True)
    .agg(
        Mean_N=("N", "mean"),
        Mean_MAE=("MAE", "mean")
    )
    .reset_index()
)

print(summary.to_string(index=False))
```

```python
# ================================================================
# APPLICABILITY DOMAIN (AD) / NEAREST-TRAINING-DISTANCE ANALYSIS
# ================================================================

import pandas as pd
import numpy as np

df = oof_df.copy()

# ------------------------------------------------
# 1. Check required columns
# ------------------------------------------------
required = [
    "Fold",
    "nearest_train_distance",
    "absolute_error",
    "AD_quartile"
]

print("Required columns:")
print([c for c in required if c in df.columns])

# Remove missing values
df = df.dropna(
    subset=["nearest_train_distance", "absolute_error"]
).copy()

print("\nTotal samples:", len(df))


# ------------------------------------------------
# 2. Performance by AD quartile
# ------------------------------------------------
print("\n" + "=" * 75)
print("PERFORMANCE BY APPLICABILITY-DOMAIN QUARTILE")
print("=" * 75)

ad_results = []

for fold in sorted(df["Fold"].unique()):

    fold_df = df[df["Fold"] == fold]

    print(f"\nFold {fold}")
    print("-" * 75)

    for quartile in ["Q1 — Closest", "Q2", "Q3", "Q4 — Most Distant"]:

        subset = fold_df[fold_df["AD_quartile"] == quartile]

        if len(subset) == 0:
            continue

        mae = subset["absolute_error"].mean()
        mean_distance = subset["nearest_train_distance"].mean()

        print(
            f"{quartile:20s} | "
            f"N = {len(subset):4d} | "
            f"Mean distance = {mean_distance:.4f} | "
            f"MAE = {mae:.4f}"
        )

        ad_results.append({
            "Fold": fold,
            "AD_quartile": quartile,
            "N": len(subset),
            "Mean_distance": mean_distance,
            "MAE": mae
        })


# ------------------------------------------------
# 3. Mean performance across folds
# ------------------------------------------------
ad_results_df = pd.DataFrame(ad_results)

print("\n")
print("=" * 75)
print("MEAN PERFORMANCE ACROSS FIVE FOLDS")
print("=" * 75)

summary = (
    ad_results_df
    .groupby("AD_quartile")
    .agg(
        Mean_N=("N", "mean"),
        Mean_distance=("Mean_distance", "mean"),
        Mean_MAE=("MAE", "mean")
    )
    .reindex(["Q1 — Closest", "Q2", "Q3", "Q4 — Most Distant"])
)

print(summary.round(4))


# ------------------------------------------------
# 4. Overall relationship between distance and error
# ------------------------------------------------
print("\n")
print("=" * 75)
print("DISTANCE–ERROR RELATIONSHIP")
print("=" * 75)

pearson = df["nearest_train_distance"].corr(
    df["absolute_error"],
    method="pearson"
)

spearman = df["nearest_train_distance"].corr(
    df["absolute_error"],
    method="spearman"
)

print(f"Pearson correlation  : {pearson:.4f}")
print(f"Spearman correlation : {spearman:.4f}")


# ------------------------------------------------
# 5. Save results for later manuscript tables
# ------------------------------------------------
ad_results_df.to_csv(
    "applicability_domain_results.csv",
    index=False
)

summary.to_csv(
    "applicability_domain_summary.csv"
)

print("\nResults saved:")
print("applicability_domain_results.csv")
print("applicability_domain_summary.csv")
```

```python
# ================================================================
# PERFORMANCE BY kcat RANGE
# ================================================================

df = oof_df.copy()

print("=" * 75)
print("PERFORMANCE BY kcat RANGE")
print("=" * 75)

results = []

for fold in sorted(df["Fold"].unique()):

    fold_df = df[df["Fold"] == fold]

    print(f"\nFold {fold}")
    print("-" * 75)

    for category in ["Low", "Medium", "High"]:

        subset = fold_df[fold_df["kcat_range"] == category]

        if len(subset) == 0:
            print(f"{category:8s} | No samples")
            continue

        mae = subset["absolute_error"].mean()
        median_ae = subset["absolute_error"].median()

        print(
            f"{category:8s} | "
            f"N = {len(subset):4d} | "
            f"MAE = {mae:.4f} | "
            f"Median AE = {median_ae:.4f}"
        )

        results.append({
            "Fold": fold,
            "kcat_range": category,
            "N": len(subset),
            "MAE": mae,
            "Median_AE": median_ae
        })


# ================================================================
# MEAN PERFORMANCE ACROSS FOLDS
# ================================================================

results_df = pd.DataFrame(results)

print("\n")
print("=" * 75)
print("MEAN PERFORMANCE ACROSS FIVE FOLDS")
print("=" * 75)

summary = (
    results_df
    .groupby("kcat_range")
    .agg(
        Mean_N=("N", "mean"),
        Mean_MAE=("MAE", "mean"),
        Mean_Median_AE=("Median_AE", "mean")
    )
    .reindex(["Low", "Medium", "High"])
)

print(summary.round(4))


# ================================================================
# ERROR RANGE
# ================================================================

print("\n")
print("=" * 75)
print("OVERALL kcat RANGE DISTRIBUTION")
print("=" * 75)

distribution = (
    df.groupby("kcat_range")["y_true"]
    .agg(["count", "mean", "std", "min", "max"])
    .reindex(["Low", "Medium", "High"])
)

print(distribution.round(4))


# ================================================================
# SAVE RESULTS
# ================================================================

results_df.to_csv(
    "kcat_range_performance.csv",
    index=False
)

summary.to_csv(
    "kcat_range_summary.csv"
)

print("\nResults saved:")
print("kcat_range_performance.csv")
print("kcat_range_summary.csv")
```

```python
import os
import re

CHECKPOINT = (
    "/content/drive/MyDrive/CatPred_kcat_Project/"
    "checkpoints/latest_checkpoint.pkl"
)

print("File exists:", os.path.exists(CHECKPOINT))
print("Size:", os.path.getsize(CHECKPOINT) / (1024**3), "GB")

# Search the binary file in chunks for useful artifact names.
# This does NOT deserialize the pickle.

patterns = [
    b"ExtraTrees",
    b"protein_disjoint",
    b"10feature",
    b"feature_names",
    b"imputer",
    b"model",
    b"n_estimators",
    b"min_samples_leaf",
]

found = {p: [] for p in patterns}

with open(CHECKPOINT, "rb") as f:
    offset = 0

    while True:
        chunk = f.read(8 * 1024 * 1024)  # 8 MB
        if not chunk:
            break

        for p in patterns:
            pos = chunk.find(p)
            if pos != -1:
                found[p].append(offset + pos)

        offset += len(chunk)

print("\n" + "=" * 80)
print("MATCHES FOUND")
print("=" * 80)

for p, positions in found.items():
    print(p.decode(), ":", positions[:20],
          "..." if len(positions) > 20 else "")
```

```python
import pickle
import os
import numpy as np

BASE = "/content/drive/MyDrive/CatPred_kcat_Project/protein_level_evaluation"

def load_pkl(name):
    path = os.path.join(BASE, name)
    with open(path, "rb") as f:
        return pickle.load(f)

X_train = load_pkl("protein_X_train.pkl")
X_val   = load_pkl("protein_X_val.pkl")
X_test  = load_pkl("protein_X_test.pkl")

y_train = load_pkl("protein_y_train.pkl")
y_val   = load_pkl("protein_y_val.pkl")
y_test  = load_pkl("protein_y_test.pkl")

imputer10 = load_pkl("protein_disjoint_10feature_imputer_reconstructed.pkl")

print("=" * 80)
print("PROTEIN-DISJOINT DATA")
print("=" * 80)

for name, obj in [
    ("X_train", X_train),
    ("X_val", X_val),
    ("X_test", X_test),
    ("y_train", y_train),
    ("y_val", y_val),
    ("y_test", y_test),
]:
    print(f"{name:8s}: shape={getattr(obj, 'shape', None)}")

print("\nX_train type:", type(X_train))

print("\n" + "=" * 80)
print("RECONSTRUCTED 10-FEATURE IMPUTER")
print("=" * 80)

print("Type:", type(imputer10))

if hasattr(imputer10, "statistics_"):
    print("statistics shape:", imputer10.statistics_.shape)
    print("statistics:", imputer10.statistics_)

if hasattr(imputer10, "feature_names_in_"):
    print("feature_names_in_:")
    print(list(imputer10.feature_names_in_))
```

```python
# ============================================================
# RECONSTRUCT LOCKED 10-FEATURE PROTEIN-DISJOINT MODEL
# ============================================================

import numpy as np
import pandas as pd
from sklearn.ensemble import ExtraTreesRegressor
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
from scipy.stats import pearsonr

FEATURES_10 = [
    "mol_weight",
    "logp",
    "tpsa",
    "hbd",
    "hba",
    "rotatable_bonds",
    "rings",
    "heavy_atoms",
    "temperature",
    "ph"
]

# ------------------------------------------------------------
# 1. Extract EXACT 10 features
# ------------------------------------------------------------

X_train_10 = X_train[FEATURES_10].copy()
X_val_10   = X_val[FEATURES_10].copy()
X_test_10  = X_test[FEATURES_10].copy()

print("X_train_10:", X_train_10.shape)
print("X_val_10:  ", X_val_10.shape)
print("X_test_10: ", X_test_10.shape)

# ------------------------------------------------------------
# 2. Apply the reconstructed locked imputer
# ------------------------------------------------------------

X_train_imp = imputer10.transform(X_train_10)
X_val_imp   = imputer10.transform(X_val_10)
X_test_imp  = imputer10.transform(X_test_10)

print("\nAfter imputation:")
print("Train:", X_train_imp.shape)
print("Val:  ", X_val_imp.shape)
print("Test: ", X_test_imp.shape)

# ------------------------------------------------------------
# 3. EXACT LOCKED PARAMETERS
# ------------------------------------------------------------

locked_model = ExtraTreesRegressor(
    n_estimators=200,
    min_samples_leaf=4,
    min_samples_split=2,
    max_features=1.0,
    max_depth=None,
    random_state=42,
    n_jobs=-1
)

# ------------------------------------------------------------
# 4. FIT ONLY ON THE ORIGINAL TRAINING SET
# ------------------------------------------------------------

locked_model.fit(X_train_imp, y_train)

print("\n" + "=" * 80)
print("LOCKED 10-FEATURE MODEL RECONSTRUCTED")
print("=" * 80)

print(type(locked_model))
print("n_features_in_:", locked_model.n_features_in_)
print("n_estimators:", locked_model.n_estimators)
print("min_samples_leaf:", locked_model.min_samples_leaf)
print("min_samples_split:", locked_model.min_samples_split)
print("max_features:", locked_model.max_features)
print("max_depth:", locked_model.max_depth)
print("random_state:", locked_model.random_state)
```

```python
# ============================================================
# SAVE FINAL LOCKED 10-FEATURE CATPRED MODEL
# ============================================================

import os
import pickle
import numpy as np
import pandas as pd
from sklearn.metrics import (
    r2_score,
    mean_squared_error,
    mean_absolute_error
)
from scipy.stats import pearsonr

# ------------------------------------------------------------
# DRIVE LOCATION
# ------------------------------------------------------------

FINAL_DIR = (
    "/content/drive/MyDrive/"
    "CatPred_kcat_Project/"
    "FINAL_LOCKED_MODEL"
)

os.makedirs(FINAL_DIR, exist_ok=True)

# ------------------------------------------------------------
# 1. VERIFY THE MODEL BEFORE SAVING
# ------------------------------------------------------------

print("=" * 90)
print("FINAL LOCKED MODEL")
print("=" * 90)

print("Model:", type(locked_model))
print("Features:", locked_model.n_features_in_)
print("Trees:", locked_model.n_estimators)
print("Min samples leaf:", locked_model.min_samples_leaf)
print("Min samples split:", locked_model.min_samples_split)
print("Max features:", locked_model.max_features)
print("Max depth:", locked_model.max_depth)
print("Random state:", locked_model.random_state)

print("\nFeatures:")
for i, f in enumerate(FEATURES_10, 1):
    print(f"{i:2d}. {f}")

# ------------------------------------------------------------
# 2. FINAL HELD-OUT TEST PREDICTIONS
# ------------------------------------------------------------

y_pred_test = locked_model.predict(X_test_imp)

r2 = r2_score(y_test, y_pred_test)
rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
mae = mean_absolute_error(y_test, y_pred_test)
pearson_r, pearson_p = pearsonr(y_test, y_pred_test)

# ------------------------------------------------------------
# 3. RESULTS TABLE
# ------------------------------------------------------------

results = pd.DataFrame([{
    "Model": "FINAL_LOCKED_10FEATURE_PROTEIN_DISJOINT_EXTRATREES",
    "N_Test": len(y_test),
    "R2": r2,
    "RMSE": rmse,
    "MAE": mae,
    "Pearson_r": pearson_r,
    "Pearson_p": pearson_p,
    "n_estimators": locked_model.n_estimators,
    "min_samples_leaf": locked_model.min_samples_leaf,
    "min_samples_split": locked_model.min_samples_split,
    "max_features": locked_model.max_features,
    "max_depth": locked_model.max_depth,
    "random_state": locked_model.random_state
}])

print("\n" + "=" * 90)
print("FINAL HELD-OUT TEST PERFORMANCE")
print("=" * 90)
print(results.to_string(index=False))

# ------------------------------------------------------------
# 4. SAVE MODEL
# ------------------------------------------------------------

MODEL_PATH = os.path.join(
    FINAL_DIR,
    "FINAL_LOCKED_10FEATURE_PROTEIN_DISJOINT_EXTRATREES.pkl"
)

with open(MODEL_PATH, "wb") as f:
    pickle.dump(locked_model, f, protocol=4)

# ------------------------------------------------------------
# 5. SAVE IMPUTER
# ------------------------------------------------------------

IMPUTER_PATH = os.path.join(
    FINAL_DIR,
    "FINAL_LOCKED_10FEATURE_IMPUTER.pkl"
)

with open(IMPUTER_PATH, "wb") as f:
    pickle.dump(imputer10, f, protocol=4)

# ------------------------------------------------------------
# 6. SAVE FEATURE NAMES
# ------------------------------------------------------------

FEATURE_PATH = os.path.join(
    FINAL_DIR,
    "FINAL_LOCKED_10FEATURE_NAMES.pkl"
)

with open(FEATURE_PATH, "wb") as f:
    pickle.dump(FEATURES_10, f, protocol=4)

# ------------------------------------------------------------
# 7. SAVE TEST RESULTS
# ------------------------------------------------------------

RESULTS_PATH = os.path.join(
    FINAL_DIR,
    "FINAL_LOCKED_10FEATURE_HELDOUT_TEST_RESULTS.csv"
)

results.to_csv(RESULTS_PATH, index=False)

# ------------------------------------------------------------
# 8. SAVE TEST PREDICTIONS
# ------------------------------------------------------------

PRED_PATH = os.path.join(
    FINAL_DIR,
    "FINAL_LOCKED_10FEATURE_HELDOUT_TEST_PREDICTIONS.csv"
)

pd.DataFrame({
    "y_true": np.asarray(y_test),
    "y_pred": y_pred_test
}).to_csv(PRED_PATH, index=False)

# ------------------------------------------------------------
# 9. SAVE HUMAN-READABLE MANIFEST
# ------------------------------------------------------------

MANIFEST_PATH = os.path.join(
    FINAL_DIR,
    "FINAL_LOCKED_10FEATURE_MODEL_MANIFEST.txt"
)

with open(MANIFEST_PATH, "w") as f:

    f.write("FINAL LOCKED CATPRED MODEL\n")
    f.write("=" * 80 + "\n\n")

    f.write("Model: ExtraTreesRegressor\n")
    f.write("Dataset split: Protein-disjoint\n")
    f.write("Status: FINAL LOCKED MODEL\n\n")

    f.write("FEATURES\n")
    f.write("-" * 40 + "\n")

    for i, feature in enumerate(FEATURES_10, 1):
        f.write(f"{i}. {feature}\n")

    f.write("\nMODEL PARAMETERS\n")
    f.write("-" * 40 + "\n")
    f.write(f"n_estimators = {locked_model.n_estimators}\n")
    f.write(f"min_samples_leaf = {locked_model.min_samples_leaf}\n")
    f.write(f"min_samples_split = {locked_model.min_samples_split}\n")
    f.write(f"max_features = {locked_model.max_features}\n")
    f.write(f"max_depth = {locked_model.max_depth}\n")
    f.write(f"random_state = {locked_model.random_state}\n")

    f.write("\nHELD-OUT TEST PERFORMANCE\n")
    f.write("-" * 40 + "\n")
    f.write(f"N = {len(y_test)}\n")
    f.write(f"R2 = {r2:.6f}\n")
    f.write(f"RMSE = {rmse:.6f}\n")
    f.write(f"MAE = {mae:.6f}\n")
    f.write(f"Pearson_r = {pearson_r:.6f}\n")
    f.write(f"Pearson_p = {pearson_p:.6e}\n")

# ------------------------------------------------------------
# 10. VERIFY FILES EXIST
# ------------------------------------------------------------

print("\n" + "=" * 90)
print("SAVED SUCCESSFULLY")
print("=" * 90)

for file in sorted(os.listdir(FINAL_DIR)):
    path = os.path.join(FINAL_DIR, file)
    print(
        f"✓ {file:<60} "
        f"{os.path.getsize(path) / (1024**2):.2f} MB"
    )

print("\nFINAL DIRECTORY:")
print(FINAL_DIR)
```

```python
import os
import glob
import pandas as pd

BASE = "/content/drive/MyDrive/CatPred_kcat_Project"

print("=" * 80)
print("SEARCHING FOR THE ~0.51 R² MODEL")
print("=" * 80)

# Search all CSV result files for R2 values around 0.51
csv_files = glob.glob(BASE + "/**/*.csv", recursive=True)

matches = []

for f in csv_files:
    try:
        df = pd.read_csv(f)

        # Check every numeric column for values near 0.51
        for col in df.columns:
            vals = pd.to_numeric(df[col], errors="coerce")

            mask = vals.notna() & (vals >= 0.45) & (vals <= 0.57)

            if mask.any():
                for idx in df.index[mask]:
                    matches.append({
                        "file": f,
                        "column": col,
                        "row": idx,
                        "value": vals.loc[idx]
                    })

    except Exception:
        pass

print("\nPotential ~0.51 results found:")
print("-" * 80)

if matches:
    result_df = pd.DataFrame(matches)
    result_df = result_df.sort_values(
        "value",
        key=lambda x: abs(x - 0.51)
    )

    print(result_df.to_string(index=False))

    out = os.path.join(
        BASE,
        "MODEL_051_SEARCH_RESULTS.csv"
    )

    result_df.to_csv(out, index=False)

    print("\nSaved:")
    print(out)

else:
    print("NO RESULTS WITH R² IN THE 0.45–0.57 RANGE WERE FOUND.")
```

```python
import os
import pandas as pd
import numpy as np

BASE = "/content/drive/MyDrive/CatPred_kcat_Project"

OUT = os.path.join(
    BASE,
    "FINAL_LOCKED_MODEL",
    "BIOLOGICAL_INTERPRETATION"
)

os.makedirs(OUT, exist_ok=True)

print("=" * 80)
print("BIOLOGICAL INTERPRETATION OF FINAL LOCKED 10-FEATURE MODEL")
print("=" * 80)

# -------------------------------------------------------------------------
# FILES ALREADY GENERATED DURING MODEL EVALUATION
# -------------------------------------------------------------------------

FILES = {
    "feature_importance":
        os.path.join(
            BASE,
            "protein_level_evaluation",
            "protein_disjoint_final_feature_importance.csv"
        ),

    "permutation_summary":
        os.path.join(
            BASE,
            "protein_level_evaluation",
            "protein_disjoint_5fold_permutation_importance_summary.csv"
        ),

    "range_bias":
        os.path.join(
            BASE,
            "protein_level_evaluation",
            "protein_disjoint_5fold_kcat_range_bias.csv"
        ),

    "range_compression":
        os.path.join(
            BASE,
            "protein_level_evaluation",
            "protein_disjoint_5fold_range_compression.csv"
        ),

    "group_importance":
        os.path.join(
            BASE,
            "protein_level_evaluation",
            "protein_disjoint_final_feature_importance.csv"
        )
}

# -------------------------------------------------------------------------
# CHECK FILES
# -------------------------------------------------------------------------

for name, path in FILES.items():
    print(f"\n{name}:")
    print(path)

    if os.path.exists(path):
        print("FOUND")
    else:
        print("NOT FOUND")

# -------------------------------------------------------------------------
# LOAD FEATURE IMPORTANCE
# -------------------------------------------------------------------------

fi_path = FILES["feature_importance"]

if not os.path.exists(fi_path):
    raise FileNotFoundError(
        "Final feature-importance file was not found:\n" + fi_path
    )

fi = pd.read_csv(fi_path)

print("\n" + "=" * 80)
print("FEATURE IMPORTANCE FILE")
print("=" * 80)

print("Shape:", fi.shape)
print("Columns:", list(fi.columns))
print("\nFirst rows:")
print(fi.head(20).to_string(index=False))

# -------------------------------------------------------------------------
# IDENTIFY FEATURE AND IMPORTANCE COLUMNS AUTOMATICALLY
# -------------------------------------------------------------------------

feature_candidates = [
    "feature",
    "Feature",
    "feature_name",
    "Feature_Name"
]

importance_candidates = [
    "importance",
    "Importance",
    "feature_importance",
    "mean_importance",
    "Mean_Importance"
]

feature_col = None
importance_col = None

for c in feature_candidates:
    if c in fi.columns:
        feature_col = c
        break

for c in importance_candidates:
    if c in fi.columns:
        importance_col = c
        break

if feature_col is None:
    raise ValueError(
        "Could not identify the feature-name column.\n"
        f"Available columns: {list(fi.columns)}"
    )

if importance_col is None:
    numeric_cols = fi.select_dtypes(include=np.number).columns.tolist()

    if len(numeric_cols) == 0:
        raise ValueError(
            "Could not identify an importance column."
        )

    importance_col = numeric_cols[0]

print("\nUsing feature column   :", feature_col)
print("Using importance column :", importance_col)

# -------------------------------------------------------------------------
# CREATE CLEAN FEATURE TABLE
# -------------------------------------------------------------------------

feature_table = fi[
    [feature_col, importance_col]
].copy()

feature_table.columns = [
    "Feature",
    "Model_Importance"
]

feature_table["Model_Importance"] = pd.to_numeric(
    feature_table["Model_Importance"],
    errors="coerce"
)

feature_table = feature_table.dropna(
    subset=["Feature", "Model_Importance"]
)

feature_table = feature_table.sort_values(
    "Model_Importance",
    ascending=False
).reset_index(drop=True)

feature_table["Importance_Rank"] = (
    np.arange(len(feature_table)) + 1
)

# -------------------------------------------------------------------------
# BIOLOGICAL INTERPRETATION MAP
# -------------------------------------------------------------------------

interpretation = {

    "mol_weight":
        "Molecular size and mass; influences substrate volume, steric accessibility, "
        "and the physical context of enzyme–substrate recognition.",

    "logp":
        "Hydrophobicity/lipophilicity; reflects the balance between hydrophobic and "
        "hydrophilic character and can influence substrate partitioning and binding.",

    "tpsa":
        "Topological polar surface area; represents the polar character of the "
        "substrate and contributes to hydrogen-bonding and polar interactions.",

    "hbd":
        "Hydrogen-bond donors; indicates the capacity of the substrate to participate "
        "in donor-mediated hydrogen-bond interactions.",

    "hba":
        "Hydrogen-bond acceptors; reflects potential acceptor sites involved in "
        "polar substrate–enzyme interactions.",

    "rotatable_bonds":
        "Molecular flexibility; reflects the number of freely rotatable bonds and "
        "therefore possible conformational adaptability during binding.",

    "rings":
        "Ring architecture; captures molecular rigidity and structural organization "
        "that can affect substrate recognition and positioning.",

    "heavy_atoms":
        "Overall molecular complexity and size; provides a complementary measure "
        "of the non-hydrogen atomic framework of the substrate.",

    "temperature":
        "Reaction temperature; captures an experimental condition that can influence "
        "enzyme catalytic dynamics and measured kcat.",

    "ph":
        "Reaction pH; captures an experimental condition affecting ionization states, "
        "active-site chemistry, and catalytic activity."
}

feature_table["Biological_Interpretation"] = (
    feature_table["Feature"]
    .astype(str)
    .str.lower()
    .map(interpretation)
)

feature_table["Biological_Interpretation"] = (
    feature_table["Biological_Interpretation"]
    .fillna(
        "Model-derived molecular or experimental feature contributing to catalytic-rate prediction."
    )
)

# -------------------------------------------------------------------------
# NORMALIZED IMPORTANCE
# -------------------------------------------------------------------------

total_importance = feature_table["Model_Importance"].sum()

if total_importance != 0:
    feature_table["Relative_Importance_%"] = (
        feature_table["Model_Importance"] /
        total_importance * 100
    )
else:
    feature_table["Relative_Importance_%"] = np.nan

# -------------------------------------------------------------------------
# PRINT FINAL TABLE
# -------------------------------------------------------------------------

print("\n" + "=" * 80)
print("FINAL BIOLOGICAL INTERPRETATION TABLE")
print("=" * 80)

print(
    feature_table[
        [
            "Importance_Rank",
            "Feature",
            "Model_Importance",
            "Relative_Importance_%",
            "Biological_Interpretation"
        ]
    ].to_string(index=False)
)

# -------------------------------------------------------------------------
# SAVE
# -------------------------------------------------------------------------

output_file = os.path.join(
    OUT,
    "FINAL_10FEATURE_BIOLOGICAL_INTERPRETATION.csv"
)

feature_table.to_csv(
    output_file,
    index=False
)

print("\n" + "=" * 80)
print("SAVED")
print("=" * 80)

print(output_file)

print("\nDone.")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ============================================================
# CATPRED — PROTEIN-DISJOINT 5-FOLD CV SUMMARY + RANGE BIAS
# Based directly on reported OOF/CV results
# ============================================================

# ------------------------------------------------------------
# 5-fold CV results
# ------------------------------------------------------------
folds = np.array([1, 2, 3, 4, 5])

r2 = np.array([
    0.1024,
    0.0357,
    0.0992,
    0.1540,
    0.0314
])

rmse = np.array([
    1.6367,
    1.6544,
    1.4739,
    1.4774,
    1.5433
])

mae = np.array([
    1.2676,
    1.2251,
    1.1139,
    1.1453,
    1.1951
])

pearson = np.array([
    0.3319,
    0.2402,
    0.3235,
    0.3953,
    0.2653
])

# Reported overall OOF values
overall_r2 = 0.0860
overall_rmse = 1.5590
overall_mae = 1.1894
overall_pearson = 0.3079

# ------------------------------------------------------------
# kcat-range bias results
# ------------------------------------------------------------
ranges = ["Low\nkcat", "Medium\nkcat", "High\nkcat"]

observed_mean = np.array([
    -0.871514,
     0.854965,
     2.546571
])

predicted_mean = np.array([
     0.630506,
     0.838344,
     1.062657
])

mean_bias = np.array([
     1.502020,
    -0.016621,
    -1.483914
])

mae_range = np.array([
    1.525034,
    0.524477,
    1.520645
])

rmse_range = np.array([
    1.833639,
    0.667242,
    1.856794
])

pearson_range = np.array([
    0.172207,
    0.070911,
    0.196756
])

# ============================================================
# FIGURE
# ============================================================

fig, axes = plt.subplots(
    1, 2,
    figsize=(11, 4.8),
    gridspec_kw={"width_ratios": [1.15, 1]}
)

# ============================================================
# PANEL A — Five-fold protein-disjoint CV performance
# ============================================================

ax = axes[0]

# Plot R² and Pearson r across folds
ax.plot(
    folds, r2,
    marker="o",
    linewidth=2,
    markersize=6,
    label="R²"
)

ax.plot(
    folds, pearson,
    marker="s",
    linewidth=2,
    markersize=6,
    label="Pearson r"
)

# Overall OOF values
ax.axhline(
    overall_r2,
    linestyle="--",
    linewidth=1.5,
    label=f"Overall R² = {overall_r2:.4f}"
)

ax.axhline(
    overall_pearson,
    linestyle=":",
    linewidth=1.5,
    label=f"Overall Pearson r = {overall_pearson:.4f}"
)

ax.set_xlabel("Protein-disjoint CV fold")
ax.set_ylabel("Correlation / R²")
ax.set_xticks(folds)
ax.set_ylim(0, 0.45)

ax.set_title(
    "A  Protein-disjoint five-fold cross-validation"
)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

ax.legend(
    frameon=False,
    fontsize=8,
    loc="upper right"
)

# Add summary text
summary_text = (
    f"OOF R² = {overall_r2:.4f}\n"
    f"RMSE = {overall_rmse:.4f}\n"
    f"MAE = {overall_mae:.4f}\n"
    f"Pearson r = {overall_pearson:.4f}"
)

ax.text(
    0.03, 0.04,
    summary_text,
    transform=ax.transAxes,
    fontsize=9,
    verticalalignment="bottom",
    bbox=dict(
        boxstyle="round,pad=0.35",
        facecolor="white",
        edgecolor="0.7",
        alpha=0.9
    )
)

# ============================================================
# PANEL B — Prediction bias across kcat ranges
# ============================================================

ax = axes[1]

x = np.arange(len(ranges))

bars = ax.bar(
    x,
    mean_bias,
    width=0.62,
    alpha=0.85
)

# Zero reference
ax.axhline(
    0,
    linestyle="--",
    linewidth=1.2
)

ax.set_xticks(x)
ax.set_xticklabels(ranges)

ax.set_ylabel("Mean prediction bias\n(Predicted − Observed)")

ax.set_title(
    "B  Prediction bias across kcat ranges"
)

ax.set_ylim(-1.9, 1.9)

ax.grid(
    axis="y",
    linestyle="--",
    alpha=0.25
)

# Add bias values
for bar, value in zip(bars, mean_bias):

    if value >= 0:
        y = value + 0.08
        va = "bottom"
    else:
        y = value - 0.08
        va = "top"

    ax.text(
        bar.get_x() + bar.get_width()/2,
        y,
        f"{value:+.4f}",
        ha="center",
        va=va,
        fontsize=9
    )

# Add MAE below the category labels through text inside plot
for i, value in enumerate(mae_range):
    ax.text(
        i,
        -1.78,
        f"MAE = {value:.3f}",
        ha="center",
        fontsize=7.5
    )

# ============================================================
# Overall formatting
# ============================================================

plt.tight_layout()

# Save high-resolution figure
output_path = "/content/CATPred_protein_disjoint_CV_bias.png"

plt.savefig(
    output_path,
    dpi=600,
    bbox_inches="tight"
)

plt.show()

print("Figure saved to:")
print(output_path)
```
